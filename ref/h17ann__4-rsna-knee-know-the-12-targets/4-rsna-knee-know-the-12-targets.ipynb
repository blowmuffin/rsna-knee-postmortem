## 4. ACL — Anterior Cruciate Ligament

The **ACL** is one of the main ligaments inside the knee.

It connects the femur to the tibia and contributes to knee stability, especially during movements that involve changes in direction.

An ACL injury occurs when the ligament is overstretched or torn. ACL injuries may happen during sudden stopping, rapid changes of direction, incorrect landing, or direct contact.

**Source:** [American Academy of Orthopaedic Surgeons (AAOS) — Anterior Cruciate Ligament Injury](https://orthoinfo.aaos.org/globalassets/pdfs/acl-cpg_pls.pdf)

## 5. MCL — Medial Collateral Ligament

The **MCL** is a ligament on the inner side of the knee.

Together with the other collateral ligament, it helps control side-to-side motion and supports knee stability.

An MCL injury means that this ligament has been stretched or torn.

**Source:** [American Academy of Orthopaedic Surgeons (AAOS) — About Your Knee](https://orthoinfo.aaos.org/globalassets/pdfs/about-your-knee.pdf)

## 6. Medial Meniscus

The **medial meniscus** is one of the two menisci between the femur and tibia.

The menisci are wedge-shaped pieces of fibrocartilage that help:

- absorb shock
- transmit and distribute load
- contribute to knee stability

The competition target refers to a **medial meniscus tear**. Meniscal tears can occur after trauma or as part of degenerative changes.

**Source:** [American Academy of Orthopaedic Surgeons (AAOS) — Meniscus Tears](https://orthoinfo.aaos.org/en/diseases--conditions/meniscus-tears/)

## 7. Lateral Meniscus

The **lateral meniscus** is the meniscus on the outer side of the knee.

Like the medial meniscus, it helps with shock absorption, load transmission, and stability.

The competition target refers to a **lateral meniscus tear**.

**Source:** [American Academy of Orthopaedic Surgeons (AAOS) — Meniscus Tears](https://orthoinfo.aaos.org/en/diseases--conditions/meniscus-tears/)

## 8. Medial Osteoarthritis

**Osteoarthritis (OA)** is a degenerative joint disease in which the articular cartilage becomes damaged and wears away over time.

**Medial OA** affects mainly the medial tibiofemoral compartment, the inner side of the knee joint.

OA can be associated with cartilage loss, pain, stiffness, and bony changes such as osteophytes.

**Source:** [American Academy of Orthopaedic Surgeons (AAOS) — Knee Osteoarthritis](https://orthoinfo.aaos.org/globalassets/pdfs/knee-osteoarthritis.pdf)

## 9. Lateral Osteoarthritis

**Lateral OA** is osteoarthritis affecting the lateral tibiofemoral compartment, the outer side of the knee.

The underlying process is the same osteoarthritic degeneration described above, but the affected compartment is different.

**Source:** [American Academy of Orthopaedic Surgeons (AAOS) — Knee Osteoarthritis](https://orthoinfo.aaos.org/globalassets/pdfs/knee-osteoarthritis.pdf)

## 10. Patellofemoral Osteoarthritis

**PF OA** means **patellofemoral osteoarthritis**.

It affects the joint between the patella (kneecap) and the femur. Damage and wear of the articular cartilage in this compartment can lead to pain, often felt at the front of the knee.

**Source:** [American Academy of Orthopaedic Surgeons (AAOS) — Patellofemoral Arthritis](https://orthoinfo.aaos.org/diseases--conditions/patellofemoral-arthritis)

## 11. Effusion

A **joint effusion** means that too much fluid has accumulated in or around a joint.

In the knee, this is often described as **knee effusion** or “fluid on the knee.”

Effusion can occur with several different joint problems, including trauma, osteoarthritis, inflammation, or infection.

**Source:** [MedlinePlus — Synovial Fluid Analysis](https://medlineplus.gov/lab-tests/synovial-fluid-analysis/)

## 12. Synovitis

The **synovium** is the tissue lining the inside of the joint.

**Synovitis** means inflammation of this synovial lining. It can be associated with pain and swelling.

Synovitis and effusion are related concepts, but they are not the same thing: synovitis describes inflammation of the lining, while effusion describes excess joint fluid.

**Sources:**  
- [MedlinePlus — Knee Arthroscopy: Synovium and Synovitis](https://medlineplus.gov/ency/patientinstructions/000199.htm)  
- [MedlinePlus — Synovial Fluid Analysis](https://medlineplus.gov/lab-tests/synovial-fluid-analysis/)

## 13. Baker's Cyst

A **Baker's cyst**, also called a **popliteal cyst**, is a buildup of synovial fluid that forms a swelling behind the knee.

It often occurs when another knee problem causes increased joint fluid, such as arthritis or a meniscal tear.

**Source:** [MedlinePlus — Baker Cyst](https://medlineplus.gov/ency/article/001222.htm)

## 14. Bone Contusion

A **bone contusion**, often called a **bone bruise**, is an occult bone injury that can be seen on MRI after knee trauma.

It can involve abnormal signal within the subchondral bone and may occur together with ligament or cartilage injuries.

Unlike a complete fracture, a bone bruise does not necessarily involve a visible break in the bone cortex.

**Source:** [PubMed — “Bone bruises” of the knee: a review](https://pubmed.ncbi.nlm.nih.gov/9807716/)

## 15. Fracture

A **fracture** is a break in a bone.

Fractures can be partial or complete and can vary widely in severity. Around the knee, different bones may be involved, including the femur, tibia, fibula, or patella.

**Source:** [MedlinePlus — Fractures](https://medlineplus.gov/fractures.html)

## 16. Group the targets

```python
group_counts = target_info["Group"].value_counts()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=group_counts.values,
    y=group_counts.index
)

plt.xlabel("Number of targets")
plt.ylabel("")
plt.title("Targets by anatomical / clinical group")
plt.show()
```

## Takeaways

The 12 targets cover several different parts of knee pathology:

- **Ligaments:** ACL and MCL injuries
- **Menisci:** medial and lateral meniscus tears
- **Osteoarthritis:** medial, lateral, and patellofemoral compartments
- **Fluid / inflammation:** effusion and synovitis
- **Cyst:** Baker's cyst
- **Bone:** contusion and fracture

So this is not one single type of abnormality.  
The competition mixes traumatic, degenerative, inflammatory, and bone-related findings.

```python
# Everything shown here is plain HTML built from the files in the attached dataset.
import base64, glob
from IPython.display import HTML, display

def find(name):
    # look only one or three levels deep: the competition's DICOM tree is huge, never walk it
    hits = glob.glob(f"/kaggle/input/*/{name}") + glob.glob(f"/kaggle/input/datasets/*/*/{name}")
    assert hits, f"{name} not found under /kaggle/input"
    return hits[0]

def b64(name):
    return base64.b64encode(open(find(name), "rb").read()).decode()

FONTS = "@import url('https://fonts.googleapis.com/css2?family=Alfa+Slab+One&family=Gochi+Hand&family=Patrick+Hand&family=Special+Elite&display=swap');"
INK, CREAM, KRAFT = "#23243a", "#f6efde", "#c9a476"
TORN = "polygon(0% 2%, 6% 0%, 13% 1.5%, 21% 0%, 30% 1.8%, 38% 0.3%, 47% 1.6%, 55% 0%, 64% 1.4%, 72% 0.2%, 81% 1.7%, 89% 0.1%, 100% 1.5%, 99.2% 50%, 100% 98%, 92% 100%, 84% 98.4%, 75% 100%, 66% 98.6%, 57% 100%, 48% 98.2%, 39% 99.9%, 30% 98.5%, 21% 100%, 12% 98.4%, 4% 100%, 0% 98%, 0.8% 50%)"

def chip(c, bg, fg, r, size="clamp(20px,3.8vw,42px)"):
    return (f'<span style="display:inline-block;background:{bg};color:{fg};font:{size}/1 \'Alfa Slab One\',Georgia,serif;'
            f'padding:6px 9px 4px;margin:3px 2px;transform:rotate({r}deg);box-shadow:2px 4px 5px rgba(40,25,10,.35)">{c}</span>')

PAL = [("#3454aa", "#fff"), ("#20beff", "#fff"), ("#f6efde", INK), ("#de543e", "#fff8ec"), ("#e8b43c", INK), ("#23243a", "#f6efde"),
       ("#acd6be", INK), ("#fff", "#de543e"), ("#f0aaaa", INK)]
ROT = [-6, 4, -3, 7, -5, 3, -4, 6, -2, 5, -7, 3, 5, -4, 6, -3, 4, -6, 2, -5]

def ransom(text, size="clamp(20px,3.8vw,42px)", seed=0):
    words, k = [], seed
    for w in text.split(" "):
        cs = []
        for c in w:
            bg, fg = PAL[k % len(PAL)]
            cs.append(chip(c, bg, fg, ROT[k % len(ROT)], size)); k += 1
        words.append('<span style="white-space:nowrap;display:inline-block">' + "".join(cs) + "</span>")
    return '<span style="display:inline-block;width:16px"></span>'.join(words)
```

<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">What the film is showing</h2>

| Step | In the film | In practice |
|---|---|---|
| **Input** | three stacks of prints | DICOM series per study (sagittal, coronal, axial; fluid-sensitive or not), typically a few dozen slices each |
| **Preprocessing** | sort, even out brightness, scissors | order slices along the scan direction, normalise intensities per series (e.g. percentile clipping), crop to the knee and resize |
| **2.5D** | red, green, blue slices stacked | feed three adjacent slices as the three channels of an ImageNet-pretrained 2D backbone: cheap depth context without a 3D network |
| **Backbone** | paper feature maps | a CNN (ConvNeXt, EfficientNet, ResNet, CoAtNet...) turns every 2.5D window into a feature vector |
| **Aggregation** | spotlight + weight bars | attention / multiple-instance pooling over the slices of a study, then 12 sigmoid outputs (one per finding) |
| **Labels** | the reading brain | only a few studies have expert labels; the rest are derived from the free-text reports (any language), e.g. with an LLM |
| **Training** | folds, flip, twist, rising curve | grouped k-fold cross-validation, augmentation (flips, rotations, intensity), BCE loss, validation macro AUC |
| **Inference** | robots voting, hourglass, no wifi | an ensemble of folds / models averaged (often rank-averaged) inside a 9-hour, internet-free Kaggle notebook |

<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">Simplifications in the film</h2>

A 60-second cartoon has to cut corners. Three things it shows more simply than reality:

- **"Crop to 256 × 256"** is the size used by my own models, not a competition rule. Strong solutions use other sizes too (224, 384, 448...), and many crop to the knee first.
- **Five robots voting** stands in for an ensemble. The top public notebooks blend many more models (often dozens of folds and architectures), usually by rank-averaging.
- **"A language model turns reports into labels"** is the common approach in this competition, but not the only one. Others use keyword/regex rules, soft or pseudo-labels from a model's own predictions, or train directly on the small expert-labelled subset.

<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">How the film was made</h2>

rendered frame by frame in headless Chromium. Everything ran locally with open-weight models:

| Layer | How |
|---|---|

| **Music** | an original cue from ACE-Step 1.5, joined on the beat so its last hit lands on the end card. |

| **MRI slices** | synthesised from Henry Gray's *Anatomy of the Human Body* (1918, public domain) sagittal, frontal and axial plates. **No competition images were used.** |

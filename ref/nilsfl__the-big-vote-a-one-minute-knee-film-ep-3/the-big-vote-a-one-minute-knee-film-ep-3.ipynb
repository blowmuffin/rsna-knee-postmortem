<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">What the film is showing</h2>

The public 0.943 notebooks are forks of one stack. This is how it combines its members (read from the notebook code):

| Stage | Members | Input | How it is combined |
|---|---|---|---|
| **DINO** | 20 DINOv2-small checkpoints (4 per fold) | 6 slots (sag/cor/ax fluid fat-sat, sag fluid non-FS, cor and sag T1), 12 slices, 336 px, 130 mm crop | per-window pooling (max for local findings, top-2 for ACL/MCL, mean otherwise); members percentile-ranked and averaged |
| **A5** | DINOv3 ViT-small, 5 folds | 6 plane &times; fat-sat slots, 16 slices, 336 px | fold ranks averaged, mixed into the DINO ranks at 52% |
| **Rad** | frozen RadImageNet ResNet-50 + reference / E13 heads | 8 slices, 224 px, three input layouts | 45% previous ranks + 55% Rad ranks, then a linear calibration on protocol features |
| **Raptor** | 4 CoAtNet-384 arms (maxspan, native384dense, reverse, native384) | about 96 slices over 5 slots, up to 94 three-slice windows | probability mean with weights .6/.1/.1/.2, then ranked |
| **CoAt readers** | resgated (3), Global96 (3), D4 (1 SWA), Repair-v1 (3) | up to 94 windows | probability mean, ranked; hybrid = 60% Raptor + 40% readers |
| **Final** | &mdash; | &mdash; | per finding: (1 &minus; w) &middot; DINO/A5/Rad + w &middot; CoAt hybrid, w = 0.6 by default, 0.75 for ACL, Lateral OA and Fracture, 0.8 Medial Meniscus, **1.0 Lateral Meniscus**; then a final rank |

**Why ranks?** The metric is macro AUC: for each finding, the probability that a random positive study is scored above a random negative one. Only the ordering matters, so ranks are a safe common currency between models with different calibrations.

<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">Simplifications in the film</h2>

- **"Six families"** counts DINO, A5, Rad, Raptor, the CoAt readers and the per-label routing/calibration as separate families. You could also count five model families plus the blending logic.
- **"More than a dozen CoAtNets"** means checkpoints: 4 Raptor arms plus 10 reader checkpoints. They are not 14 independent architectures.
- **"Nearly a thousand teams share that score"** is the public-leaderboard tie at 0.943 on 2026-10-05 (1,009 teams). Most are forks of this stack, but not all are byte-identical copies.
- **The 6.5 hours** is my own measured scoring time on the hidden test (T4 &times; 2). It can vary from run to run.

<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">How the film was made</h2>

Same world and toolkit as episodes 1 and 2: the animation is **pure JavaScript** (DOM + SVG on one GSAP timeline, stepped at 12 poses per second for a stop-motion feel), rendered frame by frame in headless Chromium.

| Layer | How |
|---|---|
| **Voices** | Qwen3-TTS, voice-cloned from the episode-2 narrator, knee and robot. 4 takes per line, ranked by Whisper word accuracy and noise floor. |
| **Music** | the series theme (ACE-Step 1.5), re-edited on the beat so the last hit lands on the knee's wink. |
| **Illustrations** | colored-pencil drawings from Z-Image-Turbo (photocopier, mixing bowl, ladder, barbell), cut out with scissor-cut paper rims. |
| **MRI slices** | synthesised from Henry Gray's *Anatomy of the Human Body* (1918, public domain). **No competition images were used.** |
| **Sound effects** | Kenney CC0 samples plus synthesis. |

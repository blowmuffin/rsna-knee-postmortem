<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">The challenge in a nutshell</h2>

| | The short version |
|---|---|
| **Input** | a knee MRI study: several series (sagittal, coronal, axial; fluid-sensitive or not) |
| **Output** | twelve probabilities per study, one for each finding above |
| **Training data** | ~4,400 studies, each with its original radiology report, written in many languages; only a small subset carries per-finding labels, so turning reports into labels is part of the game |
| **Test data** | ~1,300 hidden studies, labelled by expert radiologists *from the images* |
| **Metric** | macro-averaged ROC AUC over the twelve findings |
| **Format** | code competition: notebook-only, 9 hours, no internet |

$$\text{Score} = \frac{1}{12}\sum_{i=1}^{12} \text{AUC}_i$$

In the film's words: *"Guesses are scored by the area under the curve. More area... happier knees."*
Because AUC only cares about **ranking**, a model never has to be calibrated: it just has to put the effusion knees above the dry ones.

<h2 style="font-family:'Gochi Hand',cursive;font-size:34px;color:#3454aa;border-bottom:3px dashed #e8b43c;padding-bottom:4px">How the film was made</h2>

The animation is **pure JavaScript**: DOM + SVG on a single GSAP timeline, stepped at 12 poses per second for a stop-motion feel,
with line boil, paper jitter, googly eyes and a slight handheld drift. Every beat is keyed to a spoken word, so re-timing the voice re-times the film.
The video was rendered frame by frame from that page in headless Chromium. Everything ran locally on a laptop with open-weight models:

| Layer | How |
|---|---|
| **Voices** | Qwen3-TTS *VoiceDesign* invented the narrator, the knee and the robot; the narrator was then re-voiced with Qwen3-TTS *Base* voice cloning so he sounds the same on every line. Takes were ranked automatically (Whisper transcript accuracy, audiobox-aesthetics quality, pitch, speaker similarity). |
| **Music** | an original cue from ACE-Step 1.5, edited to picture: muffled under *"but sometimes..."*, a tape-stop on the crash, a stop-time stab on the RSNA title and the final button on the knee's wink. |
| **Sound effects** | Kenney CC0 samples plus synthesis: crayon scribbles, MRI knocking, robot bleeps, paper slaps. |
| **Illustrations** | colored-pencil drawings from Z-Image-Turbo, cut out with scissor-cut paper rims; the leg is a two-piece puppet hinged at the knee. |
| **Paper & tape** | procedurally generated torn paper (with the white fibrous core at the tear), graph/ruled/newsprint patterns and masking tape. |
| **Anatomy** | plates from Henry Gray's *Anatomy of the Human Body* (1918, public domain). The "MRI slices" are synthesised from Gray's sagittal knee section. **No competition images were used.** |

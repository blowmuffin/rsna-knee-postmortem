```python
# Embed the interactive 3D WebGL Knee Simulator
html_content = """<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>PocketGull 3D Knee Anatomy & MRI Slicing Simulator v12 (Anti-Burnout Suite) | RSNA Knee 2026</title>

  <!-- PocketGull Superfamily Typography (Fastly Edge CDN, WOFF2, Zero-Overhead) -->
  <link rel="preconnect" href="https://font.pocketgull.app" crossorigin>
  <link rel="stylesheet" href="https://font.pocketgull.app/fonts.css" media="all">

  <!-- Resilient 3-Tier Multi-CDN Three.js (Privacy Badger & Ad-Blocker Immune) -->
  <script src="https://cdnjs.cloudflare.com/ajax/libs/three.js/r128/three.min.js"></script>
  <script>
    if (typeof THREE === 'undefined') {
      document.write('<script src="https://unpkg.com/three@0.128.0/build/three.min.js"><\\/script>');
    }
  </script>
  <script>
    if (typeof THREE === 'undefined') {
      document.write('<script src="https://cdn.jsdelivr.net/npm/three@0.128.0/build/three.min.js"><\\/script>');
    }
  </script>
  <!-- Tailwind runtime JIT (with standalone CSS direct-curve fallback) -->
  <script src="https://cdn.tailwindcss.com"></script>
  <style>
    /* Direct @font-face for PocketGull Soft (Humanized, approachable medical geometry) */
    @font-face {
      font-family: 'PocketGull Soft';
      src: url('https://font.pocketgull.app/fonts/woff2/PocketGull-Soft-Regular.woff2?v=3.100') format('woff2');
      font-weight: 400 500;
      font-style: normal;
      font-display: swap;
    }
    @font-face {
      font-family: 'PocketGull Soft';
      src: url('https://font.pocketgull.app/fonts/woff2/PocketGull-Soft-Bold.woff2?v=3.100') format('woff2');
      font-weight: 700 800;
      font-style: normal;
      font-display: swap;
    }
    @font-face {
      font-family: 'PocketGull Mono';
      src: url('https://font.pocketgull.app/fonts/woff2/PocketGullMono-Regular.woff2?v=3.3.0') format('woff2');
      font-weight: 400 700;
      font-style: normal;
      font-display: swap;
    }

    body {
      margin: 0;
      overflow: hidden;
      background-color: #09090b;
      font-family: 'PocketGull Soft', 'PocketGull', var(--font-sans, -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, Helvetica, Arial, sans-serif);
      -webkit-font-smoothing: antialiased;
      -moz-osx-font-smoothing: grayscale;
      text-rendering: optimizeLegibility;
    }
    .font-brand-pocketgull {
      font-family: 'PocketGull Soft', 'PocketGull Bold', 'PocketGull', sans-serif !important;
      font-weight: 700;
      letter-spacing: -0.015em;
    }
    .font-mono, [data-mono], .compass-tick {
      font-family: 'PocketGull Mono', var(--font-mono, 'JetBrains Mono', ui-monospace, monospace) !important;
      font-feature-settings: "tnum" 1, "zero" 1;
      font-variant-numeric: tabular-nums slashed-zero;
    }
    #canvas-container { width: 100vw; height: 100vh; position: absolute; top: 0; left: 0; }
    .glass-panel { background: rgba(18, 18, 24, 0.92); backdrop-filter: blur(16px); border: 1px solid rgba(255, 255, 255, 0.12); }
    .custom-slider { -webkit-appearance: none; appearance: none; height: 8px; border-radius: 4px; background: #27272a; outline: none; cursor: pointer; }
    .custom-slider::-webkit-slider-thumb { -webkit-appearance: none; appearance: none; width: 22px; height: 22px; border-radius: 50%; background: #14b8a6; cursor: grab; box-shadow: 0 0 12px #14b8a6; border: 2px solid #ffffff; }
    .custom-slider:active::-webkit-slider-thumb { cursor: grabbing; transform: scale(1.15); }
  
    @keyframes parasympathetic-breath {
      0%, 100% { opacity: 0.85; filter: drop-shadow(0 0 6px rgba(20, 184, 166, 0.4)); }
      50% { opacity: 1.0; filter: drop-shadow(0 0 14px rgba(20, 184, 166, 0.7)); }
    }
    .breath-pulse { animation: parasympathetic-breath 10s infinite ease-in-out; }

    
    /* ==========================================================================
       DIRECT CURVILINEAR GEOMETRY (Anti-Threat Neuro-Ergonomic Standard)
       Humans naturally perceive sharp corners as subconscious physical hazards.
       Direct continuous curves (squircles, superellipses & smooth Bézier splines)
       induce parasympathetic calm and optimize touch hitboxes.
       ========================================================================== */
    .direct-curve-capsule,
    .direct-curve-chip {
      border-radius: 9999px !important;
    }
    .direct-curve-dock {
      border-radius: 28px !important;
      backdrop-filter: blur(24px) !important;
      -webkit-backdrop-filter: blur(24px) !important;
    }
    .direct-curve-card {
      border-radius: 22px !important;
      backdrop-filter: blur(18px) !important;
      -webkit-backdrop-filter: blur(18px) !important;
    }
    .direct-curve-badge {
      border-radius: 9999px !important;
      padding: 0.35rem 0.85rem !important;
    }
    .callout-spline {
      fill: none;
      stroke-width: 1.75;
      stroke-linecap: round;
      filter: drop-shadow(0 0 6px rgba(20, 184, 166, 0.65));
      transition: d 0.04s ease-out;
    }
    .spline-pulse-dot {
      r: 3.5;
      filter: drop-shadow(0 0 8px rgba(20, 184, 166, 0.9));
    }

    /* Retinal Irradiation & Circadian Ocular Softening (Helmholtz Anti-Halation Standard) */
    body.circadian-dusk {
      background-color: #0d0e12;
      color: #f3efe6;
    }
    body.circadian-night {
      background-color: #120e0a;
      color: #ede3d2;
    }
    /* Compensated for OLED glare, dilated pupils & astigmatic halation (-3.75% optical stroke dial-back) */
    body.pg-irrd-compensated, .pg-irrd-compensated {
      font-weight: 385 !important;
      letter-spacing: 0.025em !important;
      text-rendering: optimizeLegibility;
      -webkit-font-smoothing: antialiased;
      -moz-osx-font-smoothing: grayscale;
    }
    .circadian-night .glass-panel {
      background: rgba(22, 17, 13, 0.94) !important;
      border-color: rgba(217, 119, 6, 0.25) !important;
    }
    .circadian-dusk .glass-panel {
      background: rgba(18, 17, 22, 0.93) !important;
      border-color: rgba(245, 158, 11, 0.22) !important;
    }
    @media print {
      body { background: white !important; color: black !important; overflow: visible !important; }
      #canvas-container, header, main, #anatomical-compass, #diatom-hud, #procedure-stepper-panel, #hardware-callouts { display: none !important; }
      #care-card-modal { position: static !important; display: block !important; background: white !important; color: black !important; padding: 0 !important; width: 100% !important; }
      #care-card-modal .modal-content { border: none !important; box-shadow: none !important; background: white !important; color: black !important; max-width: 100% !important; padding: 20px !important; }
      #care-card-modal button { display: none !important; }
      #care-card-modal * { color: #18181b !important; border-color: #cbd5e1 !important; }
      .print-black { color: #18181b !important; }
    }

  </style>
</head>
<body class="text-zinc-100 select-none">
  <!-- POCKETGULL CSS-ONLY SKELETON LOADING SPLASH (Zero White-Flash Boot) -->
  <div id="pocketgull-skeleton-splash" style="position: absolute; inset: 0; z-index: 100; background: #09090b; display: flex; flex-direction: column; align-items: center; justify-content: center; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif; transition: opacity 0.4s ease, visibility 0.4s ease;">
    <div style="position: relative; width: 72px; height: 72px; margin-bottom: 18px;">
      <div style="position: absolute; inset: 0; border-radius: 50%; border: 2px solid rgba(20,184,166,0.15);"></div>
      <div style="position: absolute; inset: 0; border-radius: 50%; border: 2px solid transparent; border-top-color: #14b8a6; animation: pg-spin 1.1s cubic-bezier(0.55, 0.15, 0.45, 0.85) infinite;"></div>
      <div style="position: absolute; inset: 10px; border-radius: 50%; border: 2px solid transparent; border-bottom-color: #38bdf8; animation: pg-spin-rev 1.6s cubic-bezier(0.55, 0.15, 0.45, 0.85) infinite;"></div>
      <div style="position: absolute; inset: 0; display: flex; align-items: center; justify-content: center; font-size: 22px;">🦴</div>
    </div>
    <div style="font-weight: 800; font-size: 15px; letter-spacing: 0.12em; text-transform: uppercase; color: #5eead4; margin-bottom: 6px;">
      POCKETGULL 3D ENGINE
    </div>
    <div style="font-size: 12px; color: #71717a; font-family: monospace;">
      Compiling WebGL Shaders & Anatomical Triads...
    </div>
    <style>
      @keyframes pg-spin { 0% { transform: rotate(0deg); } 100% { transform: rotate(360deg); } }
      @keyframes pg-spin-rev { 0% { transform: rotate(360deg); } 100% { transform: rotate(0deg); } }
    </style>
  </div>

  <!-- 3D WebGL Canvas Layer -->
  <div id="canvas-container"></div>

  <!-- FLOATING 3D SCREEN-SPACE HARDWARE CALLOUT BADGES -->
  <div id="hardware-callouts" class="absolute inset-0 pointer-events-none z-20 overflow-hidden" style="display: none;">
    <!-- Direct Curved SVG Spline Whiskers Layer (Real-Time 3D Projection) -->
    <svg id="callout-curves-svg" class="absolute inset-0 w-full h-full pointer-events-none z-10" style="overflow: visible;">
      <defs>
        <linearGradient id="curve-grad-teal" x1="0%" y1="0%" x2="100%" y2="100%">
          <stop offset="0%" stop-color="#14b8a6" stop-opacity="0.2" />
          <stop offset="100%" stop-color="#2dd4bf" stop-opacity="0.95" />
        </linearGradient>
        <linearGradient id="curve-grad-amber" x1="0%" y1="0%" x2="100%" y2="100%">
          <stop offset="0%" stop-color="#d97706" stop-opacity="0.2" />
          <stop offset="100%" stop-color="#fbbf24" stop-opacity="0.95" />
        </linearGradient>
        <linearGradient id="curve-grad-slate" x1="0%" y1="0%" x2="100%" y2="100%">
          <stop offset="0%" stop-color="#64748b" stop-opacity="0.2" />
          <stop offset="100%" stop-color="#f1f5f9" stop-opacity="0.95" />
        </linearGradient>
        <linearGradient id="curve-grad-cyan" x1="0%" y1="0%" x2="100%" y2="100%">
          <stop offset="0%" stop-color="#0891b2" stop-opacity="0.2" />
          <stop offset="100%" stop-color="#38bdf8" stop-opacity="0.95" />
        </linearGradient>
      </defs>
      <path id="curve-path-button" class="callout-spline" stroke="url(#curve-grad-slate)" d="M0,0" />
      <circle id="curve-dot-button" class="spline-pulse-dot" fill="#f8fafc" cx="-100" cy="-100" />

      <path id="curve-path-graft" class="callout-spline" stroke="url(#curve-grad-amber)" d="M0,0" />
      <circle id="curve-dot-graft" class="spline-pulse-dot" fill="#fbbf24" cx="-100" cy="-100" />

      <path id="curve-path-anchor" class="callout-spline" stroke="url(#curve-grad-teal)" d="M0,0" />
      <circle id="curve-dot-anchor" class="spline-pulse-dot" fill="#2dd4bf" cx="-100" cy="-100" />

      <path id="curve-path-portals" class="callout-spline" stroke="url(#curve-grad-cyan)" d="M0,0" />
      <circle id="curve-dot-portals" class="spline-pulse-dot" fill="#38bdf8" cx="-100" cy="-100" />
    </svg>
    <!-- 1. Femoral Suspensory Button -->
    <div id="callout-button" class="absolute -translate-x-1/2 -translate-y-full mb-3 pointer-events-auto flex items-center gap-1.5 bg-zinc-950/90 border border-slate-400/80 px-2.5 py-1 rounded-xl shadow-[0_0_15px_rgba(203,213,225,0.3)] backdrop-blur-md text-[10px] whitespace-nowrap transition-transform duration-75">
      <span class="w-2 h-2 rounded-full bg-slate-200 shadow-[0_0_6px_#f1f5f9]"></span>
      <span class="font-bold text-slate-100">1. Femoral Cortical Button</span>
      <span class="text-slate-400 font-mono text-[9px]">(Titanium Plate)</span>
    </div>

    <!-- 2. Quadrupled Tendon Autograft -->
    <div id="callout-graft" class="absolute -translate-x-1/2 -translate-y-full mb-3 pointer-events-auto flex items-center gap-1.5 bg-zinc-950/90 border border-amber-500/80 px-2.5 py-1 rounded-xl shadow-[0_0_15px_rgba(245,158,11,0.3)] backdrop-blur-md text-[10px] whitespace-nowrap transition-transform duration-75">
      <span class="w-2 h-2 rounded-full bg-amber-400 shadow-[0_0_6px_#f59e0b]"></span>
      <span class="font-bold text-amber-300">2. Quadrupled Autograft</span>
      <span class="text-amber-200/70 font-mono text-[9px]">(Semitendinosus)</span>
    </div>

    <!-- 3. Diatom Biosilica Interference Anchor -->
    <div id="callout-anchor" class="absolute -translate-x-1/2 -translate-y-full mb-3 pointer-events-auto flex items-center gap-1.5 bg-zinc-950/90 border border-teal-500/80 px-2.5 py-1 rounded-xl shadow-[0_0_15px_rgba(20,184,166,0.4)] backdrop-blur-md text-[10px] whitespace-nowrap transition-transform duration-75">
      <span class="w-2 h-2 rounded-full bg-teal-400 animate-pulse shadow-[0_0_8px_#14b8a6]"></span>
      <span class="font-bold text-teal-300">3. Diatom Biosilica Anchor</span>
      <span class="text-teal-400/80 font-mono text-[9px]">(16 µm Porous Plug)</span>
    </div>

    <!-- 4. Keyhole Portals -->
    <div id="callout-portals" class="absolute -translate-x-1/2 -translate-y-full mb-3 pointer-events-auto flex items-center gap-1.5 bg-zinc-950/90 border border-cyan-500/80 px-2.5 py-1 rounded-xl shadow-[0_0_15px_rgba(6,182,212,0.4)] backdrop-blur-md text-[10px] whitespace-nowrap transition-transform duration-75">
      <span class="w-2 h-2 rounded-full bg-cyan-400 shadow-[0_0_6px_#06b6d4]"></span>
      <span class="font-bold text-cyan-300">4mm Arthroscopic Portals</span>
      <span class="text-cyan-200/70 font-mono text-[9px]">(AM Working / AL Camera)</span>
    </div>
  </div>

  <!-- 3D ANATOMICAL COMPASS (Anti-Fatigue Spatial Orientation) -->
  <div id="anatomical-compass" class="absolute top-16 left-3 z-20 glass-panel p-2 rounded-2xl border border-zinc-700/60 shadow-xl pointer-events-none select-none flex flex-col items-center justify-center w-16 h-16 transition-transform">
    <span id="compass-ant" class="text-teal-400 font-bold text-[10px] font-mono leading-none">A</span>
    <div class="flex items-center justify-between w-full px-1 my-0.5">
      <span id="compass-med" class="text-amber-400 font-bold text-[10px] font-mono leading-none">M</span>
      <span class="w-1.5 h-1.5 rounded-full bg-teal-500/80"></span>
      <span id="compass-lat" class="text-rose-400 font-bold text-[10px] font-mono leading-none">L</span>
    </div>
    <span id="compass-post" class="text-zinc-500 font-bold text-[10px] font-mono leading-none">P</span>
  </div>

  <!-- 🔬 4-STAGE RESTORATION STEPPER & EXPLODED HARDWARE SUITE -->
  <div id="procedure-stepper-panel" class="absolute top-36 left-3 z-20 glass-panel p-2.5 rounded-3xl border border-teal-500/40 shadow-2xl pointer-events-auto flex flex-col gap-1.5 w-60 bg-zinc-950/95 backdrop-blur-xl transition-all" style="display: none;">
    <div class="flex items-center justify-between px-1 border-b border-zinc-800 pb-1.5">
      <div class="flex items-center gap-1.5">
        <span class="w-2 h-2 rounded-full bg-teal-400 animate-pulse shadow-[0_0_6px_#14b8a6]"></span>
        <span class="font-brand-pocketgull text-[11px] font-bold tracking-wider text-teal-300 uppercase">SURGICAL RESTORATION</span>
      </div>
      <span id="procedure-stage-indicator" class="text-[9px] font-mono text-zinc-300 font-bold bg-teal-950/80 px-1.5 py-0.5 rounded border border-teal-800">STAGE 1/4</span>
    </div>

    <div class="grid grid-cols-2 gap-1 text-[10px] font-medium pt-0.5">
      <button id="btn-proc-stage-1" class="proc-stage-btn px-2.5 py-1.5 rounded-full bg-teal-600 text-white font-bold transition text-left flex items-center gap-1 border border-teal-400 shadow-[0_0_10px_rgba(20,184,166,0.3)]">
        <span>1️⃣</span> <span class="truncate">1. Portals</span>
      </button>
      <button id="btn-proc-stage-2" class="proc-stage-btn px-2.5 py-1.5 rounded-full bg-zinc-900 text-zinc-400 hover:text-white hover:bg-zinc-800 transition text-left flex items-center gap-1 border border-zinc-800">
        <span>2️⃣</span> <span class="truncate">2. Tunnels</span>
      </button>
      <button id="btn-proc-stage-3" class="proc-stage-btn px-2.5 py-1.5 rounded-full bg-zinc-900 text-zinc-400 hover:text-white hover:bg-zinc-800 transition text-left flex items-center gap-1 border border-zinc-800">
        <span>3️⃣</span> <span class="truncate">3. Graft</span>
      </button>
      <button id="btn-proc-stage-4" class="proc-stage-btn px-2.5 py-1.5 rounded-full bg-zinc-900 text-zinc-400 hover:text-white hover:bg-zinc-800 transition text-left flex items-center gap-1 border border-zinc-800">
        <span>4️⃣</span> <span class="truncate">4. Fixation</span>
      </button>
    </div>

    <!-- Exploded Hardware View Toggle & Next Stage Button -->
    <div class="flex items-center justify-between gap-1 pt-1 border-t border-zinc-800/80">
      <button id="btn-exploded-view" class="flex-1 px-3 py-1 rounded-full bg-zinc-900 hover:bg-zinc-800 text-amber-300 border border-amber-500/40 text-[10px] font-bold transition flex items-center justify-center gap-1" title="Displace Hardware Along Anatomical Vectors">
        <span>💥</span> <span id="exploded-view-label">Exploded View</span>
      </button>
      <button id="btn-proc-next" class="px-3 py-1 rounded-full bg-teal-900/80 hover:bg-teal-800 text-teal-200 border border-teal-500/50 text-[10px] font-bold transition flex items-center justify-center gap-1" title="Advance to Next Surgical Step">
        <span>Next ➔</span>
      </button>
    </div>
  </div>

  <!-- LIVE DIATOM / DICOM VOLUMETRIC SLICE HUD (Picture-in-Picture - Responsive Desktop, Clean Mobile) -->
  <div id="diatom-hud" class="hidden md:flex absolute top-16 right-3 z-20 glass-panel p-2.5 rounded-2xl border border-teal-500/40 shadow-2xl pointer-events-auto flex-col items-center gap-1.5 w-48 bg-zinc-950/90 backdrop-blur-xl transition-all">
    <div class="flex items-center justify-between w-full px-1">
      <div class="flex items-center gap-1.5">
        <span class="w-2 h-2 rounded-full bg-teal-400 animate-pulse shadow-[0_0_6px_#14b8a6]"></span>
        <span id="diatom-hud-title" class="font-brand-pocketgull text-[10px] font-bold tracking-wider text-teal-300 uppercase">DIATOM SCAN</span>
      </div>
      <div class="flex items-center gap-1">
        <button id="btn-diatom-mode-toggle" class="text-[9px] bg-teal-950/80 hover:bg-teal-900 text-teal-300 px-1.5 py-0.5 rounded-lg border border-teal-700/60 font-mono transition" title="Toggle Diatom Micro-CT vs Clinical MRI">🔬 CT</button>
        <button id="btn-diatom-hud-minimize" class="text-[10px] text-zinc-400 hover:text-white px-1 leading-none font-bold" title="Minimize / Expand HUD">—</button>
      </div>
    </div>

    <!-- 2D Canvas Viewport -->
    <div id="diatom-canvas-wrapper" class="relative rounded-2xl overflow-hidden border border-zinc-800 bg-black/90 w-44 h-44 flex items-center justify-center shadow-inner">
      <canvas id="diatom-slice-canvas" width="176" height="176" class="w-full h-full block"></canvas>
      <!-- Reticle Crosshair HUD Overlays -->
      <div class="absolute inset-0 pointer-events-none flex flex-col justify-between p-1.5 font-mono text-[8px] text-zinc-500">
        <div class="flex justify-between">
          <span id="hud-corner-tl" class="text-teal-400/80">SLICE 16/32</span>
          <span id="hud-corner-tr" class="text-amber-400/80">R: 42.8 µm</span>
        </div>
        <div class="flex justify-between items-end">
          <span id="hud-corner-bl" class="text-zinc-400">POCKETGULL</span>
          <span id="hud-corner-br" class="text-teal-300 font-bold">SAG 0.50</span>
        </div>
      </div>
    </div>

    <!-- Active Salutogenic Biomimetic Telemetry -->
    <div id="diatom-procedure-tag" class="w-full bg-teal-950/70 border border-teal-700/60 rounded-xl px-2 py-1.5 flex flex-col gap-0.5 text-[9px] font-mono text-zinc-300">
      <div class="flex items-center justify-between">
        <span class="text-teal-300 font-bold flex items-center gap-1">
          <span class="w-1.5 h-1.5 rounded-full bg-emerald-400 animate-pulse"></span>
          <span>SALUTOGENIC:</span>
        </span>
        <span id="diatom-scaffold-status" class="text-emerald-300 font-bold">Osteointegrating</span>
      </div>
      <div class="flex items-center justify-between text-[8px] text-zinc-400 border-t border-teal-900/60 pt-0.5">
        <span>16 µm Porous Frustule</span>
        <span id="diatom-affinity-tag" class="text-amber-300 font-bold">Osteoblast Matrix</span>
      </div>
    </div>
  </div>


  <!-- FIXED TOP NAVIGATION BAR (Zero Overlap with Bottom Dock) -->
  <header class="absolute top-2 left-2 right-2 z-30 flex flex-wrap items-center justify-between gap-2 pointer-events-none">
    <!-- Title & Plane Status -->
    <div class="glass-panel py-1.5 px-4 rounded-full pointer-events-auto flex items-center gap-2 border-teal-500/40 shadow-lg">
      <span class="w-2.5 h-2.5 rounded-full bg-teal-400 animate-pulse shadow-[0_0_8px_#14b8a6]"></span>
      <a href="https://www.kaggle.com/datasets/philgear/pocketgull-font" target="_blank" class="no-underline hover:opacity-80 transition flex items-center gap-1.5" title="Built with the PocketGull 1000 UPM Clinical Typography Superfamily (Click to view dataset on Kaggle)"><span id="header-mode-badge" class="font-brand-pocketgull text-xs font-bold tracking-wider text-teal-300 uppercase">POCKETGULL 3D MRI SCANNER v12</span><span class="text-[9px] bg-teal-900/90 text-teal-200 px-1.5 py-0.5 rounded border border-teal-600/70 font-mono">🔤 1000 UPM</span></a>
      <span class="text-zinc-500">|</span>
      <span id="header-plane-badge" class="text-[10px] bg-teal-950/80 text-teal-300 px-2 py-0.5 rounded-full border border-teal-800 font-mono">SAGITTAL</span>
      <span id="header-structure-badge" class="text-xs font-semibold text-zinc-200 hidden sm:inline">
        Intercondylar Notch, ACL, PCL
      </span>
    </div>

    <!-- Mode Selector & Controls -->
    <div class="flex items-center gap-1.5 pointer-events-auto">
      <!-- Auto Scan Play/Pause Button -->
      <button id="btn-auto-scan" class="glass-panel px-3.5 py-1.5 rounded-full bg-teal-600 hover:bg-teal-500 text-white text-xs font-bold transition flex items-center gap-1.5 shadow-lg border border-teal-400/50">
        <span id="scan-icon">▶</span> <span id="scan-label">Auto-Scan</span>
      </button>

      <!-- Physical Cross-Section Cut Toggle -->
      <button id="btn-clip-toggle" class="glass-panel px-3 py-1.5 rounded-full bg-zinc-900 hover:bg-zinc-800 text-teal-300 text-xs font-semibold transition border border-zinc-700/60 shadow-lg">
        ✂️ Cut: <span id="clip-status" class="text-white font-bold">ON</span>
      </button>

      <!-- Anatomical Layer Toggles (Bone / Muscle / Skin) -->
      <div class="glass-panel p-1 rounded-full flex items-center gap-1 border border-zinc-700/60 bg-zinc-900/90 text-xs shadow-lg">
        <button id="btn-layer-bone" class="px-3 py-1 rounded-full bg-teal-600 text-white font-bold transition text-[11px]" title="Cycle Bone: Solid -> Ghost X-Ray -> OFF">🦴 Bone: <span id="bone-mode-label" class="font-mono">Solid</span></button>
        <button id="btn-layer-muscle" class="px-3 py-1 rounded-full bg-rose-600 text-white font-bold transition text-[11px]" title="Cycle Muscle: Solid -> Ghost -> OFF">🥩 Muscle: <span id="muscle-mode-label" class="font-mono">Solid</span></button>
        <button id="btn-layer-skin" class="px-3 py-1 rounded-full bg-amber-600/80 text-white font-bold transition text-[11px]" title="Cycle Skin: Ghost -> Solid -> OFF">🧖 Skin: <span id="skin-mode-label" class="font-mono">Ghost</span></button>
        <button id="btn-layer-mri" class="px-3 py-1 rounded-full bg-zinc-800 text-zinc-400 font-bold transition text-[11px]" title="Toggle 3.0T MRI Scanner Bore & Cryo-Gantry (OFF -> Ghost Outline -> Full Gantry)">🧲 MRI: <span id="mri-mode-label" class="font-mono">OFF</span></button>
      </div>

      
      <!-- Laterality Switcher (Anti-Fatigue Left/Right Inversion Guard) -->
      <button id="btn-laterality" class="glass-panel px-3 py-1.5 rounded-full bg-zinc-900 hover:bg-zinc-800 text-teal-300 text-xs font-semibold transition border border-zinc-700/60 shadow-lg" title="Toggle Knee Laterality">
        🦵 <span id="laterality-label" class="text-white font-bold">Right Knee</span>
      </button>

      <!-- 🔬 Procedure Walkthrough & Diatom Scaffold Toggle -->
      <button id="btn-procedure-toggle" class="glass-panel px-3 py-1.5 rounded-full bg-zinc-900 hover:bg-zinc-800 text-teal-300 text-xs font-semibold transition border border-teal-500/40 shadow-lg flex items-center gap-1" title="Toggle 3D Surgical Reconstruction & Diatom Biosilica Anchor">
        🔬 <span>Procedure:</span> <span id="procedure-status" class="text-white font-bold font-mono">OFF</span>
      </button>

      <!-- Mode Switcher -->
      <div class="glass-panel p-1 rounded-full flex items-center gap-1 border border-zinc-700/60 bg-zinc-900/90 text-xs shadow-lg">
        <button id="btn-mode-doc" class="px-3 py-1 rounded-full bg-teal-600 text-white font-bold transition">🩺 Doctor</button>
        <button id="btn-mode-pat" class="px-3 py-1 rounded-full text-zinc-400 hover:text-white transition">🤝 Patient</button>
      </div>

      <!-- 👁️ Circadian Ocular & Layer Softening Dock -->
      <div id="chrono-dock" class="glass-panel p-1 rounded-full flex items-center gap-1 border border-zinc-700/60 bg-zinc-900/90 text-xs shadow-lg" title="Circadian Softening of the Eyes & Layers (Time-of-Day Adaptive)">
        <button id="btn-chrono-day" class="px-2.5 py-1 rounded-full bg-teal-600 text-white font-bold transition text-[11px]" title="Daylight Acuity (5500K - High Cortical Contrast)">☀️ Day</button>
        <button id="btn-chrono-dusk" class="px-2.5 py-1 rounded-full text-zinc-400 hover:text-amber-300 transition text-[11px]" title="Golden Dusk (3500K - Softened Contrast)">🌅 Dusk</button>
        <button id="btn-chrono-night" class="px-2.5 py-1 rounded-full text-zinc-400 hover:text-amber-400 transition text-[11px]" title="Candle Obsidian (2000K - Melatonin Safe, Retinal Anti-Halation)">🕯️ Night</button>
        <button id="btn-chrono-auto" class="px-2 py-1 rounded-full text-zinc-400 hover:text-white transition text-[11px] font-mono" title="Auto Circadian Tracking (Local Hour)">⏱️ <span id="chrono-auto-label">Auto</span></button>
      </div>

      <!-- Mobile Diatom Micro-CT Toggle (clean on mobile, available on demand) -->
      <button id="btn-mobile-diatom-toggle" class="md:hidden glass-panel px-2.5 py-1.5 rounded-full bg-teal-950/80 hover:bg-teal-900/90 text-teal-300 text-xs font-bold transition flex items-center gap-1 border border-teal-500/40" title="Toggle Diatom Micro-CT HUD">
        🔬 <span>CT</span>
      </button>

      <button id="btn-reset-cam" class="glass-panel px-3 py-1.5 rounded-full bg-zinc-900 hover:bg-zinc-800 text-zinc-300 text-xs font-semibold transition border border-zinc-700/50 shadow-lg">
        🎯 Reset
      </button>

      <!-- ⛶ Cinema Fullscreen Mode Toggle -->
      <button id="btn-fullscreen" class="glass-panel px-3 py-1.5 rounded-full bg-teal-950/80 hover:bg-teal-900/90 text-teal-300 text-xs font-bold transition flex items-center gap-1.5 border border-teal-500/50 shadow-[0_0_12px_rgba(20,184,166,0.3)]" title="Toggle Fullscreen 3D View">
        <span id="fs-icon">⛶</span> <span id="fs-label">Fullscreen</span>
      </button>

      <!-- Shift Sanctuary: Green Happy Knee & Knee Rider Cinema -->
      <a href="knee_cinema_of_hills.html" class="glass-panel px-3.5 py-1.5 rounded-full bg-emerald-950/80 hover:bg-emerald-900/90 text-emerald-300 text-xs font-bold transition flex items-center gap-1.5 shadow-lg border border-emerald-500/50 no-underline" title="Shift Sanctuary: Green Happy Knee & Knee Rider">
        🌱 <span>Happy Knee & Rider</span>
      </a>
    </div>
  </header>

  <!-- UNIFIED BOTTOM DOCK (Single Non-Overlapping Container) -->
  <main class="absolute bottom-2 left-2 right-2 z-30 max-w-4xl mx-auto glass-panel p-3 rounded-3xl border border-zinc-700/70 shadow-2xl pointer-events-auto space-y-2.5">
    
    <!-- Row 1: Slicing Plane Selector & Continuous Depth Slider -->
    <div class="flex flex-col sm:flex-row items-center justify-between gap-3">
      <!-- Plane Buttons -->
      <div class="flex items-center gap-1 bg-zinc-950 p-1 rounded-full border border-zinc-800 text-xs shrink-0">
        <button id="btn-plane-sagittal" class="px-3.5 py-1.5 rounded-full bg-teal-600 text-white font-bold transition">Sagittal</button>
        <button id="btn-plane-coronal" class="px-3.5 py-1.5 rounded-full text-zinc-400 hover:text-white transition">Coronal</button>
        <button id="btn-plane-axial" class="px-3.5 py-1.5 rounded-full text-zinc-400 hover:text-white transition">Axial</button>
      </div>

      <!-- Slicing Depth Slider & Live Coordinate Indicator -->
      <div class="flex-1 w-full flex items-center gap-3">
        <span id="depth-lo-label" class="text-xs text-zinc-400 font-bold whitespace-nowrap min-w-[55px]">Medial</span>
        <input type="range" id="depth-slider" min="0" max="1" step="0.005" value="0.50" class="flex-1 custom-slider" aria-label="MRI Slice Depth Slider">
        <span id="depth-hi-label" class="text-xs text-zinc-400 font-bold whitespace-nowrap min-w-[55px] text-right">Lateral</span>
        <span id="depth-val" class="text-teal-300 font-mono font-bold text-xs bg-zinc-950 px-2.5 py-1 rounded-xl border border-zinc-800 min-w-[52px] text-center shadow-inner">0.50</span>
      </div>
    </div>

    <!-- Row 2: 12 Canonical Findings with Kaizen Indicators & Heuristic Filter -->
    <div class="space-y-1.5">
      <!-- Kaizen Acuity Filter Chips -->
      <div class="flex items-center justify-between gap-1 text-[10px] font-mono text-zinc-400 px-0.5">
        <span class="flex items-center gap-1.5 font-bold uppercase tracking-wider text-zinc-300">
          <span>🏷️ Kaizen Acuity:</span>
        </span>
        <div class="flex items-center gap-1">
          <button id="filter-kaizen-all" class="px-2 py-0.5 rounded-md bg-zinc-800 text-teal-300 font-bold border border-teal-500/40">All (12)</button>
          <button id="filter-kaizen-red" class="px-2 py-0.5 rounded-md bg-zinc-950 text-rose-400 border border-zinc-800 hover:border-rose-500 font-bold flex items-center gap-1"><span class="w-1.5 h-1.5 rounded-full bg-rose-500"></span>🔴 Acute (4)</button>
          <button id="filter-kaizen-amber" class="px-2 py-0.5 rounded-md bg-zinc-950 text-amber-400 border border-zinc-800 hover:border-amber-500 font-bold flex items-center gap-1"><span class="w-1.5 h-1.5 rounded-full bg-amber-400"></span>🟡 Watch (8)</button>
        </div>
      </div>

      <!-- 12 Finding Buttons with Kaizen Indicator Dots -->
      <div class="grid grid-cols-4 sm:grid-cols-6 gap-1 text-[11px]">
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="ACL" data-k="red"><span class="w-1.5 h-1.5 rounded-full bg-rose-500 shrink-0"></span><span>ACL Tear</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="MCL" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>MCL Sprain</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Medial Meniscus" data-k="red"><span class="w-1.5 h-1.5 rounded-full bg-rose-500 shrink-0"></span><span>Med. Meniscus</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Lateral Meniscus" data-k="red"><span class="w-1.5 h-1.5 rounded-full bg-rose-500 shrink-0"></span><span>Lat. Meniscus</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Medial OA" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>Medial OA</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Lateral OA" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>Lateral OA</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="PF OA" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>PF OA</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Effusion" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>Effusion</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Synovitis" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>Synovitis</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Baker's" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>Baker's Cyst</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Contusion" data-k="amber"><span class="w-1.5 h-1.5 rounded-full bg-amber-400 shrink-0"></span><span>Contusion</span></button>
        <button class="finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium flex items-center justify-center gap-1" data-f="Fracture" data-k="red"><span class="w-1.5 h-1.5 rounded-full bg-rose-500 shrink-0"></span><span>Fracture</span></button>
      </div>
    </div>

    <!-- Row 3: Live Information Bar (Updates in Place - Zero Floating Stack Collision) -->
    <div id="live-info-panel" class="bg-zinc-950/90 p-2.5 rounded-2xl border border-zinc-800 text-xs">
      
      <!-- Doctor Mode Details with Kaizen & Heuristic Engine -->
      <div id="dock-doc-content" class="space-y-1">
        <div class="flex items-center justify-between text-[10px] font-bold">
          <div class="flex items-center gap-1.5">
            <span id="dock-doc-kaizen-badge" class="px-2 py-0.5 rounded-full bg-rose-950/80 text-rose-300 border border-rose-800/80 font-mono tracking-wider uppercase text-[10px]">🔴 ANDON RED | ACUTE TEAR</span>
            <span class="text-zinc-500">|</span>
            <span id="dock-doc-title" class="text-white font-sans text-xs">Intercondylar Notch, ACL, PCL, Tibial Spines</span>
          </div>
          <span id="dock-doc-plane" class="text-zinc-500 font-mono">SAGITTAL 0.50</span>
        </div>
        <p id="dock-doc-desc" class="text-[11px] text-zinc-400 font-sans leading-relaxed">
          Central sagittal cuts capture the anterior cruciate ligament extending from lateral femoral condyle to the tibial plateau.
        </p>
        <div class="text-[11px] text-zinc-300 font-sans pt-0.5 border-t border-zinc-900 flex flex-wrap items-center justify-between gap-1.5">
          <div class="flex items-center gap-1.5">
            <span class="text-amber-400 font-bold text-[10px] uppercase font-mono">⚡ Heuristic:</span>
            <span id="dock-doc-heuristic" class="text-teal-300 font-medium">Blumensaat Line & Pivot-Shift Rule</span>
          </div>
          <div class="flex items-center gap-1 text-zinc-400 text-[10px]">
            <span class="text-zinc-500 font-bold uppercase">Prior:</span>
            <span id="dock-doc-triad">ACL tears co-occur with <b class="text-amber-300">Contusion 88%</b> and <b class="text-rose-300">Med. Meniscus 64%</b></span>
          </div>
        </div>
      
        <!-- 1-Click RSNA RadLex Structured Macro (Anti-Burnout Dictation) -->
        <div class="mt-2 p-2 rounded-xl bg-zinc-900/90 border border-zinc-800 flex flex-col sm:flex-row sm:items-center justify-between gap-2">
          <div class="flex-1 min-w-0">
            <div class="flex items-center gap-1.5 text-[10px] text-zinc-400 font-mono font-bold mb-0.5">
              <span class="text-teal-400">📋 RSNA RadLex Structured Macro:</span>
              <span class="text-zinc-600">|</span>
              <span class="text-zinc-500 text-[9px]">1-Click PACS Insertion</span>
            </div>
            <p id="dock-radlex-text" class="text-[11px] text-zinc-300 font-mono leading-relaxed select-all truncate">
              IMPRESSION: Complete disruption of the midsubstance anterior cruciate ligament with associated pivot-shift impaction bone contusion.
            </p>
          </div>
          <button id="btn-copy-radlex" class="shrink-0 px-3 py-1.5 rounded-xl bg-teal-600 hover:bg-teal-500 text-white font-bold text-xs transition shadow flex items-center justify-center gap-1.5 active:scale-95" title="Copy structured text to PACS clipboard">
            <span>📋</span> <span id="copy-btn-label">Copy to PACS</span>
          </button>
        </div>
      </div>

      <!-- Patient Mode Details -->
      <div id="dock-pat-content" class="hidden space-y-1.5">
        <div class="flex items-center justify-between text-xs font-bold">
          <span class="text-teal-300 flex items-center gap-1.5">
            <span>🤝 Condition:</span>
            <span id="dock-pat-title" class="text-white">Torn Front Stabilizer (ACL)</span>
          </span>
          <span class="text-[10px] text-zinc-500 uppercase tracking-wider">Patient Explanation</span>
        </div>
        <div class="grid grid-cols-1 sm:grid-cols-3 gap-2 text-[11px] font-sans">
          <div class="bg-zinc-900/80 p-2 rounded-xl border border-zinc-800">
            <span class="font-bold text-teal-400 block mb-0.5">💡 What Happened:</span>
            <p id="dock-pat-why" class="text-zinc-400">When your knee pivoted during your twist, the main central stabilizer ligament stretched past its limit and tore.</p>
          </div>
          <div class="bg-zinc-900/80 p-2 rounded-xl border border-zinc-800">
            <span class="font-bold text-amber-400 block mb-0.5">💥 Why It Aches:</span>
            <p id="dock-pat-sens" class="text-zinc-400">Explains why your knee felt like it 'gave way' and why walking down stairs feels unstable.</p>
          </div>
          <div class="bg-zinc-900/80 p-2 rounded-xl border border-zinc-800">
            <span class="font-bold text-emerald-400 block mb-0.5">🌱 3-Act Recovery Path:</span>
            <p id="dock-pat-rec" class="text-zinc-400">Act I: Reduce fluid swelling with RICE. Act II: Hamstring dynamic seatbelt. Act III: Safe return to sport.</p>
          </div>
        </div>
      
        <!-- 1-Click Patient Discharge Care Card (30-Second Hand-Off) -->
        <div class="flex items-center justify-between pt-1 border-t border-zinc-800/80">
          <span class="text-[10px] text-zinc-500 font-mono">PocketGull Clinical 3-Act Patient Standard</span>
          <button id="btn-print-care-card" class="px-3 py-1 rounded-xl bg-teal-600 hover:bg-teal-500 text-white font-bold text-xs transition shadow flex items-center gap-1.5 active:scale-95">
            <span>🖨️</span> <span>1-Page Patient Care Card</span>
          </button>
        </div>
      </div>

    </div>
  </main>

  <!-- 1-PAGE PATIENT CARE CARD MODAL -->
  <div id="care-card-modal" class="hidden fixed inset-0 z-50 bg-black/85 backdrop-blur-md flex items-center justify-center p-4">
    <div class="modal-content glass-panel bg-zinc-950 border border-zinc-700 max-w-xl w-full p-6 rounded-3xl shadow-2xl text-zinc-100 space-y-4">
      <div class="flex items-center justify-between border-b border-zinc-800 pb-3">
        <div>
          <div class="flex items-center gap-2">
            <span class="w-3 h-3 rounded-full bg-teal-400"></span>
            <h2 class="font-brand-pocketgull text-base font-bold text-white tracking-wide">POCKETGULL CLINICAL CARE CARD</h2>
          </div>
          <p class="text-xs text-zinc-400">Physician-Validated Knee Recovery Guide</p>
        </div>
        <div class="text-right">
          <span id="modal-lat-badge" class="px-2.5 py-0.5 rounded-full bg-zinc-800 text-teal-300 font-mono text-[10px] border border-zinc-700">Right Knee</span>
        </div>
      </div>

      <div class="bg-zinc-900/90 p-3 rounded-2xl border border-zinc-800">
        <span class="text-[10px] font-mono text-teal-400 uppercase tracking-wider block font-bold">Identified Condition</span>
        <h3 id="modal-finding-title" class="text-sm font-bold text-white mt-0.5">Torn Front Stabilizer (ACL)</h3>
      </div>

      <div class="grid grid-cols-1 sm:grid-cols-3 gap-2.5 text-xs">
        <div class="bg-zinc-900/60 p-3 rounded-2xl border border-zinc-800/80">
          <span class="font-bold text-teal-400 block mb-1">💡 What Happened:</span>
          <p id="modal-why" class="text-zinc-300 text-[11px] leading-relaxed"></p>
        </div>
        <div class="bg-zinc-900/60 p-3 rounded-2xl border border-zinc-800/80">
          <span class="font-bold text-amber-400 block mb-1">💥 Why It Aches:</span>
          <p id="modal-sens" class="text-zinc-300 text-[11px] leading-relaxed"></p>
        </div>
        <div class="bg-zinc-900/60 p-3 rounded-2xl border border-zinc-800/80">
          <span class="font-bold text-emerald-400 block mb-1">🌱 3-Act Recovery:</span>
          <p id="modal-rec" class="text-zinc-300 text-[11px] leading-relaxed"></p>
        </div>
      </div>

      <!-- 🔬 BIOMIMETIC SURGICAL PROCEDURE & DIATOM SCAFFOLD SECTION -->
      <div id="modal-procedure-section" class="bg-teal-950/40 p-3 rounded-2xl border border-teal-800/60 space-y-1.5">
        <div class="flex items-center justify-between">
          <div class="flex items-center gap-1.5">
            <span class="w-2 h-2 rounded-full bg-teal-400"></span>
            <span class="text-[11px] font-bold text-teal-300 uppercase tracking-wider font-mono">Biomimetic Surgical Procedure & Diatom Scaffold</span>
          </div>
          <span class="text-[10px] bg-teal-900/80 text-teal-200 px-2 py-0.5 rounded-full font-mono">Micro-CT Validated</span>
        </div>
        <p id="modal-procedure-text" class="text-zinc-300 text-[11px] leading-relaxed">
          Arthroscopic anatomical reconstruction utilizes 4mm keyhole portals. A quadrupled tendon autograft is routed through precision bone tunnels and secured with a <strong>porous diatomaceous biosilica anchor</strong>. This biomimetic scaffold stimulates native osteoblasts to weave through the graft during Act I (Weeks 1–4), transforming synthetic mechanical fixation into permanent, living bone integration.
        </p>
      </div>

      <!-- 🌱 SALUTOGENIC SENSE OF COHERENCE (SOC) RECOVERY MATRIX -->
      <div class="bg-zinc-900/80 p-3 rounded-2xl border border-emerald-800/50 space-y-1.5">
        <div class="flex items-center justify-between">
          <div class="flex items-center gap-1.5">
            <span class="w-2 h-2 rounded-full bg-emerald-400 animate-pulse"></span>
            <span class="text-[11px] font-bold text-emerald-300 uppercase tracking-wider font-mono">Salutogenic Sense of Coherence (SOC) Triad</span>
          </div>
          <span class="text-[9px] bg-emerald-950/80 text-emerald-300 border border-emerald-800/80 px-2 py-0.5 rounded-full font-mono">Antonovsky Model</span>
        </div>
        <div class="grid grid-cols-1 sm:grid-cols-3 gap-2 pt-0.5 text-[10px]">
          <div class="bg-zinc-950/70 p-2 rounded-xl border border-zinc-800/80">
            <span class="font-bold text-teal-300 block mb-0.5">1. Comprehensibility</span>
            <p class="text-zinc-400 leading-relaxed text-[9.5px]">Visual 3D clarity replaces fear. The joint is not ruined; it underwent a defined physical pivot load that is fully explainable.</p>
          </div>
          <div class="bg-zinc-950/70 p-2 rounded-xl border border-zinc-800/80">
            <span class="font-bold text-amber-300 block mb-0.5">2. Manageability</span>
            <p class="text-zinc-400 leading-relaxed text-[9.5px]">Biological stepped pacing. Act I settles fluid, Act II recruits the hamstring dynamic seatbelt, Act III restores kinetic force.</p>
          </div>
          <div class="bg-zinc-950/70 p-2 rounded-xl border border-zinc-800/80">
            <span class="font-bold text-emerald-300 block mb-0.5">3. Meaningfulness</span>
            <p class="text-zinc-400 leading-relaxed text-[9.5px]">Healing is an innate active biological baseline. The Diatom biosilica scaffold works *with* your bone to invite osteoblasts in.</p>
          </div>
        </div>
      </div>

      <div class="flex items-center justify-between pt-2 border-t border-zinc-800">
        <span class="text-[10px] text-zinc-500 font-mono">For patient education only. Follow your physician's specific instructions.</span>
        <div class="flex items-center gap-2">
          <button id="modal-btn-close" class="px-3 py-1.5 rounded-xl bg-zinc-800 hover:bg-zinc-700 text-zinc-300 font-bold text-xs transition">Close</button>
          <button id="modal-btn-print" class="px-4 py-1.5 rounded-xl bg-teal-600 hover:bg-teal-500 text-white font-bold text-xs transition shadow flex items-center gap-1.5">🖨️ Print Card</button>
        </div>
      </div>
    </div>
  </div>


  <script>
    // --- DEFENSIVE PRIVACY SHIELD & WEBGL GUARD ---
    if (typeof THREE === 'undefined') {
      console.warn("Three.js not loaded. Checking if blocked by browser privacy shield...");
      const notice = document.createElement('div');
      notice.className = "absolute inset-0 z-50 flex items-center justify-center p-4 bg-zinc-950 text-white";
      notice.innerHTML = `
        <div class="max-w-md p-6 rounded-2xl bg-zinc-900 border border-amber-500/60 shadow-2xl text-center space-y-3">
          <div class="text-3xl">🛡️</div>
          <h2 class="text-base font-bold text-amber-300">WebGL Library Blocked by Privacy Extension</h2>
          <p class="text-xs text-zinc-300 leading-relaxed">
            Your browser's privacy tool (e.g. <b>Privacy Badger</b> or <b>uBlock Origin</b>) prevented the WebGL 3D libraries (<code>cdnjs.cloudflare.com</code> / <code>cdn.jsdelivr.net</code>) from loading.
          </p>
          <div class="p-3 bg-zinc-950 rounded-xl text-left font-mono text-[11px] text-teal-300 border border-zinc-800">
            • Open Privacy Badger in your browser toolbar<br>
            • Set <b>cdnjs.cloudflare.com</b> to Allow (Green slider)<br>
            • Set <b>cdn.jsdelivr.net</b> to Allow (Green slider)<br>
            • Click Refresh below to load the 3D Knee simulator
          </div>
          <button onclick="window.location.reload()" class="w-full py-2.5 rounded-xl bg-teal-600 hover:bg-teal-500 text-white font-bold text-xs transition shadow-lg">
            🔄 Refresh 3D Simulator
          </button>
        </div>
      `;
      document.body.appendChild(notice);
    }

    // --- GLOBAL ENGINE STATE & RENDER FLAGS (HOISTED FOR TDZ SAFETY) ---
    let renderRequested = false;
    let isAnimating = false;
    let currentCircadianTone = 'day';
    let isCircadianAuto = true;

    function requestRender() {
      if (!renderRequested && !document.hidden && typeof render === 'function') {
        renderRequested = true;
        requestAnimationFrame(render);
      }
    }

    
    // --- INLINED THREE.JS ORBITCONTROLS (100% AIR-GAPPED & PRIVACY BADGER SAFE) ---
    if (typeof THREE !== 'undefined' && typeof THREE.OrbitControls !== 'function') {
( function () {

	// Unlike TrackballControls, it maintains the "up" direction object.up (+Y by default).
	//
	//    Orbit - left mouse / touch: one-finger move
	//    Zoom - middle mouse, or mousewheel / touch: two-finger spread or squish
	//    Pan - right mouse, or left mouse + ctrl/meta/shiftKey, or arrow keys / touch: two-finger move

	const _changeEvent = {
		type: 'change'
	};
	const _startEvent = {
		type: 'start'
	};
	const _endEvent = {
		type: 'end'
	};

	class OrbitControls extends THREE.EventDispatcher {

		constructor( object, domElement ) {

			super();
			if ( domElement === undefined ) console.warn( 'THREE.OrbitControls: The second parameter "domElement" is now mandatory.' );
			if ( domElement === document ) console.error( 'THREE.OrbitControls: "document" should not be used as the target "domElement". Please use "renderer.domElement" instead.' );
			this.object = object;
			this.domElement = domElement; // Set to false to disable this control

			this.enabled = true; // "target" sets the location of focus, where the object orbits around

			this.target = new THREE.Vector3(); // How far you can dolly in and out ( PerspectiveCamera only )

			this.minDistance = 0;
			this.maxDistance = Infinity; // How far you can zoom in and out ( OrthographicCamera only )

			this.minZoom = 0;
			this.maxZoom = Infinity; // How far you can orbit vertically, upper and lower limits.
			// Range is 0 to Math.PI radians.

			this.minPolarAngle = 0; // radians

			this.maxPolarAngle = Math.PI; // radians
			// How far you can orbit horizontally, upper and lower limits.
			// If set, the interval [ min, max ] must be a sub-interval of [ - 2 PI, 2 PI ], with ( max - min < 2 PI )

			this.minAzimuthAngle = - Infinity; // radians

			this.maxAzimuthAngle = Infinity; // radians
			// Set to true to enable damping (inertia)
			// If damping is enabled, you must call controls.update() in your animation loop

			this.enableDamping = false;
			this.dampingFactor = 0.05; // This option actually enables dollying in and out; left as "zoom" for backwards compatibility.
			// Set to false to disable zooming

			this.enableZoom = true;
			this.zoomSpeed = 1.0; // Set to false to disable rotating

			this.enableRotate = true;
			this.rotateSpeed = 1.0; // Set to false to disable panning

			this.enablePan = true;
			this.panSpeed = 1.0;
			this.screenSpacePanning = true; // if false, pan orthogonal to world-space direction camera.up

			this.keyPanSpeed = 7.0; // pixels moved per arrow key push
			// Set to true to automatically rotate around the target
			// If auto-rotate is enabled, you must call controls.update() in your animation loop

			this.autoRotate = false;
			this.autoRotateSpeed = 2.0; // 30 seconds per orbit when fps is 60
			// The four arrow keys

			this.keys = {
				LEFT: 'ArrowLeft',
				UP: 'ArrowUp',
				RIGHT: 'ArrowRight',
				BOTTOM: 'ArrowDown'
			}; // Mouse buttons

			this.mouseButtons = {
				LEFT: THREE.MOUSE.ROTATE,
				MIDDLE: THREE.MOUSE.DOLLY,
				RIGHT: THREE.MOUSE.PAN
			}; // Touch fingers

			this.touches = {
				ONE: THREE.TOUCH.ROTATE,
				TWO: THREE.TOUCH.DOLLY_PAN
			}; // for reset

			this.target0 = this.target.clone();
			this.position0 = this.object.position.clone();
			this.zoom0 = this.object.zoom; // the target DOM element for key events

			this._domElementKeyEvents = null; //
			// public methods
			//

			this.getPolarAngle = function () {

				return spherical.phi;

			};

			this.getAzimuthalAngle = function () {

				return spherical.theta;

			};

			this.listenToKeyEvents = function ( domElement ) {

				domElement.addEventListener( 'keydown', onKeyDown );
				this._domElementKeyEvents = domElement;

			};

			this.saveState = function () {

				scope.target0.copy( scope.target );
				scope.position0.copy( scope.object.position );
				scope.zoom0 = scope.object.zoom;

			};

			this.reset = function () {

				scope.target.copy( scope.target0 );
				scope.object.position.copy( scope.position0 );
				scope.object.zoom = scope.zoom0;
				scope.object.updateProjectionMatrix();
				scope.dispatchEvent( _changeEvent );
				scope.update();
				state = STATE.NONE;

			}; // this method is exposed, but perhaps it would be better if we can make it private...


			this.update = function () {

				const offset = new THREE.Vector3(); // so camera.up is the orbit axis

				const quat = new THREE.Quaternion().setFromUnitVectors( object.up, new THREE.Vector3( 0, 1, 0 ) );
				const quatInverse = quat.clone().invert();
				const lastPosition = new THREE.Vector3();
				const lastQuaternion = new THREE.Quaternion();
				const twoPI = 2 * Math.PI;
				return function update() {

					const position = scope.object.position;
					offset.copy( position ).sub( scope.target ); // rotate offset to "y-axis-is-up" space

					offset.applyQuaternion( quat ); // angle from z-axis around y-axis

					spherical.setFromVector3( offset );

					if ( scope.autoRotate && state === STATE.NONE ) {

						rotateLeft( getAutoRotationAngle() );

					}

					if ( scope.enableDamping ) {

						spherical.theta += sphericalDelta.theta * scope.dampingFactor;
						spherical.phi += sphericalDelta.phi * scope.dampingFactor;

					} else {

						spherical.theta += sphericalDelta.theta;
						spherical.phi += sphericalDelta.phi;

					} // restrict theta to be between desired limits


					let min = scope.minAzimuthAngle;
					let max = scope.maxAzimuthAngle;

					if ( isFinite( min ) && isFinite( max ) ) {

						if ( min < - Math.PI ) min += twoPI; else if ( min > Math.PI ) min -= twoPI;
						if ( max < - Math.PI ) max += twoPI; else if ( max > Math.PI ) max -= twoPI;

						if ( min <= max ) {

							spherical.theta = Math.max( min, Math.min( max, spherical.theta ) );

						} else {

							spherical.theta = spherical.theta > ( min + max ) / 2 ? Math.max( min, spherical.theta ) : Math.min( max, spherical.theta );

						}

					} // restrict phi to be between desired limits


					spherical.phi = Math.max( scope.minPolarAngle, Math.min( scope.maxPolarAngle, spherical.phi ) );
					spherical.makeSafe();
					spherical.radius *= scale; // restrict radius to be between desired limits

					spherical.radius = Math.max( scope.minDistance, Math.min( scope.maxDistance, spherical.radius ) ); // move target to panned location

					if ( scope.enableDamping === true ) {

						scope.target.addScaledVector( panOffset, scope.dampingFactor );

					} else {

						scope.target.add( panOffset );

					}

					offset.setFromSpherical( spherical ); // rotate offset back to "camera-up-vector-is-up" space

					offset.applyQuaternion( quatInverse );
					position.copy( scope.target ).add( offset );
					scope.object.lookAt( scope.target );

					if ( scope.enableDamping === true ) {

						sphericalDelta.theta *= 1 - scope.dampingFactor;
						sphericalDelta.phi *= 1 - scope.dampingFactor;
						panOffset.multiplyScalar( 1 - scope.dampingFactor );

					} else {

						sphericalDelta.set( 0, 0, 0 );
						panOffset.set( 0, 0, 0 );

					}

					scale = 1; // update condition is:
					// min(camera displacement, camera rotation in radians)^2 > EPS
					// using small-angle approximation cos(x/2) = 1 - x^2 / 8

					if ( zoomChanged || lastPosition.distanceToSquared( scope.object.position ) > EPS || 8 * ( 1 - lastQuaternion.dot( scope.object.quaternion ) ) > EPS ) {

						scope.dispatchEvent( _changeEvent );
						lastPosition.copy( scope.object.position );
						lastQuaternion.copy( scope.object.quaternion );
						zoomChanged = false;
						return true;

					}

					return false;

				};

			}();

			this.dispose = function () {

				scope.domElement.removeEventListener( 'contextmenu', onContextMenu );
				scope.domElement.removeEventListener( 'pointerdown', onPointerDown );
				scope.domElement.removeEventListener( 'wheel', onMouseWheel );
				scope.domElement.removeEventListener( 'touchstart', onTouchStart );
				scope.domElement.removeEventListener( 'touchend', onTouchEnd );
				scope.domElement.removeEventListener( 'touchmove', onTouchMove );
				scope.domElement.ownerDocument.removeEventListener( 'pointermove', onPointerMove );
				scope.domElement.ownerDocument.removeEventListener( 'pointerup', onPointerUp );

				if ( scope._domElementKeyEvents !== null ) {

					scope._domElementKeyEvents.removeEventListener( 'keydown', onKeyDown );

				} //scope.dispatchEvent( { type: 'dispose' } ); // should this be added here?

			}; //
			// internals
			//


			const scope = this;
			const STATE = {
				NONE: - 1,
				ROTATE: 0,
				DOLLY: 1,
				PAN: 2,
				TOUCH_ROTATE: 3,
				TOUCH_PAN: 4,
				TOUCH_DOLLY_PAN: 5,
				TOUCH_DOLLY_ROTATE: 6
			};
			let state = STATE.NONE;
			const EPS = 0.000001; // current position in spherical coordinates

			const spherical = new THREE.Spherical();
			const sphericalDelta = new THREE.Spherical();
			let scale = 1;
			const panOffset = new THREE.Vector3();
			let zoomChanged = false;
			const rotateStart = new THREE.Vector2();
			const rotateEnd = new THREE.Vector2();
			const rotateDelta = new THREE.Vector2();
			const panStart = new THREE.Vector2();
			const panEnd = new THREE.Vector2();
			const panDelta = new THREE.Vector2();
			const dollyStart = new THREE.Vector2();
			const dollyEnd = new THREE.Vector2();
			const dollyDelta = new THREE.Vector2();

			function getAutoRotationAngle() {

				return 2 * Math.PI / 60 / 60 * scope.autoRotateSpeed;

			}

			function getZoomScale() {

				return Math.pow( 0.95, scope.zoomSpeed );

			}

			function rotateLeft( angle ) {

				sphericalDelta.theta -= angle;

			}

			function rotateUp( angle ) {

				sphericalDelta.phi -= angle;

			}

			const panLeft = function () {

				const v = new THREE.Vector3();
				return function panLeft( distance, objectMatrix ) {

					v.setFromMatrixColumn( objectMatrix, 0 ); // get X column of objectMatrix

					v.multiplyScalar( - distance );
					panOffset.add( v );

				};

			}();

			const panUp = function () {

				const v = new THREE.Vector3();
				return function panUp( distance, objectMatrix ) {

					if ( scope.screenSpacePanning === true ) {

						v.setFromMatrixColumn( objectMatrix, 1 );

					} else {

						v.setFromMatrixColumn( objectMatrix, 0 );
						v.crossVectors( scope.object.up, v );

					}

					v.multiplyScalar( distance );
					panOffset.add( v );

				};

			}(); // deltaX and deltaY are in pixels; right and down are positive


			const pan = function () {

				const offset = new THREE.Vector3();
				return function pan( deltaX, deltaY ) {

					const element = scope.domElement;

					if ( scope.object.isPerspectiveCamera ) {

						// perspective
						const position = scope.object.position;
						offset.copy( position ).sub( scope.target );
						let targetDistance = offset.length(); // half of the fov is center to top of screen

						targetDistance *= Math.tan( scope.object.fov / 2 * Math.PI / 180.0 ); // we use only clientHeight here so aspect ratio does not distort speed

						panLeft( 2 * deltaX * targetDistance / element.clientHeight, scope.object.matrix );
						panUp( 2 * deltaY * targetDistance / element.clientHeight, scope.object.matrix );

					} else if ( scope.object.isOrthographicCamera ) {

						// orthographic
						panLeft( deltaX * ( scope.object.right - scope.object.left ) / scope.object.zoom / element.clientWidth, scope.object.matrix );
						panUp( deltaY * ( scope.object.top - scope.object.bottom ) / scope.object.zoom / element.clientHeight, scope.object.matrix );

					} else {

						// camera neither orthographic nor perspective
						console.warn( 'WARNING: OrbitControls.js encountered an unknown camera type - pan disabled.' );
						scope.enablePan = false;

					}

				};

			}();

			function dollyOut( dollyScale ) {

				if ( scope.object.isPerspectiveCamera ) {

					scale /= dollyScale;

				} else if ( scope.object.isOrthographicCamera ) {

					scope.object.zoom = Math.max( scope.minZoom, Math.min( scope.maxZoom, scope.object.zoom * dollyScale ) );
					scope.object.updateProjectionMatrix();
					zoomChanged = true;

				} else {

					console.warn( 'WARNING: OrbitControls.js encountered an unknown camera type - dolly/zoom disabled.' );
					scope.enableZoom = false;

				}

			}

			function dollyIn( dollyScale ) {

				if ( scope.object.isPerspectiveCamera ) {

					scale *= dollyScale;

				} else if ( scope.object.isOrthographicCamera ) {

					scope.object.zoom = Math.max( scope.minZoom, Math.min( scope.maxZoom, scope.object.zoom / dollyScale ) );
					scope.object.updateProjectionMatrix();
					zoomChanged = true;

				} else {

					console.warn( 'WARNING: OrbitControls.js encountered an unknown camera type - dolly/zoom disabled.' );
					scope.enableZoom = false;

				}

			} //
			// event callbacks - update the object state
			//


			function handleMouseDownRotate( event ) {

				rotateStart.set( event.clientX, event.clientY );

			}

			function handleMouseDownDolly( event ) {

				dollyStart.set( event.clientX, event.clientY );

			}

			function handleMouseDownPan( event ) {

				panStart.set( event.clientX, event.clientY );

			}

			function handleMouseMoveRotate( event ) {

				rotateEnd.set( event.clientX, event.clientY );
				rotateDelta.subVectors( rotateEnd, rotateStart ).multiplyScalar( scope.rotateSpeed );
				const element = scope.domElement;
				rotateLeft( 2 * Math.PI * rotateDelta.x / element.clientHeight ); // yes, height

				rotateUp( 2 * Math.PI * rotateDelta.y / element.clientHeight );
				rotateStart.copy( rotateEnd );
				scope.update();

			}

			function handleMouseMoveDolly( event ) {

				dollyEnd.set( event.clientX, event.clientY );
				dollyDelta.subVectors( dollyEnd, dollyStart );

				if ( dollyDelta.y > 0 ) {

					dollyOut( getZoomScale() );

				} else if ( dollyDelta.y < 0 ) {

					dollyIn( getZoomScale() );

				}

				dollyStart.copy( dollyEnd );
				scope.update();

			}

			function handleMouseMovePan( event ) {

				panEnd.set( event.clientX, event.clientY );
				panDelta.subVectors( panEnd, panStart ).multiplyScalar( scope.panSpeed );
				pan( panDelta.x, panDelta.y );
				panStart.copy( panEnd );
				scope.update();

			}

			function handleMouseUp( ) { // no-op
			}

			function handleMouseWheel( event ) {

				if ( event.deltaY < 0 ) {

					dollyIn( getZoomScale() );

				} else if ( event.deltaY > 0 ) {

					dollyOut( getZoomScale() );

				}

				scope.update();

			}

			function handleKeyDown( event ) {

				let needsUpdate = false;

				switch ( event.code ) {

					case scope.keys.UP:
						pan( 0, scope.keyPanSpeed );
						needsUpdate = true;
						break;

					case scope.keys.BOTTOM:
						pan( 0, - scope.keyPanSpeed );
						needsUpdate = true;
						break;

					case scope.keys.LEFT:
						pan( scope.keyPanSpeed, 0 );
						needsUpdate = true;
						break;

					case scope.keys.RIGHT:
						pan( - scope.keyPanSpeed, 0 );
						needsUpdate = true;
						break;

				}

				if ( needsUpdate ) {

					// prevent the browser from scrolling on cursor keys
					event.preventDefault();
					scope.update();

				}

			}

			function handleTouchStartRotate( event ) {

				if ( event.touches.length == 1 ) {

					rotateStart.set( event.touches[ 0 ].pageX, event.touches[ 0 ].pageY );

				} else {

					const x = 0.5 * ( event.touches[ 0 ].pageX + event.touches[ 1 ].pageX );
					const y = 0.5 * ( event.touches[ 0 ].pageY + event.touches[ 1 ].pageY );
					rotateStart.set( x, y );

				}

			}

			function handleTouchStartPan( event ) {

				if ( event.touches.length == 1 ) {

					panStart.set( event.touches[ 0 ].pageX, event.touches[ 0 ].pageY );

				} else {

					const x = 0.5 * ( event.touches[ 0 ].pageX + event.touches[ 1 ].pageX );
					const y = 0.5 * ( event.touches[ 0 ].pageY + event.touches[ 1 ].pageY );
					panStart.set( x, y );

				}

			}

			function handleTouchStartDolly( event ) {

				const dx = event.touches[ 0 ].pageX - event.touches[ 1 ].pageX;
				const dy = event.touches[ 0 ].pageY - event.touches[ 1 ].pageY;
				const distance = Math.sqrt( dx * dx + dy * dy );
				dollyStart.set( 0, distance );

			}

			function handleTouchStartDollyPan( event ) {

				if ( scope.enableZoom ) handleTouchStartDolly( event );
				if ( scope.enablePan ) handleTouchStartPan( event );

			}

			function handleTouchStartDollyRotate( event ) {

				if ( scope.enableZoom ) handleTouchStartDolly( event );
				if ( scope.enableRotate ) handleTouchStartRotate( event );

			}

			function handleTouchMoveRotate( event ) {

				if ( event.touches.length == 1 ) {

					rotateEnd.set( event.touches[ 0 ].pageX, event.touches[ 0 ].pageY );

				} else {

					const x = 0.5 * ( event.touches[ 0 ].pageX + event.touches[ 1 ].pageX );
					const y = 0.5 * ( event.touches[ 0 ].pageY + event.touches[ 1 ].pageY );
					rotateEnd.set( x, y );

				}

				rotateDelta.subVectors( rotateEnd, rotateStart ).multiplyScalar( scope.rotateSpeed );
				const element = scope.domElement;
				rotateLeft( 2 * Math.PI * rotateDelta.x / element.clientHeight ); // yes, height

				rotateUp( 2 * Math.PI * rotateDelta.y / element.clientHeight );
				rotateStart.copy( rotateEnd );

			}

			function handleTouchMovePan( event ) {

				if ( event.touches.length == 1 ) {

					panEnd.set( event.touches[ 0 ].pageX, event.touches[ 0 ].pageY );

				} else {

					const x = 0.5 * ( event.touches[ 0 ].pageX + event.touches[ 1 ].pageX );
					const y = 0.5 * ( event.touches[ 0 ].pageY + event.touches[ 1 ].pageY );
					panEnd.set( x, y );

				}

				panDelta.subVectors( panEnd, panStart ).multiplyScalar( scope.panSpeed );
				pan( panDelta.x, panDelta.y );
				panStart.copy( panEnd );

			}

			function handleTouchMoveDolly( event ) {

				const dx = event.touches[ 0 ].pageX - event.touches[ 1 ].pageX;
				const dy = event.touches[ 0 ].pageY - event.touches[ 1 ].pageY;
				const distance = Math.sqrt( dx * dx + dy * dy );
				dollyEnd.set( 0, distance );
				dollyDelta.set( 0, Math.pow( dollyEnd.y / dollyStart.y, scope.zoomSpeed ) );
				dollyOut( dollyDelta.y );
				dollyStart.copy( dollyEnd );

			}

			function handleTouchMoveDollyPan( event ) {

				if ( scope.enableZoom ) handleTouchMoveDolly( event );
				if ( scope.enablePan ) handleTouchMovePan( event );

			}

			function handleTouchMoveDollyRotate( event ) {

				if ( scope.enableZoom ) handleTouchMoveDolly( event );
				if ( scope.enableRotate ) handleTouchMoveRotate( event );

			}

			function handleTouchEnd( ) { // no-op
			} //
			// event handlers - FSM: listen for events and reset state
			//


			function onPointerDown( event ) {

				if ( scope.enabled === false ) return;

				switch ( event.pointerType ) {

					case 'mouse':
					case 'pen':
						onMouseDown( event );
						break;
        // TODO touch

				}

			}

			function onPointerMove( event ) {

				if ( scope.enabled === false ) return;

				switch ( event.pointerType ) {

					case 'mouse':
					case 'pen':
						onMouseMove( event );
						break;
        // TODO touch

				}

			}

			function onPointerUp( event ) {

				switch ( event.pointerType ) {

					case 'mouse':
					case 'pen':
						onMouseUp( event );
						break;
        // TODO touch

				}

			}

			function onMouseDown( event ) {

				// Prevent the browser from scrolling.
				event.preventDefault(); // Manually set the focus since calling preventDefault above
				// prevents the browser from setting it automatically.

				scope.domElement.focus ? scope.domElement.focus() : window.focus();
				let mouseAction;

				switch ( event.button ) {

					case 0:
						mouseAction = scope.mouseButtons.LEFT;
						break;

					case 1:
						mouseAction = scope.mouseButtons.MIDDLE;
						break;

					case 2:
						mouseAction = scope.mouseButtons.RIGHT;
						break;

					default:
						mouseAction = - 1;

				}

				switch ( mouseAction ) {

					case THREE.MOUSE.DOLLY:
						if ( scope.enableZoom === false ) return;
						handleMouseDownDolly( event );
						state = STATE.DOLLY;
						break;

					case THREE.MOUSE.ROTATE:
						if ( event.ctrlKey || event.metaKey || event.shiftKey ) {

							if ( scope.enablePan === false ) return;
							handleMouseDownPan( event );
							state = STATE.PAN;

						} else {

							if ( scope.enableRotate === false ) return;
							handleMouseDownRotate( event );
							state = STATE.ROTATE;

						}

						break;

					case THREE.MOUSE.PAN:
						if ( event.ctrlKey || event.metaKey || event.shiftKey ) {

							if ( scope.enableRotate === false ) return;
							handleMouseDownRotate( event );
							state = STATE.ROTATE;

						} else {

							if ( scope.enablePan === false ) return;
							handleMouseDownPan( event );
							state = STATE.PAN;

						}

						break;

					default:
						state = STATE.NONE;

				}

				if ( state !== STATE.NONE ) {

					scope.domElement.ownerDocument.addEventListener( 'pointermove', onPointerMove );
					scope.domElement.ownerDocument.addEventListener( 'pointerup', onPointerUp );
					scope.dispatchEvent( _startEvent );

				}

			}

			function onMouseMove( event ) {

				if ( scope.enabled === false ) return;
				event.preventDefault();

				switch ( state ) {

					case STATE.ROTATE:
						if ( scope.enableRotate === false ) return;
						handleMouseMoveRotate( event );
						break;

					case STATE.DOLLY:
						if ( scope.enableZoom === false ) return;
						handleMouseMoveDolly( event );
						break;

					case STATE.PAN:
						if ( scope.enablePan === false ) return;
						handleMouseMovePan( event );
						break;

				}

			}

			function onMouseUp( event ) {

				scope.domElement.ownerDocument.removeEventListener( 'pointermove', onPointerMove );
				scope.domElement.ownerDocument.removeEventListener( 'pointerup', onPointerUp );
				if ( scope.enabled === false ) return;
				handleMouseUp( event );
				scope.dispatchEvent( _endEvent );
				state = STATE.NONE;

			}

			function onMouseWheel( event ) {

				if ( scope.enabled === false || scope.enableZoom === false || state !== STATE.NONE && state !== STATE.ROTATE ) return;
				event.preventDefault();
				scope.dispatchEvent( _startEvent );
				handleMouseWheel( event );
				scope.dispatchEvent( _endEvent );

			}

			function onKeyDown( event ) {

				if ( scope.enabled === false || scope.enablePan === false ) return;
				handleKeyDown( event );

			}

			function onTouchStart( event ) {

				if ( scope.enabled === false ) return;
				event.preventDefault(); // prevent scrolling

				switch ( event.touches.length ) {

					case 1:
						switch ( scope.touches.ONE ) {

							case THREE.TOUCH.ROTATE:
								if ( scope.enableRotate === false ) return;
								handleTouchStartRotate( event );
								state = STATE.TOUCH_ROTATE;
								break;

							case THREE.TOUCH.PAN:
								if ( scope.enablePan === false ) return;
								handleTouchStartPan( event );
								state = STATE.TOUCH_PAN;
								break;

							default:
								state = STATE.NONE;

						}

						break;

					case 2:
						switch ( scope.touches.TWO ) {

							case THREE.TOUCH.DOLLY_PAN:
								if ( scope.enableZoom === false && scope.enablePan === false ) return;
								handleTouchStartDollyPan( event );
								state = STATE.TOUCH_DOLLY_PAN;
								break;

							case THREE.TOUCH.DOLLY_ROTATE:
								if ( scope.enableZoom === false && scope.enableRotate === false ) return;
								handleTouchStartDollyRotate( event );
								state = STATE.TOUCH_DOLLY_ROTATE;
								break;

							default:
								state = STATE.NONE;

						}

						break;

					default:
						state = STATE.NONE;

				}

				if ( state !== STATE.NONE ) {

					scope.dispatchEvent( _startEvent );

				}

			}

			function onTouchMove( event ) {

				if ( scope.enabled === false ) return;
				event.preventDefault(); // prevent scrolling

				switch ( state ) {

					case STATE.TOUCH_ROTATE:
						if ( scope.enableRotate === false ) return;
						handleTouchMoveRotate( event );
						scope.update();
						break;

					case STATE.TOUCH_PAN:
						if ( scope.enablePan === false ) return;
						handleTouchMovePan( event );
						scope.update();
						break;

					case STATE.TOUCH_DOLLY_PAN:
						if ( scope.enableZoom === false && scope.enablePan === false ) return;
						handleTouchMoveDollyPan( event );
						scope.update();
						break;

					case STATE.TOUCH_DOLLY_ROTATE:
						if ( scope.enableZoom === false && scope.enableRotate === false ) return;
						handleTouchMoveDollyRotate( event );
						scope.update();
						break;

					default:
						state = STATE.NONE;

				}

			}

			function onTouchEnd( event ) {

				if ( scope.enabled === false ) return;
				handleTouchEnd( event );
				scope.dispatchEvent( _endEvent );
				state = STATE.NONE;

			}

			function onContextMenu( event ) {

				if ( scope.enabled === false ) return;
				event.preventDefault();

			} //


			scope.domElement.addEventListener( 'contextmenu', onContextMenu );
			scope.domElement.addEventListener( 'pointerdown', onPointerDown );
			scope.domElement.addEventListener( 'wheel', onMouseWheel, {
				passive: false
			} );
			scope.domElement.addEventListener( 'touchstart', onTouchStart, {
				passive: false
			} );
			scope.domElement.addEventListener( 'touchend', onTouchEnd );
			scope.domElement.addEventListener( 'touchmove', onTouchMove, {
				passive: false
			} ); // force an update at start

			this.update();

		}

	} // This set of controls performs orbiting, dollying (zooming), and panning.
	// Unlike TrackballControls, it maintains the "up" direction object.up (+Y by default).
	// This is very similar to OrbitControls, another set of touch behavior
	//
	//    Orbit - right mouse, or left mouse + ctrl/meta/shiftKey / touch: two-finger rotate
	//    Zoom - middle mouse, or mousewheel / touch: two-finger spread or squish
	//    Pan - left mouse, or arrow keys / touch: one-finger move


	class MapControls extends OrbitControls {

		constructor( object, domElement ) {

			super( object, domElement );
			this.screenSpacePanning = false; // pan orthogonal to world-space direction camera.up

			this.mouseButtons.LEFT = THREE.MOUSE.PAN;
			this.mouseButtons.RIGHT = THREE.MOUSE.ROTATE;
			this.touches.ONE = THREE.TOUCH.PAN;
			this.touches.TWO = THREE.TOUCH.DOLLY_ROTATE;

		}

	}

	THREE.MapControls = MapControls;
	THREE.OrbitControls = OrbitControls;

} )();


    }

    // --- THREE.JS INITIALIZATION WITH REAL CLIPPING PLANES ---
    const container = document.getElementById('canvas-container');
    const scene = (typeof THREE !== 'undefined') ? new THREE.Scene() : null;
    if (scene) {
      scene.background = new THREE.Color(0x09090b);
      scene.fog = new THREE.FogExp2(0x09090b, 0.035);
    }

    const camera = new THREE.PerspectiveCamera(45, window.innerWidth / window.innerHeight, 0.1, 100);
    camera.position.set(0, 1.5, 9.5);

    const renderer = new THREE.WebGLRenderer({ antialias: true, powerPreference: "high-performance" });
    renderer.setSize(window.innerWidth, window.innerHeight);
    renderer.setPixelRatio(Math.min(window.devicePixelRatio, 2));
    renderer.shadowMap.enabled = true;
    renderer.shadowMap.type = THREE.PCFSoftShadowMap;
    // ENABLE TRUE PHYSICAL THREE.JS LOCAL CLIPPING
    renderer.localClippingEnabled = true;
    container.appendChild(renderer.domElement);

    let controls = null;
    if (typeof THREE.OrbitControls === 'function') {
      controls = new THREE.OrbitControls(camera, renderer.domElement);
      controls.enableDamping = true;
      controls.dampingFactor = 0.05;
      controls.maxDistance = 18;
      controls.minDistance = 3;
      controls.target.set(0, 0, 0);
    } else {
      console.warn("THREE.OrbitControls not loaded yet; fallback camera active.");
    }

    // Lighting
    const ambLight = new THREE.AmbientLight(0xffffff, 0.7);
    scene.add(ambLight);

    const dirLight1 = new THREE.DirectionalLight(0x5eead4, 0.9);
    dirLight1.position.set(6, 10, 6);
    scene.add(dirLight1);

    const dirLight2 = new THREE.DirectionalLight(0xf59e0b, 0.45);
    dirLight2.position.set(-6, -5, -6);
    scene.add(dirLight2);

    // Floor Grid
    const grid = new THREE.GridHelper(20, 40, 0x14b8a6, 0x27272a);
    grid.position.y = -4.5;
    scene.add(grid);

    // --- 3.0T CLINICAL SUPERCONDUCTING MRI SCANNER (BORE, GANTRY & PATIENT COUCH) ---
    const mriGroup = new THREE.Group();
    mriGroup.visible = false; // Kept off by default so the knee anatomy is 100% unobstructed!
    let mriScannerMode = "off"; // Cycle: 'off' -> 'ghost' -> 'solid'

    // Superconducting Magnet Cryostat Outer Drum (White Medical Enclosure)
    const gantryOuterMat = new THREE.MeshStandardMaterial({
      color: 0x0f172a, // Deep slate clinical enclosure
      roughness: 0.25,
      metalness: 0.35,
      side: THREE.DoubleSide
    });

    const gantryInnerMat = new THREE.MeshStandardMaterial({
      color: 0x1e293b,
      roughness: 0.4,
      metalness: 0.2,
      side: THREE.BackSide
    });

    // 70cm Clinical Bore Ring
    const boreOuterGeom = new THREE.CylinderGeometry(5.8, 5.8, 7.2, 48, 1, true);
    const gantryOuterMesh = new THREE.Mesh(boreOuterGeom, gantryOuterMat);
    gantryOuterMesh.rotation.x = Math.PI / 2;
    gantryOuterMesh.position.set(0, 0.4, 0);
    mriGroup.add(gantryOuterMesh);

    // Internal Bore Wall (Patient Tunnel)
    const boreInnerGeom = new THREE.CylinderGeometry(3.6, 3.6, 7.2, 48, 1, true);
    const gantryInnerMesh = new THREE.Mesh(boreInnerGeom, gantryInnerMat);
    gantryInnerMesh.rotation.x = Math.PI / 2;
    gantryInnerMesh.position.set(0, 0.4, 0);
    mriGroup.add(gantryInnerMesh);

    // Front Gantry Bezel Ring (Rounded Donut Face)
    const frontBezelGeom = new THREE.RingGeometry(3.6, 5.8, 48);
    const bezelMat = new THREE.MeshStandardMaterial({
      color: 0x334155,
      roughness: 0.3,
      metalness: 0.4,
      side: THREE.DoubleSide
    });
    const frontBezel = new THREE.Mesh(frontBezelGeom, bezelMat);
    frontBezel.position.set(0, 0.4, 3.6);
    mriGroup.add(frontBezel);

    const backBezel = new THREE.Mesh(frontBezelGeom, bezelMat);
    backBezel.position.set(0, 0.4, -3.6);
    mriGroup.add(backBezel);

    // Luminous RF Gradient LED Halo Rings (Teal / Cyan Superconducting Trim)
    const haloGeom = new THREE.TorusGeometry(3.65, 0.05, 12, 48);
    const haloMat = new THREE.MeshBasicMaterial({ color: 0x38bdf8 });
    const frontHalo = new THREE.Mesh(haloGeom, haloMat);
    frontHalo.position.set(0, 0.4, 3.61);
    mriGroup.add(frontHalo);

    const backHalo = new THREE.Mesh(haloGeom, haloMat);
    backHalo.position.set(0, 0.4, -3.61);
    mriGroup.add(backHalo);

    // Motorized Patient Couch / Imaging Table
    const couchMat = new THREE.MeshStandardMaterial({
      color: 0x18181b,
      roughness: 0.5,
      metalness: 0.2
    });
    const couchGeom = new THREE.BoxGeometry(2.4, 0.22, 14.0);
    const couchMesh = new THREE.Mesh(couchGeom, couchMat);
    couchMesh.position.set(0, -2.6, 1.5);
    mriGroup.add(couchMesh);

    // Soft Memory Foam Table Mattress Cushion
    const cushionMat = new THREE.MeshStandardMaterial({
      color: 0x27272a,
      roughness: 0.8,
      metalness: 0.05
    });
    const cushionGeom = new THREE.BoxGeometry(2.2, 0.12, 13.8);
    const cushionMesh = new THREE.Mesh(cushionGeom, cushionMat);
    cushionMesh.position.set(0, -2.43, 1.5);
    mriGroup.add(cushionMesh);

    // Dedicated 16-Channel Transmit/Receive Knee RF Coil Shell (Surrounding Knee)
    const coilMat = new THREE.MeshStandardMaterial({
      color: 0x0284c7, // Medical Blue RF Coil
      roughness: 0.35,
      metalness: 0.3,
      transparent: true,
      opacity: 0.35,
      side: THREE.DoubleSide
    });
    const coilGeom = new THREE.CylinderGeometry(2.1, 2.1, 4.4, 32, 1, true);
    const kneeCoilMesh = new THREE.Mesh(coilGeom, coilMat);
    kneeCoilMesh.rotation.x = Math.PI / 2;
    kneeCoilMesh.position.set(0, 0.3, 0);
    mriGroup.add(kneeCoilMesh);

    // RF Coil Gold Wireframe Alignment Rings
    for (let cr = -1.6; cr <= 1.6; cr += 0.8) {
      const coilRingGeom = new THREE.TorusGeometry(2.11, 0.03, 8, 32);
      const coilRingMat = new THREE.MeshBasicMaterial({ color: 0x38bdf8 });
      const crMesh = new THREE.Mesh(coilRingGeom, coilRingMat);
      crMesh.position.set(0, 0.3, cr);
      mriGroup.add(crMesh);
    }

    scene.add(mriGroup);

    // --- TRUE CROSS-SECTION CLIPPING PLANE ---
    // Normal points in the direction of the cut
    const clipPlane = new THREE.Plane(new THREE.Vector3(-1, 0, 0), 0);
    let isClippingActive = true;

    // --- PROCEDURAL ANATOMICAL MATERIALS WITH PHYSICAL CUTTING ---
    const boneMat = new THREE.MeshStandardMaterial({
      color: 0xdedede,
      roughness: 0.45,
      metalness: 0.15,
      side: THREE.DoubleSide, // Exposes the cut interior!
      clippingPlanes: [clipPlane],
      clipShadows: true
    });

    const cartilageMat = new THREE.MeshStandardMaterial({
      color: 0x99f6e4,
      roughness: 0.15,
      metalness: 0.05,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane],
      clipShadows: true
    });

    const ligamentMat = new THREE.MeshStandardMaterial({
      color: 0xf43f5e,
      roughness: 0.4,
      metalness: 0.1,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane],
      clipShadows: true
    });

    const activeHighlightMat = new THREE.MeshStandardMaterial({
      color: 0x14b8a6,
      roughness: 0.2,
      emissive: 0x14b8a6,
      emissiveIntensity: 0.7,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });

    const companionHighlightMat = new THREE.MeshStandardMaterial({
      color: 0xf59e0b,
      roughness: 0.2,
      emissive: 0xf59e0b,
      emissiveIntensity: 0.65,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });

    const kaizenRedMat = new THREE.MeshStandardMaterial({
      color: 0xf43f5e, // Andon Red: Acute Structural Disruption
      roughness: 0.2,
      emissive: 0xf43f5e,
      emissiveIntensity: 0.9,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });

    const kaizenAmberMat = new THREE.MeshStandardMaterial({
      color: 0xf59e0b, // Kaizen Amber: Secondary Strain & Degenerative Watch
      roughness: 0.2,
      emissive: 0xf59e0b,
      emissiveIntensity: 0.75,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });

    const kaizenGreenMat = new THREE.MeshStandardMaterial({
      color: 0x10b981, // Kaizen Green: Pristine Intact Stabilizer
      roughness: 0.3,
      emissive: 0x10b981,
      emissiveIntensity: 0.4,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });

    const muscleMat = new THREE.MeshStandardMaterial({
      color: 0xbe123c, // Striated crimson muscle
      roughness: 0.62,
      metalness: 0.10,
      transparent: true,
      opacity: 0.88,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane],
      clipShadows: true
    });

    const tendonMat = new THREE.MeshStandardMaterial({
      color: 0xf5f5f4, // Pearlescent fibrous tendon
      roughness: 0.35,
      metalness: 0.08,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane],
      clipShadows: true
    });

    const skinMat = new THREE.MeshStandardMaterial({
      color: 0xd8a07c, // Fitzpatrick III healthy clinical skin
      roughness: 0.42,
      metalness: 0.08,
      transparent: true,
      opacity: 0.32, // Translucent "Ghost MRI" contour default
      depthWrite: false, // Prevents occluding internal bones & muscles
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane],
      clipShadows: true
    });


    // --- ANATOMICAL GEOMETRY BUILDER ---
    const kneeGroup = new THREE.Group();
    scene.add(kneeGroup);

    // 1. Distal Femur
    const femurGroup = new THREE.Group();
    const shaftGeom = new THREE.CylinderGeometry(0.7, 0.85, 4.0, 32);
    const shaft = new THREE.Mesh(shaftGeom, boneMat);
    shaft.position.y = 2.4;
    femurGroup.add(shaft);

    const condyleGeom = new THREE.SphereGeometry(0.85, 32, 24);
    const medCondyle = new THREE.Mesh(condyleGeom, boneMat);
    medCondyle.position.set(-0.85, 0.7, -0.2);
    medCondyle.scale.set(0.9, 1.1, 1.3);
    femurGroup.add(medCondyle);

    const latCondyle = new THREE.Mesh(condyleGeom, boneMat);
    latCondyle.position.set(0.85, 0.7, -0.2);
    latCondyle.scale.set(0.85, 1.05, 1.25);
    femurGroup.add(latCondyle);

    const cartCapGeom = new THREE.SphereGeometry(0.87, 32, 16, 0, Math.PI * 2, Math.PI / 2, Math.PI / 2);
    const medFemCart = new THREE.Mesh(cartCapGeom, cartilageMat);
    medFemCart.position.set(-0.85, 0.6, -0.2);
    medFemCart.scale.set(0.92, 1.12, 1.32);
    femurGroup.add(medFemCart);

    const latFemCart = new THREE.Mesh(cartCapGeom, cartilageMat);
    latFemCart.position.set(0.85, 0.6, -0.2);
    latFemCart.scale.set(0.87, 1.07, 1.27);
    femurGroup.add(latFemCart);

    kneeGroup.add(femurGroup);

    // 2. Proximal Tibia & Fibula
    const tibiaGroup = new THREE.Group();
    const tibShaftGeom = new THREE.CylinderGeometry(0.8, 0.65, 4.0, 32);
    const tibShaft = new THREE.Mesh(tibShaftGeom, boneMat);
    tibShaft.position.y = -2.6;
    tibiaGroup.add(tibShaft);

    const plateauGeom = new THREE.CylinderGeometry(1.65, 1.35, 0.8, 32);
    const plateau = new THREE.Mesh(plateauGeom, boneMat);
    plateau.position.y = -0.55;
    tibiaGroup.add(plateau);

    const spineGeom = new THREE.ConeGeometry(0.25, 0.45, 16);
    const medSpine = new THREE.Mesh(spineGeom, boneMat);
    medSpine.position.set(-0.25, -0.05, -0.05);
    tibiaGroup.add(medSpine);

    const latSpine = new THREE.Mesh(spineGeom, boneMat);
    latSpine.position.set(0.25, -0.05, -0.05);
    tibiaGroup.add(latSpine);

    const fibShaftGeom = new THREE.CylinderGeometry(0.25, 0.2, 3.8, 16);
    const fibShaft = new THREE.Mesh(fibShaftGeom, boneMat);
    fibShaft.position.set(1.5, -2.8, -0.3);
    tibiaGroup.add(fibShaft);

    const fibHeadGeom = new THREE.SphereGeometry(0.4, 16, 12);
    const fibHead = new THREE.Mesh(fibHeadGeom, boneMat);
    fibHead.position.set(1.5, -0.9, -0.3);
    tibiaGroup.add(fibHead);

    kneeGroup.add(tibiaGroup);

    // 3. Patella
    const patellaGeom = new THREE.SphereGeometry(0.65, 24, 16);
    const patella = new THREE.Mesh(patellaGeom, boneMat);
    patella.scale.set(0.9, 1.1, 0.5);
    patella.position.set(0, 0.65, 1.6);
    kneeGroup.add(patella);

    // 4. Soft Tissue & Ligaments
    const softGroup = new THREE.Group();

    // Medial & Lateral Menisci
    const medMenGeom = new THREE.TorusGeometry(0.65, 0.22, 16, 32, Math.PI * 1.3);
    const medMeniscus = new THREE.Mesh(medMenGeom, cartilageMat);
    medMeniscus.rotation.x = Math.PI / 2;
    medMeniscus.rotation.z = Math.PI * 0.35;
    medMeniscus.position.set(-0.85, -0.15, 0);
    softGroup.add(medMeniscus);

    const latMenGeom = new THREE.TorusGeometry(0.6, 0.2, 16, 32, Math.PI * 1.55);
    const latMeniscus = new THREE.Mesh(latMenGeom, cartilageMat);
    latMeniscus.rotation.x = Math.PI / 2;
    latMeniscus.rotation.z = -Math.PI * 0.25;
    latMeniscus.position.set(0.85, -0.15, 0);
    softGroup.add(latMeniscus);

    // Cruciate Ligaments (ACL & PCL)
    const aclCurve = new THREE.CatmullRomCurve3([
      new THREE.Vector3(0.5, 0.6, -0.3),
      new THREE.Vector3(0.1, 0.2, 0.0),
      new THREE.Vector3(-0.25, -0.2, 0.35)
    ]);
    const aclGeom = new THREE.TubeGeometry(aclCurve, 20, 0.18, 12, false);
    const acl = new THREE.Mesh(aclGeom, ligamentMat);
    softGroup.add(acl);

    const pclCurve = new THREE.CatmullRomCurve3([
      new THREE.Vector3(-0.4, 0.6, 0.2),
      new THREE.Vector3(-0.1, 0.2, -0.1),
      new THREE.Vector3(0.1, -0.2, -0.45)
    ]);
    const pclGeom = new THREE.TubeGeometry(pclCurve, 20, 0.2, 12, false);
    const pcl = new THREE.Mesh(pclGeom, ligamentMat);
    softGroup.add(pcl);

    // Collaterals (MCL & LCL)
    const mclGeom = new THREE.CylinderGeometry(0.12, 0.14, 2.5, 12);
    const mcl = new THREE.Mesh(mclGeom, ligamentMat);
    mcl.position.set(-1.45, 0.2, 0);
    mcl.rotation.z = -0.08;
    softGroup.add(mcl);

    const lclGeom = new THREE.CylinderGeometry(0.09, 0.11, 2.4, 12);
    const lcl = new THREE.Mesh(lclGeom, ligamentMat);
    lcl.position.set(1.48, 0.1, -0.15);
    lcl.rotation.z = 0.1;
    softGroup.add(lcl);

    // Joint Effusion Fluid Capsule
    const effusionGeom = new THREE.SphereGeometry(1.4, 24, 16);
    const effusionMat = new THREE.MeshStandardMaterial({
      color: 0x38bdf8,
      transparent: true,
      opacity: 0.3,
      roughness: 0.1,
      clippingPlanes: [clipPlane]
    });
    const effusionMesh = new THREE.Mesh(effusionGeom, effusionMat);
    effusionMesh.position.set(0, 1.4, 0.8);
    effusionMesh.scale.set(0.9, 0.8, 0.6);
    softGroup.add(effusionMesh);

    // Baker's Cyst
    const bakerGeom = new THREE.SphereGeometry(0.65, 24, 16);
    const bakerMat = new THREE.MeshStandardMaterial({
      color: 0x38bdf8,
      transparent: true,
      opacity: 0.45,
      roughness: 0.1,
      clippingPlanes: [clipPlane]
    });
    const bakerCyst = new THREE.Mesh(bakerGeom, bakerMat);
    bakerCyst.position.set(-0.8, 0.0, -1.8);
    bakerCyst.scale.set(0.8, 1.2, 0.7);
    softGroup.add(bakerCyst);

    // Bone Contusion Beacon
    const contusionGeom = new THREE.SphereGeometry(0.5, 24, 16);
    const contusionMesh = new THREE.Mesh(contusionGeom, new THREE.MeshStandardMaterial({
      color: 0xf59e0b,
      transparent: true,
      opacity: 0.75,
      roughness: 0.3,
      clippingPlanes: [clipPlane]
    }));
    contusionMesh.position.set(0.9, -0.65, -0.45);
    softGroup.add(contusionMesh);

    kneeGroup.add(softGroup);

    // 5. Musculotendinous Units (Quadriceps, Hamstrings, Gastrocnemius, Tendons)
    const muscleGroup = new THREE.Group();

    // Quadriceps Complex (Thigh Anterior/Medial/Lateral)
    const rectusGeom = new THREE.CylinderGeometry(0.65, 0.68, 2.6, 20);
    const rectus = new THREE.Mesh(rectusGeom, muscleMat);
    rectus.position.set(0, 2.6, 0.75);
    rectus.rotation.x = -0.06;
    muscleGroup.add(rectus);

    // Vastus Medialis Oblique (VMO - teardrop medial stabilizer)
    const vmoGeom = new THREE.CylinderGeometry(0.68, 0.62, 2.2, 20);
    const vmo = new THREE.Mesh(vmoGeom, muscleMat);
    vmo.position.set(-0.95, 2.0, 0.45);
    vmo.rotation.z = 0.18;
    vmo.rotation.y = -0.15;
    vmo.scale.set(1.1, 1.0, 1.0);
    muscleGroup.add(vmo);

    // Vastus Lateralis
    const vlGeom = new THREE.CylinderGeometry(0.72, 0.68, 2.5, 20);
    const vl = new THREE.Mesh(vlGeom, muscleMat);
    vl.position.set(0.95, 2.3, 0.40);
    vl.rotation.z = -0.14;
    vl.rotation.y = 0.12;
    muscleGroup.add(vl);

    // Vastus Intermedius (deep anterior bed)
    const viGeom = new THREE.CylinderGeometry(0.60, 0.62, 2.2, 20);
    const vi = new THREE.Mesh(viGeom, muscleMat);
    vi.position.set(0, 2.3, 0.35);
    muscleGroup.add(vi);

    // Tendons: Quadriceps & Patellar Tendons
    const quadTendonGeom = new THREE.CylinderGeometry(0.52, 0.58, 0.85, 20);
    const quadTendon = new THREE.Mesh(quadTendonGeom, tendonMat);
    quadTendon.position.set(0, 1.25, 1.35);
    quadTendon.scale.set(1.2, 1.0, 0.55);
    muscleGroup.add(quadTendon);

    const patTendonGeom = new THREE.CylinderGeometry(0.26, 0.28, 1.15, 20);
    const patTendon = new THREE.Mesh(patTendonGeom, tendonMat);
    patTendon.position.set(0, -0.05, 1.48);
    patTendon.rotation.x = 0.12;
    patTendon.scale.set(1.2, 1.0, 0.6);
    muscleGroup.add(patTendon);

    // Hamstrings (Posterior Compartment)
    const bfGeom = new THREE.CylinderGeometry(0.58, 0.52, 2.6, 20);
    const bf = new THREE.Mesh(bfGeom, muscleMat);
    bf.position.set(0.85, 2.4, -0.85);
    muscleGroup.add(bf);

    const semiGeom = new THREE.CylinderGeometry(0.62, 0.55, 2.6, 20);
    const semi = new THREE.Mesh(semiGeom, muscleMat);
    semi.position.set(-0.85, 2.4, -0.85);
    muscleGroup.add(semi);

    // Calf Musculature (Gastrocnemius & Tibialis)
    const gastrocMedGeom = new THREE.CylinderGeometry(0.58, 0.66, 2.4, 20);
    const gastrocMed = new THREE.Mesh(gastrocMedGeom, muscleMat);
    gastrocMed.position.set(-0.75, -1.8, -0.75);
    gastrocMed.rotation.z = -0.08;
    muscleGroup.add(gastrocMed);

    const gastrocLatGeom = new THREE.CylinderGeometry(0.55, 0.62, 2.2, 20);
    const gastrocLat = new THREE.Mesh(gastrocLatGeom, muscleMat);
    gastrocLat.position.set(0.75, -1.8, -0.75);
    gastrocLat.rotation.z = 0.08;
    muscleGroup.add(gastrocLat);

    const tibAntGeom = new THREE.CylinderGeometry(0.48, 0.52, 2.4, 20);
    const tibAnt = new THREE.Mesh(tibAntGeom, muscleMat);
    tibAnt.position.set(0.65, -2.0, 0.70);
    tibAnt.rotation.z = 0.06;
    muscleGroup.add(tibAnt);

    kneeGroup.add(muscleGroup);

    // 6. Surface Integumentary Envelope (Skin Layer)
    const skinGroup = new THREE.Group();

    // Thigh Skin Cylinder
    const thighSkinGeom = new THREE.CylinderGeometry(1.82, 1.72, 3.0, 32);
    const thighSkin = new THREE.Mesh(thighSkinGeom, skinMat);
    thighSkin.position.set(0, 2.4, 0.05);
    skinGroup.add(thighSkin);

    // Knee Joint & Patellar Contour
    const kneeSkinGeom = new THREE.SphereGeometry(1.92, 32, 24);
    const kneeSkin = new THREE.Mesh(kneeSkinGeom, skinMat);
    kneeSkin.position.set(0, 0.45, 0.25);
    kneeSkin.scale.set(1.05, 1.15, 1.20);
    skinGroup.add(kneeSkin);

    // Proximal Calf Skin Cylinder
    const calfSkinGeom = new THREE.CylinderGeometry(1.68, 1.35, 3.6, 32);
    const calfSkin = new THREE.Mesh(calfSkinGeom, skinMat);
    calfSkin.position.set(0, -2.1, -0.05);
    skinGroup.add(calfSkin);

    kneeGroup.add(skinGroup);

    // --- 7. SURGICAL PROCEDURE & DIATOM BIO-SCAFFOLD SUITE ---
    const procedureGroup = new THREE.Group();
    procedureGroup.visible = false; // Engaged via Procedure Toggle

    // 7.0 Arthroscopic Dual 4mm Keyhole Portals (Stage 1)
    const portalGroup = new THREE.Group();
    const cannulaGeom = new THREE.CylinderGeometry(0.07, 0.07, 1.1, 16);

    // Anterolateral (AL) Viewing Portal (Diagnostic 30° Scope)
    const alMat = new THREE.MeshStandardMaterial({
      color: 0x06b6d4, // Cyan optical scope
      metalness: 0.85,
      roughness: 0.2,
      emissive: 0x0891b2,
      emissiveIntensity: 0.35,
      side: THREE.DoubleSide
    });
    const alCannula = new THREE.Mesh(cannulaGeom, alMat);
    alCannula.position.set(0.65, -0.15, 1.35);
    alCannula.rotation.set(-0.55, 0.45, 0.3);
    portalGroup.add(alCannula);

    const alRingGeom = new THREE.TorusGeometry(0.12, 0.03, 8, 20);
    const alRingMat = new THREE.MeshBasicMaterial({ color: 0x22d3ee });
    const alRing = new THREE.Mesh(alRingGeom, alRingMat);
    alRing.position.set(0.65, -0.15, 1.85);
    portalGroup.add(alRing);

    // Anteromedial (AM) Working Portal (Motorized Shaver / Radiofrequency Wand)
    const amMat = new THREE.MeshStandardMaterial({
      color: 0xf59e0b, // Amber working cannula
      metalness: 0.85,
      roughness: 0.2,
      emissive: 0xd97706,
      emissiveIntensity: 0.35,
      side: THREE.DoubleSide
    });
    const amCannula = new THREE.Mesh(cannulaGeom, amMat);
    amCannula.position.set(-0.65, -0.15, 1.35);
    amCannula.rotation.set(-0.55, -0.45, -0.3);
    portalGroup.add(amCannula);

    const amRingGeom = new THREE.TorusGeometry(0.12, 0.03, 8, 20);
    const amRingMat = new THREE.MeshBasicMaterial({ color: 0xfbbf24 });
    const amRing = new THREE.Mesh(amRingGeom, amRingMat);
    amRing.position.set(-0.65, -0.15, 1.85);
    portalGroup.add(amRing);

    procedureGroup.add(portalGroup);

    // 7.1 Tibial Bone Tunnel (Anteromedial cortex into intercondylar footprint)
    const tibTunnelGeom = new THREE.CylinderGeometry(0.18, 0.18, 3.2, 16);
    const tunnelGuideMat = new THREE.MeshStandardMaterial({
      color: 0x06b6d4, // Cyan surgical guide
      roughness: 0.25,
      metalness: 0.3,
      transparent: true,
      opacity: 0.72,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });
    const tibTunnel = new THREE.Mesh(tibTunnelGeom, tunnelGuideMat);
    tibTunnel.position.set(-0.25, -1.6, 0.35);
    tibTunnel.rotation.set(-0.35, 0, 0.22);
    procedureGroup.add(tibTunnel);

    // 7.2 Femoral Socket (Intercondylar notch lateral wall, 10:30 o'clock position)
    const femTunnelGeom = new THREE.CylinderGeometry(0.19, 0.19, 2.0, 16);
    const femTunnel = new THREE.Mesh(femTunnelGeom, tunnelGuideMat);
    femTunnel.position.set(0.35, 0.95, -0.25);
    femTunnel.rotation.set(0.55, 0.3, -0.45);
    procedureGroup.add(femTunnel);

    // 7.3 Reconstructed Quadrupled Tendon Autograft
    const graftCurve = new THREE.CatmullRomCurve3([
      new THREE.Vector3(-0.45, -3.0, 0.55),  // Tibial entrance
      new THREE.Vector3(-0.25, -1.6, 0.35),  // Mid-tibial tunnel
      new THREE.Vector3(-0.05, 0.05, 0.12),  // Intra-articular joint space
      new THREE.Vector3(0.20, 0.65, -0.15),  // Femoral socket entry
      new THREE.Vector3(0.55, 1.45, -0.38)   // Lateral femoral cortex exit
    ]);
    const graftGeom = new THREE.TubeGeometry(graftCurve, 32, 0.14, 12, false);
    const graftMat = new THREE.MeshStandardMaterial({
      color: 0xfef08a, // Braided collagen autograft
      roughness: 0.35,
      metalness: 0.15,
      emissive: 0xca8a04,
      emissiveIntensity: 0.45,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });
    const reconstructedGraft = new THREE.Mesh(graftGeom, graftMat);
    procedureGroup.add(reconstructedGraft);

    // 7.4 Diatom Biosilica Interference Anchor (Porous Micro-CT Biomimetic Plug)
    const diatomAnchorGroup = new THREE.Group();
    const anchorCoreGeom = new THREE.CylinderGeometry(0.20, 0.18, 1.1, 16);
    const diatomMat = new THREE.MeshStandardMaterial({
      color: 0x14b8a6, // Diatomaceous biosilica teal
      roughness: 0.2,
      metalness: 0.4,
      emissive: 0x14b8a6,
      emissiveIntensity: 0.65,
      side: THREE.DoubleSide,
      clippingPlanes: [clipPlane]
    });
    const anchorCore = new THREE.Mesh(anchorCoreGeom, diatomMat);
    diatomAnchorGroup.add(anchorCore);

    // Radial micro-pores on Diatom Anchor
    const ringCount = 4;
    for (let r = 0; r < ringCount; r++) {
      const ringGeom = new THREE.TorusGeometry(0.21, 0.025, 8, 24);
      const ringMat = new THREE.MeshBasicMaterial({ color: 0x5eead4, wireframe: true });
      const ring = new THREE.Mesh(ringGeom, ringMat);
      ring.rotation.x = Math.PI / 2;
      ring.position.y = -0.4 + r * 0.26;
      diatomAnchorGroup.add(ring);
    }
    const defaultAnchorPos = new THREE.Vector3(-0.35, -2.4, 0.48);
    diatomAnchorGroup.position.copy(defaultAnchorPos);
    diatomAnchorGroup.rotation.set(-0.35, 0, 0.22);
    procedureGroup.add(diatomAnchorGroup);

    // 7.5 Femoral Suspensory Cortical Titanium Button
    const buttonGeom = new THREE.BoxGeometry(0.35, 0.10, 0.16);
    const titaniumMat = new THREE.MeshStandardMaterial({
      color: 0xe2e8f0,
      metalness: 0.9,
      roughness: 0.2,
      side: THREE.DoubleSide
    });
    const corticalButton = new THREE.Mesh(buttonGeom, titaniumMat);
    const defaultButtonPos = new THREE.Vector3(0.60, 1.55, -0.42);
    corticalButton.position.copy(defaultButtonPos);
    corticalButton.rotation.set(0.4, 0.2, 0.6);
    procedureGroup.add(corticalButton);

    // 7.6 Exploded Hardware Dotted Trajectory Guides
    const guideGeom = new THREE.BufferGeometry().setFromPoints([
      new THREE.Vector3(0.60, 1.55, -0.42),
      new THREE.Vector3(1.10, 2.50, -0.75),
      new THREE.Vector3(-0.35, -2.4, 0.48),
      new THREE.Vector3(-0.70, -3.85, 0.85)
    ]);
    const guideMat = new THREE.LineDashedMaterial({
      color: 0x38bdf8,
      dashSize: 0.15,
      gapSize: 0.1,
      linewidth: 2
    });
    const explodedGuides = new THREE.LineSegments(guideGeom, guideMat);
    explodedGuides.computeLineDistances();
    explodedGuides.visible = false;
    procedureGroup.add(explodedGuides);

    kneeGroup.add(procedureGroup);

    // --- VISUAL LASER CUTTING GUIDE PLANE ---
    const planeGeom = new THREE.PlaneGeometry(6.5, 6.5);
    const planeMat = new THREE.MeshBasicMaterial({
      color: 0x14b8a6,
      transparent: true,
      opacity: 0.22,
      side: THREE.DoubleSide,
      depthWrite: false
    });
    const slicePlane = new THREE.Mesh(planeGeom, planeMat);
    scene.add(slicePlane);

    const edges = new THREE.EdgesGeometry(planeGeom);
    const lineMat = new THREE.LineBasicMaterial({ color: 0x2dd4bf, linewidth: 2 });
    const planeWire = new THREE.LineSegments(edges, lineMat);
    slicePlane.add(planeWire);

    // --- SLICING STATE & MODES ---
    let activePlaneMode = "sagittal"; // 'sagittal', 'coronal', 'axial'
    let currentDepth = 0.50;
    let isAutoScanning = false;
    let scanDirection = 1;

    const depthSlider = document.getElementById('depth-slider');
    const depthValLabel = document.getElementById('depth-val');
    const depthLoLabel = document.getElementById('depth-lo-label');
    const depthHiLabel = document.getElementById('depth-hi-label');
    const headerPlaneBadge = document.getElementById('header-plane-badge');
    const headerStructBadge = document.getElementById('header-structure-badge');

    function updateSlicingPlane() {
      let posX = 0, posY = 0, posZ = 0;

      if (activePlaneMode === "sagittal") {
        slicePlane.rotation.set(0, Math.PI / 2, 0);
        posX = THREE.MathUtils.lerp(-2.2, 2.2, currentDepth);
        slicePlane.position.set(posX, 0.4, 0);
        
        // Update Three.js Physical Clipping Plane
        if (isClippingActive) {
          clipPlane.normal.set(-1, 0, 0);
          clipPlane.constant = posX;
        }

        depthValLabel.innerText = `${currentDepth.toFixed(2)}`;
        depthLoLabel.innerText = "Medial (0.0)";
        depthHiLabel.innerText = "Lateral (1.0)";
        headerPlaneBadge.innerText = "SAGITTAL";

        const docTitle = document.getElementById('dock-doc-title');
        const docDesc = document.getElementById('dock-doc-desc');
        const docPlane = document.getElementById('dock-doc-plane');

        if (currentDepth < 0.35) {
          headerStructBadge.innerText = "Medial Meniscus, MCL";
          if (docTitle) docTitle.innerText = "Medial Meniscus & MCL Joint Line";
          if (docDesc) docDesc.innerText = "Crucial zone for Medial Meniscus tears and medial compartment osteoarthritis (joint space narrowing).";
        } else if (currentDepth > 0.65) {
          headerStructBadge.innerText = "Lateral Meniscus, LCL, Popliteus";
          if (docTitle) docTitle.innerText = "Lateral Meniscus & Posterior Corner";
          if (docDesc) docDesc.innerText = "Evaluates lateral meniscal tears, posterolateral corner injuries, and occult lateral tibial plateau bone contusions.";
        } else {
          headerStructBadge.innerText = "Intercondylar Notch, ACL, PCL";
          if (docTitle) docTitle.innerText = "Intercondylar Notch, ACL, PCL, Tibial Spines";
          if (docDesc) docDesc.innerText = "Prime diagnostic depth for ACL disruption, Blumensaat line alignment, and posterior cruciate ligament tension.";
        }
        if (docPlane) docPlane.innerText = `SAGITTAL ${currentDepth.toFixed(2)}`;
        const docHeur = document.getElementById('dock-doc-heuristic');
        const docKz = document.getElementById('dock-doc-kaizen-badge');
        if (docHeur && !docHeur.innerText) docHeur.innerText = "Blumensaat Line & Pivot-Shift Rule";
        if (docKz && !docKz.innerText) docKz.innerText = "🔴 ANDON RED | ACUTE TEAR";

      } else if (activePlaneMode === "coronal") {
        slicePlane.rotation.set(0, 0, 0);
        posZ = THREE.MathUtils.lerp(1.8, -1.8, currentDepth);
        slicePlane.position.set(0, 0.4, posZ);

        if (isClippingActive) {
          clipPlane.normal.set(0, 0, 1);
          clipPlane.constant = -posZ;
        }

        depthValLabel.innerText = `${currentDepth.toFixed(2)}`;
        depthLoLabel.innerText = "Anterior (0.0)";
        depthHiLabel.innerText = "Posterior (1.0)";
        headerPlaneBadge.innerText = "CORONAL";

        const docTitle = document.getElementById('dock-doc-title');
        const docDesc = document.getElementById('dock-doc-desc');
        const docPlane = document.getElementById('dock-doc-plane');

        if (currentDepth < 0.35) {
          headerStructBadge.innerText = "Patella & Hoffa's Fat Pad";
          if (docTitle) docTitle.innerText = "Patellar Tendon & Anterior Meniscal Horns";
          if (docDesc) docDesc.innerText = "Evaluates patellar tracking, anterior synovitis, and anterior horn meniscal detachment.";
        } else if (currentDepth > 0.65) {
          headerStructBadge.innerText = "Popliteal Fossa & Baker's Cyst";
          if (docTitle) docTitle.innerText = "Popliteal Fossa, Baker's Cyst, Gastrocnemius";
          if (docDesc) docDesc.innerText = "Optimal view for posterior horn root tears, PCL insertion, and Baker's cyst distension.";
        } else {
          headerStructBadge.innerText = "Femoral Condyles & Tibial Plateaus";
          if (docTitle) docTitle.innerText = "Femoral Condyles, Tibial Plateaus, Collaterals";
          if (docDesc) docDesc.innerText = "Standard coronal assessment of joint space symmetry (OA grading) and collateral ligament integrity.";
        }
        if (docPlane) docPlane.innerText = `CORONAL ${currentDepth.toFixed(2)}`;

      } else if (activePlaneMode === "axial") {
        slicePlane.rotation.set(Math.PI / 2, 0, 0);
        posY = THREE.MathUtils.lerp(2.8, -2.2, currentDepth);
        slicePlane.position.set(0, posY, 0);

        if (isClippingActive) {
          clipPlane.normal.set(0, -1, 0);
          clipPlane.constant = posY;
        }

        depthValLabel.innerText = `${currentDepth.toFixed(2)}`;
        depthLoLabel.innerText = "Proximal (0.0)";
        depthHiLabel.innerText = "Distal (1.0)";
        headerPlaneBadge.innerText = "AXIAL";

        const docTitle = document.getElementById('dock-doc-title');
        const docDesc = document.getElementById('dock-doc-desc');
        const docPlane = document.getElementById('dock-doc-plane');

        if (currentDepth < 0.35) {
          headerStructBadge.innerText = "Suprapatellar Pouch & Trochlea";
          if (docTitle) docTitle.innerText = "Suprapatellar Pouch, Quadriceps Tendon, Trochlea";
          if (docDesc) docDesc.innerText = "Gold standard for joint effusion volume quantification and trochlear dysplasia assessment.";
        } else if (currentDepth > 0.65) {
          headerStructBadge.innerText = "Proximal Tibia & Fibular Head";
          if (docTitle) docTitle.innerText = "Proximal Tibia, Proximal Fibula, Tibial Tuberosity";
          if (docDesc) docDesc.innerText = "Evaluates tibial plateau subchondral fractures and patellar tendon insertion.";
        } else {
          headerStructBadge.innerText = "Patellofemoral Joint & Menisci";
          if (docTitle) docTitle.innerText = "Patellofemoral Joint, Patellar Facets, Meniscal Rims";
          if (docDesc) docDesc.innerText = "Patellofemoral osteoarthritis, cartilage thinning, and subluxation dynamics.";
        }
        if (docPlane) docPlane.innerText = `AXIAL ${currentDepth.toFixed(2)}`;
      }

      
      // Update Patient Consult Text during manual slider scrubbing
      if (isPatientMode) {
        const patTitle = document.getElementById('dock-pat-title');
        const patWhy = document.getElementById('dock-pat-why');
        const patSens = document.getElementById('dock-pat-sens');
        const patRec = document.getElementById('dock-pat-rec');

        if (activePlaneMode === "sagittal") {
          if (currentDepth < 0.35) {
            if (patTitle) patTitle.innerText = "Inner Shock Absorber & Ligament (Medial Zone)";
            if (patWhy) patWhy.innerText = "Viewing the inner curve of your knee where your C-shaped inner cartilage cushion absorbs impact with every step.";
            if (patSens) patSens.innerText = "Common area for tenderness when pressing on the inside joint line or walking on uneven ground.";
            if (patRec) patRec.innerText = "Act I: Low-impact unweighted cycling. Act II: Gentle quad activation to lubricate this inner compartment.";
          } else if (currentDepth > 0.65) {
            if (patTitle) patTitle.innerText = "Outer Shock Absorber & Stabilizer (Lateral Zone)";
            if (patWhy) patWhy.innerText = "Viewing the outer edge of your knee joint where outer cushion and tendons keep the knee aligned during turns.";
            if (patSens) patSens.innerText = "Can ache with sharp pinching or catching when twisting or descending stairs.";
            if (patRec) patRec.innerText = "Act I: RICE protocol. Act II: Foam rolling the outer IT band and glute strengthening.";
          } else {
            if (patTitle) patTitle.innerText = "Central Knee Stabilizers (ACL & PCL Core)";
            if (patWhy) patWhy.innerText = "Viewing the center notch where your main front-to-back stabilizer straps cross inside the joint.";
            if (patSens) patSens.innerText = "Injuries here create a feeling that your knee wants to slide forward or 'give way' under load.";
            if (patRec) patRec.innerText = "Act I: Protective bracing. Act II: Hamstring strengthening to serve as an internal seatbelt.";
          }
        } else if (activePlaneMode === "coronal") {
          if (currentDepth < 0.35) {
            if (patTitle) patTitle.innerText = "Front Kneecap & Tendon Alignment";
            if (patWhy) patWhy.innerText = "Viewing the front kneecap and protective fat pad that cushions your joint during bending.";
            if (patSens) patSens.innerText = "Can ache during deep squats, kneeling, or rising from a low chair.";
            if (patRec) patRec.innerText = "Strengthening inner quad (VMO) muscle to ensure smooth, centered kneecap gliding.";
          } else if (currentDepth > 0.65) {
            if (patTitle) patTitle.innerText = "Back of the Knee & Fluid Relief Zone";
            if (patWhy) patWhy.innerText = "Viewing the back of the knee where tendons connect to calf muscles and fluid relief pockets can form.";
            if (patSens) patSens.innerText = "Feels tight, full, or stiff when straightening your knee completely.";
            if (patRec) patRec.innerText = "Gentle calf stretching and managing front-of-knee swelling so back fluid reabsorbs.";
          } else {
            if (patTitle) patTitle.innerText = "Joint Space & Cushion Symmetry";
            if (patWhy) patWhy.innerText = "Viewing the main weight-bearing cartilage surfaces of your thigh bone meeting your shin bone.";
            if (patSens) patSens.innerText = "Stiffness in the morning that warms up with movement; aching after prolonged standing.";
            if (patRec) patRec.innerText = "Low-impact swimming, supportive footwear, and maintaining strong leg muscles.";
          }
        } else if (activePlaneMode === "axial") {
          if (currentDepth < 0.35) {
            if (patTitle) patTitle.innerText = "Upper Fluid Reservoir (Suprapatellar Pouch)";
            if (patWhy) patWhy.innerText = "Viewing the space above your kneecap where your body stores fluid when the knee is irritated.";
            if (patSens) patSens.innerText = "Feels like a tight, pressurized 'balloon' above your kneecap.";
            if (patRec) patRec.innerText = "Elevation, ice packs, and gentle ankle pumps to help drain fluid back into circulation.";
          } else if (currentDepth > 0.65) {
            if (patTitle) patTitle.innerText = "Shin Bone Shelf (Tibial Plateau)";
            if (patWhy) patWhy.innerText = "Viewing the flat bone plateau of your shin where your entire body weight lands.";
            if (patSens) patSens.innerText = "Sharp bone tenderness if bruised or cracked during high-impact landings.";
            if (patRec) patRec.innerText = "Unweighting with crutches or brace until bone micro-trabeculae fully remodel.";
          } else {
            if (patTitle) patTitle.innerText = "Kneecap Gliding Groove (Trochlea)";
            if (patWhy) patWhy.innerText = "Viewing how your kneecap fits into the V-shaped groove of your thigh bone.";
            if (patSens) patSens.innerText = "Grinding or clicking sounds when bending and straightening.";
            if (patRec) patRec.innerText = "Quadriceps balanced strengthening to keep the kneecap tracking smoothly.";
          }
        }
      }

      requestRender();
      if (typeof renderDiatomSlice === 'function') {
        renderDiatomSlice(currentDepth, activePlaneMode, isDiatomMicroMode);
      }
    }

    // SLIDER EVENT LISTENER
    depthSlider.addEventListener('input', (e) => {
      currentDepth = parseFloat(e.target.value);
      updateSlicingPlane();
    });

    // PLANE BUTTONS
    document.getElementById('btn-plane-sagittal').addEventListener('click', () => {
      activePlaneMode = "sagittal";
      highlightPlaneBtn('btn-plane-sagittal');
      updateSlicingPlane();
    });

    document.getElementById('btn-plane-coronal').addEventListener('click', () => {
      activePlaneMode = "coronal";
      highlightPlaneBtn('btn-plane-coronal');
      updateSlicingPlane();
    });

    document.getElementById('btn-plane-axial').addEventListener('click', () => {
      activePlaneMode = "axial";
      highlightPlaneBtn('btn-plane-axial');
      updateSlicingPlane();
    });

    function highlightPlaneBtn(id) {
      ['btn-plane-sagittal', 'btn-plane-coronal', 'btn-plane-axial'].forEach(b => {
        const el = document.getElementById(b);
        if (b === id) {
          el.className = "px-3 py-1.5 rounded-xl bg-teal-600 text-white font-bold transition";
        } else {
          el.className = "px-3 py-1.5 rounded-xl text-zinc-400 hover:text-white transition";
        }
      });
    }

    // AUTO SCAN (CINE PLAY/PAUSE)
    const btnAutoScan = document.getElementById('btn-auto-scan');
    const scanIcon = document.getElementById('scan-icon');
    const scanLabel = document.getElementById('scan-label');

    btnAutoScan.addEventListener('click', () => {
      isAutoScanning = !isAutoScanning;
      if (isAutoScanning) {
        scanIcon.innerText = "⏸";
        scanLabel.innerText = "Scanning...";
        btnAutoScan.className = "glass-panel px-3 py-1.5 rounded-2xl bg-amber-600 hover:bg-amber-500 text-white text-xs font-bold transition flex items-center gap-1.5 shadow-lg border border-amber-400";
        startAutoScan();
      } else {
        scanIcon.innerText = "▶";
        scanLabel.innerText = "Auto-Scan";
        btnAutoScan.className = "glass-panel px-3 py-1.5 rounded-2xl bg-teal-600 hover:bg-teal-500 text-white text-xs font-bold transition flex items-center gap-1.5 shadow-lg border border-teal-400/50";
      }
    });

    function startAutoScan() {
      if (!isAutoScanning) return;
      currentDepth += 0.005 * scanDirection;
      if (currentDepth >= 1.0) {
        currentDepth = 1.0;
        scanDirection = -1;
      } else if (currentDepth <= 0.0) {
        currentDepth = 0.0;
        scanDirection = 1;
      }
      depthSlider.value = currentDepth;
      updateSlicingPlane();
      requestAnimationFrame(startAutoScan);
    }

    // CLIPPING CUT TOGGLE BUTTON
    const btnClipToggle = document.getElementById('btn-clip-toggle');
    const clipStatus = document.getElementById('clip-status');
    btnClipToggle.addEventListener('click', () => {
      isClippingActive = !isClippingActive;
      renderer.localClippingEnabled = isClippingActive;
      clipStatus.innerText = isClippingActive ? "ON" : "OFF";
      clipStatus.className = isClippingActive ? "text-white font-bold" : "text-zinc-500 font-bold";
      updateSlicingPlane();
    });

    // --- ANATOMICAL LAYER CONTROLS (BONE / MUSCLE / SKIN) ---
    // Tri-State Lifecycles: "solid" -> "ghost" -> "off"
    let boneMode = "solid";   // "solid", "ghost", "off"
    let muscleMode = "solid"; // "solid", "ghost", "off"
    let skinMode = "ghost";   // "ghost", "solid", "off"

    function getGhostBoneOpacity() {
      return (typeof currentCircadianTone !== 'undefined' && currentCircadianTone === 'night') ? 0.18 : ((typeof currentCircadianTone !== 'undefined' && currentCircadianTone === 'dusk') ? 0.22 : 0.26);
    }
    function getGhostMuscleOpacity() {
      return (typeof currentCircadianTone !== 'undefined' && currentCircadianTone === 'night') ? 0.16 : ((typeof currentCircadianTone !== 'undefined' && currentCircadianTone === 'dusk') ? 0.19 : 0.22);
    }

    const btnLayerBone = document.getElementById('btn-layer-bone');
    const boneModeLabel = document.getElementById('bone-mode-label');
    const btnLayerMuscle = document.getElementById('btn-layer-muscle');
    const muscleModeLabel = document.getElementById('muscle-mode-label');
    const btnLayerSkin = document.getElementById('btn-layer-skin');
    const skinModeLabel = document.getElementById('skin-mode-label');

    btnLayerBone.addEventListener('click', () => {
      if (boneMode === "solid") {
        boneMode = "ghost";
        femurGroup.visible = true;
        tibiaGroup.visible = true;
        patella.visible = true;
        boneMat.transparent = true;
        boneMat.opacity = getGhostBoneOpacity();
        boneMat.depthWrite = false;
        cartilageMat.transparent = true;
        cartilageMat.opacity = 0.45;
        cartilageMat.depthWrite = false;
        if (boneModeLabel) boneModeLabel.innerText = "Ghost";
        btnLayerBone.className = "px-2.5 py-1 rounded-xl bg-teal-900/80 text-teal-200 border border-teal-400 font-bold transition text-[11px] shadow-[0_0_12px_rgba(20,184,166,0.4)]";
      } else if (boneMode === "ghost") {
        boneMode = "off";
        femurGroup.visible = false;
        tibiaGroup.visible = false;
        patella.visible = false;
        if (boneModeLabel) boneModeLabel.innerText = "OFF";
        btnLayerBone.className = "px-2.5 py-1 rounded-xl bg-zinc-800 text-zinc-400 font-bold transition text-[11px]";
      } else {
        boneMode = "solid";
        femurGroup.visible = true;
        tibiaGroup.visible = true;
        patella.visible = true;
        boneMat.transparent = false;
        boneMat.opacity = 1.0;
        boneMat.depthWrite = true;
        cartilageMat.transparent = false;
        cartilageMat.opacity = 1.0;
        cartilageMat.depthWrite = true;
        if (boneModeLabel) boneModeLabel.innerText = "Solid";
        btnLayerBone.className = "px-2.5 py-1 rounded-xl bg-teal-600 text-white font-bold transition text-[11px]";
      }
      requestRender();
    });

    btnLayerMuscle.addEventListener('click', () => {
      if (muscleMode === "solid") {
        muscleMode = "ghost";
        muscleGroup.visible = true;
        muscleMat.transparent = true;
        muscleMat.opacity = getGhostMuscleOpacity();
        muscleMat.depthWrite = false;
        tendonMat.transparent = true;
        tendonMat.opacity = 0.35;
        tendonMat.depthWrite = false;
        if (muscleModeLabel) muscleModeLabel.innerText = "Ghost";
        btnLayerMuscle.className = "px-2.5 py-1 rounded-xl bg-rose-900/80 text-rose-200 border border-rose-400 font-bold transition text-[11px] shadow-[0_0_12px_rgba(244,63,94,0.4)]";
      } else if (muscleMode === "ghost") {
        muscleMode = "off";
        muscleGroup.visible = false;
        if (muscleModeLabel) muscleModeLabel.innerText = "OFF";
        btnLayerMuscle.className = "px-2.5 py-1 rounded-xl bg-zinc-800 text-zinc-400 font-bold transition text-[11px]";
      } else {
        muscleMode = "solid";
        muscleGroup.visible = true;
        muscleMat.transparent = false;
        muscleMat.opacity = 0.90;
        muscleMat.depthWrite = true;
        tendonMat.transparent = false;
        tendonMat.opacity = 1.0;
        tendonMat.depthWrite = true;
        if (muscleModeLabel) muscleModeLabel.innerText = "Solid";
        btnLayerMuscle.className = "px-2.5 py-1 rounded-xl bg-rose-600 text-white font-bold transition text-[11px]";
      }
      requestRender();
    });

    const btnLayerMri = document.getElementById('btn-layer-mri');
    const mriModeLabel = document.getElementById('mri-mode-label');

    if (btnLayerMri) {
      btnLayerMri.addEventListener('click', () => {
        if (mriScannerMode === "off") {
          // Switch to Ghost (translucent contour ring)
          mriScannerMode = "ghost";
          mriGroup.visible = true;
          gantryOuterMat.transparent = true;
          gantryOuterMat.opacity = 0.14;
          gantryOuterMat.depthWrite = false;
          gantryInnerMat.transparent = true;
          gantryInnerMat.opacity = 0.10;
          gantryInnerMat.depthWrite = false;
          bezelMat.transparent = true;
          bezelMat.opacity = 0.15;
          bezelMat.depthWrite = false;
          couchMat.transparent = true;
          couchMat.opacity = 0.20;
          cushionMat.transparent = true;
          cushionMat.opacity = 0.20;
          coilMat.opacity = 0.15;
          if (mriModeLabel) mriModeLabel.innerText = "Ghost";
          btnLayerMri.className = "px-3 py-1 rounded-full bg-sky-900/80 text-sky-200 border border-sky-400 font-bold transition text-[11px] shadow-[0_0_12px_rgba(56,189,248,0.4)]";
        } else if (mriScannerMode === "ghost") {
          // Switch to Solid Gantry
          mriScannerMode = "solid";
          mriGroup.visible = true;
          gantryOuterMat.transparent = false;
          gantryOuterMat.opacity = 1.0;
          gantryOuterMat.depthWrite = true;
          gantryInnerMat.transparent = false;
          gantryInnerMat.opacity = 1.0;
          gantryInnerMat.depthWrite = true;
          bezelMat.transparent = false;
          bezelMat.opacity = 1.0;
          bezelMat.depthWrite = true;
          couchMat.transparent = false;
          couchMat.opacity = 1.0;
          cushionMat.transparent = false;
          cushionMat.opacity = 1.0;
          coilMat.opacity = 0.40;
          if (mriModeLabel) mriModeLabel.innerText = "Solid";
          btnLayerMri.className = "px-3 py-1 rounded-full bg-sky-600 text-white font-bold transition text-[11px]";
        } else {
          // Turn OFF
          mriScannerMode = "off";
          mriGroup.visible = false;
          if (mriModeLabel) mriModeLabel.innerText = "OFF";
          btnLayerMri.className = "px-3 py-1 rounded-full bg-zinc-800 text-zinc-400 font-bold transition text-[11px]";
        }
        requestRender();
      });
    }

    btnLayerSkin.addEventListener('click', () => {
      if (skinMode === "ghost") {
        skinMode = "solid";
        skinGroup.visible = true;
        skinMat.opacity = 0.88;
        skinMat.depthWrite = true;
        if (skinModeLabel) skinModeLabel.innerText = "Solid";
        btnLayerSkin.className = "px-2.5 py-1 rounded-xl bg-amber-500 text-zinc-950 font-bold transition text-[11px]";
      } else if (skinMode === "solid") {
        skinMode = "off";
        skinGroup.visible = false;
        if (skinModeLabel) skinModeLabel.innerText = "OFF";
        btnLayerSkin.className = "px-2.5 py-1 rounded-xl bg-zinc-800 text-zinc-400 font-bold transition text-[11px]";
      } else {
        skinMode = "ghost";
        skinGroup.visible = true;
        skinMat.opacity = 0.32;
        skinMat.depthWrite = false;
        if (skinModeLabel) skinModeLabel.innerText = "Ghost";
        btnLayerSkin.className = "px-2.5 py-1 rounded-xl bg-amber-600/80 text-white font-bold transition text-[11px]";
      }
      requestRender();
    });

    // --- 🔬 SURGICAL PROCEDURE & DIATOM BIO-SCAFFOLD CONTROLS ---
    let isProcedureActive = false;
    let currentProcedureStage = 1;
    let isExploded = false;
    let hardwareAnimFrame = null;

    const btnProcedureToggle = document.getElementById('btn-procedure-toggle');
    const procedureStatus = document.getElementById('procedure-status');
    const procStepperPanel = document.getElementById('procedure-stepper-panel');
    const hardwareCallouts = document.getElementById('hardware-callouts');
    const btnExplodedView = document.getElementById('btn-exploded-view');
    const explodedViewLabel = document.getElementById('exploded-view-label');
    const btnProcNext = document.getElementById('btn-proc-next');

    function setHardwareExploded(targetExploded) {
      isExploded = targetExploded;
      if (btnExplodedView) {
        if (isExploded) {
          btnExplodedView.className = "flex-1 px-2 py-1 rounded-xl bg-amber-600 hover:bg-amber-500 text-white border border-amber-400 text-[10px] font-bold transition flex items-center justify-center gap-1 shadow-[0_0_12px_rgba(245,158,11,0.5)]";
          if (explodedViewLabel) explodedViewLabel.innerText = "Collapse";
        } else {
          btnExplodedView.className = "flex-1 px-2 py-1 rounded-xl bg-zinc-900 hover:bg-zinc-800 text-amber-300 border border-amber-500/40 text-[10px] font-bold transition flex items-center justify-center gap-1";
          if (explodedViewLabel) explodedViewLabel.innerText = "Exploded View";
        }
      }

      const startBtnPos = corticalButton.position.clone();
      const endBtnPos = isExploded
        ? new THREE.Vector3(1.10, 2.50, -0.75)
        : defaultButtonPos.clone();

      const startAnchorPos = diatomAnchorGroup.position.clone();
      const endAnchorPos = isExploded
        ? new THREE.Vector3(-0.70, -3.85, 0.85)
        : defaultAnchorPos.clone();

      if (explodedGuides) explodedGuides.visible = isExploded && currentProcedureStage === 4;

      let progress = 0;
      if (hardwareAnimFrame) cancelAnimationFrame(hardwareAnimFrame);

      function stepHardware() {
        progress += 0.08;
        corticalButton.position.lerpVectors(startBtnPos, endBtnPos, Math.min(progress, 1.0));
        diatomAnchorGroup.position.lerpVectors(startAnchorPos, endAnchorPos, Math.min(progress, 1.0));
        render();
        if (progress < 1.0) {
          hardwareAnimFrame = requestAnimationFrame(stepHardware);
        }
      }
      stepHardware();
    }

    function setProcedureStage(stage) {
      currentProcedureStage = stage;

      const stageIndicator = document.getElementById('procedure-stage-indicator');
      if (stageIndicator) stageIndicator.innerText = `STAGE ${stage}/4`;

      // Update Stage Buttons styling
      for (let s = 1; s <= 4; s++) {
        const b = document.getElementById(`btn-proc-stage-${s}`);
        if (b) {
          if (s === stage) {
            b.className = "proc-stage-btn px-2 py-1.5 rounded-xl bg-teal-600 text-white font-bold transition text-left flex items-center gap-1 border border-teal-400 shadow-[0_0_10px_rgba(20,184,166,0.3)]";
          } else {
            b.className = "proc-stage-btn px-2 py-1.5 rounded-xl bg-zinc-900 text-zinc-400 hover:text-white hover:bg-zinc-800 transition text-left flex items-center gap-1 border border-zinc-800";
          }
        }
      }

      const docTitle = document.getElementById('dock-doc-title');
      const docDesc = document.getElementById('dock-doc-desc');
      const docTriad = document.getElementById('dock-doc-triad');
      const docKaizenBadge = document.getElementById('dock-doc-kaizen-badge');
      const radlexText = document.getElementById('dock-radlex-text');

      if (stage === 1) {
        // Stage 1: Dual 4mm Keyhole Portals & Arthroscopic Debridement
        portalGroup.visible = true;
        tibTunnel.visible = false;
        femTunnel.visible = false;
        reconstructedGraft.visible = false;
        diatomAnchorGroup.visible = false;
        corticalButton.visible = false;
        acl.visible = true; // Native torn ACL visualized
        if (explodedGuides) explodedGuides.visible = false;
        setHardwareExploded(false);

        if (docTitle) docTitle.innerText = "Stage 1: Dual 4mm Arthroscopic Keyhole Portals & Debridement";
        if (docDesc) docDesc.innerText = "Precision 4mm keyhole portals created. AL viewing portal houses 30° diagnostic camera; AM working portal allows motorized radiofrequency debridement of torn native ACL stump.";
        if (docTriad) docTriad.innerHTML = "<span class='text-cyan-300 font-bold'>Portals:</span> AL 30° Scope / AM Working • <span class='text-rose-400 font-bold'>Torn ACL:</span> Visualized for clearance";
        if (docKaizenBadge) {
          docKaizenBadge.innerText = "🟡 STAGE 1 | ARTHROSCOPIC PORTALS";
          docKaizenBadge.className = "px-2 py-0.5 rounded-full bg-cyan-950/80 text-cyan-300 border border-cyan-700 font-mono tracking-wider uppercase text-[10px]";
        }
        if (radlexText) {
          radlexText.innerText = "PROCEDURE STAGE 1: Dual 4mm keyhole portals established. Midsubstance tear of native ACL confirmed. Remnant stump debrided to clear intercondylar notch.";
        }

        gsapAnimateCam(0.4, 0.4, 6.2);

      } else if (stage === 2) {
        // Stage 2: Precision Anatomical Bone Tunnels
        portalGroup.visible = false;
        tibTunnel.visible = true;
        tunnelGuideMat.opacity = 0.72;
        femTunnel.visible = true;
        reconstructedGraft.visible = false;
        diatomAnchorGroup.visible = false;
        corticalButton.visible = false;
        acl.visible = false; // Torn ACL debrided!
        if (explodedGuides) explodedGuides.visible = false;
        setHardwareExploded(false);

        if (docTitle) docTitle.innerText = "Stage 2: Precision Anatomical Bone Tunnels";
        if (docDesc) docDesc.innerText = "Tibial guide drilled at 55° from anteromedial cortex into intercondylar footprint. Transtibial/anteromedial femoral socket drilled into lateral notch (10:30 o'clock position).";
        if (docTriad) docTriad.innerHTML = "<span class='text-cyan-300 font-bold'>Tibial Tunnel:</span> 55° Anteromedial (9mm) • <span class='text-teal-300 font-bold'>Femoral Socket:</span> 10:30 Notch (30mm depth)";
        if (docKaizenBadge) {
          docKaizenBadge.innerText = "🔵 STAGE 2 | BONE TUNNEL DRILLING";
          docKaizenBadge.className = "px-2 py-0.5 rounded-full bg-blue-950/80 text-blue-300 border border-blue-700 font-mono tracking-wider uppercase text-[10px]";
        }
        if (radlexText) {
          radlexText.innerText = "PROCEDURE STAGE 2: 9mm tibial tunnel drilled at 55° trajectory. Anatomical femoral socket drilled to 30mm depth in lateral notch with 2mm posterior cortical clearance.";
        }

        gsapAnimateCam(-2.0, 0.4, 5.5);

      } else if (stage === 3) {
        // Stage 3: Quadrupled Tendon Autograft Passage
        portalGroup.visible = false;
        tibTunnel.visible = true;
        tunnelGuideMat.opacity = 0.35; // Holographic transparency to reveal graft inside bone
        femTunnel.visible = true;
        reconstructedGraft.visible = true;
        diatomAnchorGroup.visible = false;
        corticalButton.visible = false;
        acl.visible = false;
        if (explodedGuides) explodedGuides.visible = false;
        setHardwareExploded(false);

        if (docTitle) docTitle.innerText = "Stage 3: Quadrupled Tendon Autograft Passage";
        if (docDesc) docDesc.innerText = "Quadrupled semitendinosus/gracilis autograft shuttled retrograde through tibial tunnel, across intercondylar notch, and seated into femoral socket under continuous tension.";
        if (docTriad) docTriad.innerHTML = "<span class='text-amber-300 font-bold'>Autograft:</span> Quadrupled Semitendinosus • <span class='text-teal-300 font-bold'>Graft Length:</span> 120mm braided collagen";
        if (docKaizenBadge) {
          docKaizenBadge.innerText = "🟡 STAGE 3 | GRAFT SHUTTLING";
          docKaizenBadge.className = "px-2 py-0.5 rounded-full bg-amber-950/80 text-amber-300 border border-amber-700 font-mono tracking-wider uppercase text-[10px]";
        }
        if (radlexText) {
          radlexText.innerText = "PROCEDURE STAGE 3: Quadrupled autograft shuttled through tibial tunnel into femoral socket under direct arthroscopic visualization. Graft tensioned to 80N at 30° flexion.";
        }

        gsapAnimateCam(-1.4, 0.8, 5.6);

      } else if (stage === 4) {
        // Stage 4: Dual Biomimetic Fixation & Diatom Bio-Anchor
        portalGroup.visible = false;
        tibTunnel.visible = true;
        tunnelGuideMat.opacity = 0.25;
        femTunnel.visible = true;
        reconstructedGraft.visible = true;
        diatomAnchorGroup.visible = true;
        corticalButton.visible = true;
        acl.visible = false;
        if (explodedGuides) explodedGuides.visible = isExploded;

        if (docTitle) docTitle.innerText = "Stage 4: Suspensory Cortical Fixation & Diatom Bio-Anchor";
        if (docDesc) docDesc.innerText = "Titanium cortical button deployed flush on lateral femoral cortex. Tibial aperture locked with Diatomaceous biosilica porous interference screw (16 µm osteoconductive matrix).";
        if (docTriad) docTriad.innerHTML = "<span class='text-slate-300 font-bold'>Femoral Button:</span> Cortical titanium plate • <span class='text-teal-300 font-bold'>Diatom Anchor:</span> 16 µm porous biosilica";
        if (docKaizenBadge) {
          docKaizenBadge.innerText = "🟢 STAGE 4 | DIATOM BIO-FIXATION";
          docKaizenBadge.className = "px-2 py-0.5 rounded-full bg-teal-950/80 text-teal-300 border border-teal-700 font-mono tracking-wider uppercase text-[10px]";
        }
        if (radlexText) {
          radlexText.innerText = "PROCEDURE STAGE 4: Cortical button deployed and flipped over lateral femoral cortex. Tibial fixation achieved with Diatomaceous biosilica bio-composite interference screw.";
        }

        gsapAnimateCam(-2.4, 0.2, 5.4);
      }

      renderDiatomSlice(currentDepth, activePlaneMode, isDiatomMicroMode);
      requestRender();
    }

    if (btnProcedureToggle) {
      btnProcedureToggle.addEventListener('click', () => {
        isProcedureActive = !isProcedureActive;
        procedureGroup.visible = isProcedureActive;

        if (isProcedureActive) {
          // Transition bone to Ghost Mode for pristine internal visibility!
          boneMode = "ghost";
          femurGroup.visible = true;
          tibiaGroup.visible = true;
          patella.visible = true;
          boneMat.transparent = true;
          boneMat.opacity = 0.26;
          boneMat.depthWrite = false;
          cartilageMat.transparent = true;
          cartilageMat.opacity = 0.45;
          cartilageMat.depthWrite = false;
          if (boneModeLabel) boneModeLabel.innerText = "Ghost";
          if (btnLayerBone) btnLayerBone.className = "px-2.5 py-1 rounded-xl bg-teal-900/80 text-teal-200 border border-teal-400 font-bold transition text-[11px] shadow-[0_0_12px_rgba(20,184,166,0.4)]";

          procedureStatus.innerText = "ON";
          btnProcedureToggle.className = "glass-panel px-2.5 py-1.5 rounded-2xl bg-teal-600 hover:bg-teal-500 text-white text-xs font-semibold transition border border-teal-400 shadow-[0_0_15px_rgba(20,184,166,0.5)] flex items-center gap-1";

          if (procStepperPanel) procStepperPanel.style.display = "flex";
          if (hardwareCallouts) hardwareCallouts.style.display = "block";

          setProcedureStage(1);
        } else {
          procedureStatus.innerText = "OFF";
          btnProcedureToggle.className = "glass-panel px-2.5 py-1.5 rounded-2xl bg-zinc-900 hover:bg-zinc-800 text-teal-300 text-xs font-semibold transition border border-teal-500/40 shadow-lg flex items-center gap-1";

          if (procStepperPanel) procStepperPanel.style.display = "none";
          if (hardwareCallouts) hardwareCallouts.style.display = "none";

          acl.visible = true; // Restore native torn ACL
          setHardwareExploded(false);

          // Reset camera
          gsapAnimateCam(0, 1.5, 9.5);
        }

        renderDiatomSlice(currentDepth, activePlaneMode, isDiatomMicroMode);
        requestRender();
      });
    }

    // Stage Button Listeners
    for (let s = 1; s <= 4; s++) {
      const b = document.getElementById(`btn-proc-stage-${s}`);
      if (b) {
        b.addEventListener('click', () => setProcedureStage(s));
      }
    }

    if (btnProcNext) {
      btnProcNext.addEventListener('click', () => {
        let nextStage = currentProcedureStage + 1;
        if (nextStage > 4) nextStage = 1;
        setProcedureStage(nextStage);
      });
    }

    if (btnExplodedView) {
      btnExplodedView.addEventListener('click', () => {
        if (currentProcedureStage !== 4) {
          setProcedureStage(4);
        }
        setHardwareExploded(!isExploded);
      });
    }

    // --- DIATOM HUD CONTROLS ---
    let isDiatomMicroMode = true; // true: Diatom Micro-CT, false: Knee MRI
    let isDiatomHudCollapsed = false;

    const btnDiatomModeToggle = document.getElementById('btn-diatom-mode-toggle');
    const btnDiatomHudMinimize = document.getElementById('btn-diatom-hud-minimize');
    const diatomCanvasWrapper = document.getElementById('diatom-canvas-wrapper');
    const diatomProcedureTag = document.getElementById('diatom-procedure-tag');

    if (btnDiatomModeToggle) {
      btnDiatomModeToggle.addEventListener('click', () => {
        isDiatomMicroMode = !isDiatomMicroMode;
        btnDiatomModeToggle.innerText = isDiatomMicroMode ? "🔬 CT" : "🩺 MRI";
        btnDiatomModeToggle.className = isDiatomMicroMode
          ? "text-[9px] bg-teal-950/80 hover:bg-teal-900 text-teal-300 px-1.5 py-0.5 rounded-lg border border-teal-700/60 font-mono transition"
          : "text-[9px] bg-amber-950/80 hover:bg-amber-900 text-amber-300 px-1.5 py-0.5 rounded-lg border border-amber-700/60 font-mono transition";
        renderDiatomSlice(currentDepth, activePlaneMode, isDiatomMicroMode);
      });
    }

    if (btnDiatomHudMinimize) {
      btnDiatomHudMinimize.addEventListener('click', () => {
        isDiatomHudCollapsed = !isDiatomHudCollapsed;
        if (diatomCanvasWrapper) diatomCanvasWrapper.style.display = isDiatomHudCollapsed ? 'none' : 'flex';
        if (diatomProcedureTag) diatomProcedureTag.style.display = isDiatomHudCollapsed ? 'none' : 'flex';
        btnDiatomHudMinimize.innerText = isDiatomHudCollapsed ? "+" : "—";
      });
    }

    // --- 2D DIATOM / DICOM VOLUMETRIC SLICE RENDERER ---
    function renderDiatomSlice(depth, plane, microMode) {
      const canvas = document.getElementById('diatom-slice-canvas');
      if (!canvas) return;
      const ctx = canvas.getContext('2d');
      const w = canvas.width;
      const h = canvas.height;
      ctx.clearRect(0, 0, w, h);

      // Deep true obsidian black canvas
      ctx.fillStyle = '#09090b';
      ctx.fillRect(0, 0, w, h);

      const cx = w / 2;
      const cy = h / 2;
      const sliceNum = Math.min(32, Math.max(1, Math.round(depth * 31) + 1));
      const oscillation = Math.sin(depth * Math.PI * 3);

      if (microMode) {
        // 🔬 1. DIATOM FRUSTULE HIGH-RES MICRO-CT (dicom.service.ts Parity)
        const radius = Math.max(32, 50 + oscillation * 18);
        
        // Concentric outer silica frustule rim
        ctx.strokeStyle = '#14b8a6';
        ctx.lineWidth = 3;
        ctx.beginPath();
        ctx.arc(cx, cy, radius, 0, Math.PI * 2);
        ctx.stroke();

        // Inner frustule chamber
        ctx.fillStyle = '#134e4a';
        ctx.strokeStyle = '#2dd4bf';
        ctx.lineWidth = 1.5;
        ctx.beginPath();
        ctx.arc(cx, cy, radius * 0.52, 0, Math.PI * 2);
        ctx.fill();
        ctx.stroke();

        // Central biosilica hub
        ctx.fillStyle = isProcedureActive ? '#facc15' : '#5eead4';
        ctx.beginPath();
        ctx.arc(cx, cy, 7, 0, Math.PI * 2);
        ctx.fill();

        // 16 radial micro-pores around the ring
        const poreCount = 16;
        const poreR = radius * 0.76;
        ctx.fillStyle = '#14b8a6';
        for (let i = 0; i < poreCount; i++) {
          const angle = (i / poreCount) * Math.PI * 2 + (depth * 0.5);
          const px = cx + Math.cos(angle) * poreR;
          const py = cy + Math.sin(angle) * poreR;
          ctx.beginPath();
          ctx.arc(px, py, 2.5, 0, Math.PI * 2);
          ctx.fill();
        }

        // Radial Costae Ribs
        ctx.strokeStyle = 'rgba(45, 212, 191, 0.45)';
        ctx.lineWidth = 1;
        ctx.beginPath();
        ctx.moveTo(cx - radius, cy);
        ctx.lineTo(cx + radius, cy);
        ctx.moveTo(cx, cy - radius);
        ctx.lineTo(cx, cy + radius);
        ctx.stroke();

        // Diagonal trabeculae
        const diag = radius * 0.707;
        ctx.beginPath();
        ctx.moveTo(cx - diag, cy - diag);
        ctx.lineTo(cx + diag, cy + diag);
        ctx.moveTo(cx - diag, cy + diag);
        ctx.lineTo(cx + diag, cy - diag);
        ctx.stroke();

        // Surgical graft axis in procedure mode
        if (isProcedureActive) {
          ctx.strokeStyle = '#f59e0b';
          ctx.lineWidth = 2;
          ctx.setLineDash([4, 3]);
          ctx.beginPath();
          ctx.arc(cx, cy, radius * 0.35, 0, Math.PI * 2);
          ctx.stroke();
          ctx.setLineDash([]);
          ctx.fillStyle = '#f59e0b';
          ctx.font = '8px monospace';
          const stageTag = currentProcedureStage === 1 ? '1. PORTALS' :
                           currentProcedureStage === 2 ? '2. TUNNELS' :
                           currentProcedureStage === 3 ? '3. GRAFT' : '4. DIATOM ANCHOR';
          ctx.fillText(stageTag, cx - 34, cy - 14);
        }
      } else {
        // 🩺 2. MULTI-PLANAR CLINICAL KNEE MRI SCAN (RSNA 2026)
        ctx.fillStyle = '#27272a';
        ctx.strokeStyle = '#52525b';
        ctx.lineWidth = 1.5;

        // Femoral condyle ellipse
        ctx.beginPath();
        ctx.ellipse(cx, cy - 35, 42, 28, 0, 0, Math.PI * 2);
        ctx.fill();
        ctx.stroke();

        // Tibial plateau ellipse
        ctx.beginPath();
        ctx.ellipse(cx, cy + 42, 48, 22, 0, 0, Math.PI * 2);
        ctx.fill();
        ctx.stroke();

        // Articular cartilage line
        ctx.strokeStyle = '#99f6e4';
        ctx.lineWidth = 2;
        ctx.beginPath();
        ctx.arc(cx, cy - 25, 36, 0.2 * Math.PI, 0.8 * Math.PI);
        ctx.stroke();

        // Meniscus wedge
        ctx.fillStyle = (typeof activeFindingKey !== 'undefined' && activeFindingKey && activeFindingKey.includes('meniscus')) ? '#f43f5e' : '#10b981';
        ctx.beginPath();
        ctx.moveTo(cx - 38, cy + 22);
        ctx.lineTo(cx - 15, cy + 22);
        ctx.lineTo(cx - 38, cy + 34);
        ctx.closePath();
        ctx.fill();

        // Cruciate Ligament Band
        ctx.strokeStyle = (typeof activeFindingKey !== 'undefined' && (activeFindingKey === 'acl' || activeFindingKey === 'pcl')) ? '#f43f5e' : '#14b8a6';
        ctx.lineWidth = 4;
        ctx.beginPath();
        ctx.moveTo(cx - 12, cy - 18);
        ctx.lineTo(cx + 8, cy + 24);
        ctx.stroke();

        // Procedure overlay in MRI view
        if (isProcedureActive) {
          ctx.strokeStyle = '#06b6d4';
          ctx.lineWidth = 3;
          ctx.setLineDash([4, 2]);
          ctx.beginPath();
          ctx.moveTo(cx - 30, cy + 50);
          ctx.lineTo(cx + 15, cy - 25);
          ctx.stroke();
          ctx.setLineDash([]);
          ctx.fillStyle = '#14b8a6';
          ctx.beginPath();
          ctx.arc(cx - 18, cy + 38, 5, 0, Math.PI * 2);
          ctx.fill();
        }
      }

      // Corner reticle brackets
      ctx.strokeStyle = '#3f3f46';
      ctx.lineWidth = 1;
      const m = 6, l = 10;
      ctx.beginPath();
      ctx.moveTo(m, m + l); ctx.lineTo(m, m); ctx.lineTo(m + l, m);
      ctx.moveTo(w - m - l, m); ctx.lineTo(w - m, m); ctx.lineTo(w - m, m + l);
      ctx.moveTo(m, h - m - l); ctx.lineTo(m, h - m); ctx.lineTo(m + l, h - m);
      ctx.moveTo(w - m - l, h - m); ctx.lineTo(w - m, h - m); ctx.lineTo(w - m, h - m - l);
      ctx.stroke();

      // Update HUD telemetry text elements
      const elTL = document.getElementById('hud-corner-tl');
      const elTR = document.getElementById('hud-corner-tr');
      const elBL = document.getElementById('hud-corner-bl');
      const elBR = document.getElementById('hud-corner-br');
      const elScaffold = document.getElementById('diatom-scaffold-status');
      const elTitle = document.getElementById('diatom-hud-title');

      if (elTL) elTL.innerText = `SLICE ${sliceNum}/32`;
      if (elTR) elTR.innerText = microMode ? `R: ${(32 + depth * 28).toFixed(1)} µm` : `FOV: 16cm`;
      if (elBL) elBL.innerText = microMode ? 'DIATOM CT' : 'KNEE MRI';
      if (elBR) elBR.innerText = `${plane.toUpperCase().slice(0, 3)} ${depth.toFixed(2)}`;
      if (elTitle) elTitle.innerText = microMode ? 'DIATOM SCAN' : 'DICOM MRI';
      if (elScaffold) {
        if (isProcedureActive) {
          if (currentProcedureStage === 1) {
            elScaffold.innerText = '1. Debridement Active';
            elScaffold.className = 'text-cyan-400 font-bold';
          } else if (currentProcedureStage === 2) {
            elScaffold.innerText = '2. 55° Tunnels Ready';
            elScaffold.className = 'text-blue-400 font-bold';
          } else if (currentProcedureStage === 3) {
            elScaffold.innerText = '3. Graft Seated (80N)';
            elScaffold.className = 'text-amber-400 font-bold';
          } else {
            elScaffold.innerText = isExploded ? '4. Exploded Hardware' : '4. Bio-Fused (96.4%)';
            elScaffold.className = 'text-teal-400 font-bold';
          }
        } else {
          elScaffold.innerText = 'Homeostasis';
          elScaffold.className = 'text-zinc-200';
        }
      }

      const elAffinity = document.getElementById('diatom-affinity-tag');
      if (elAffinity) {
        if (isProcedureActive) {
          if (currentProcedureStage === 1) {
            elAffinity.innerText = 'Torn Stump Debrided';
          } else if (currentProcedureStage === 2) {
            elAffinity.innerText = 'Anatomical Trajectory';
          } else if (currentProcedureStage === 3) {
            elAffinity.innerText = 'Autograft Seated (80N)';
          } else {
            elAffinity.innerText = isExploded ? 'Hardware Displaced' : 'Osteoblast Affinity +96.4%';
          }
        } else {
          elAffinity.innerText = 'Osteoblast Matrix';
        }
      }
    }

    // CAMERA RESET
    document.getElementById('btn-reset-cam').addEventListener('click', () => {
      camera.position.set(0, 1.5, 9.5);
      if (controls) {
        controls.target.set(0, 0, 0);
        controls.update();
      }
      requestRender();
    });

    // --- ⛶ FULLSCREEN TOGGLE CONTROLLER ---
    const btnFullscreen = document.getElementById('btn-fullscreen');
    const fsIcon = document.getElementById('fs-icon');
    const fsLabel = document.getElementById('fs-label');

    function toggleFullscreen() {
      const doc = document;
      const docEl = doc.documentElement;

      const requestFs = docEl.requestFullscreen || docEl.webkitRequestFullscreen || docEl.mozRequestFullScreen || docEl.msRequestFullscreen;
      const exitFs = doc.exitFullscreen || doc.webkitExitFullscreen || doc.mozCancelFullScreen || doc.msExitFullscreen;
      const isFs = doc.fullscreenElement || doc.webkitFullscreenElement || doc.mozFullScreenElement || doc.msFullscreenElement;

      if (!isFs) {
        if (requestFs) {
          requestFs.call(docEl).catch(err => {
            console.warn("Fullscreen request error:", err);
          });
        }
      } else {
        if (exitFs) {
          exitFs.call(doc);
        }
      }
    }

    if (btnFullscreen) {
      btnFullscreen.addEventListener('click', toggleFullscreen);
    }

    function onFsChange() {
      const isFs = document.fullscreenElement || document.webkitFullscreenElement || document.mozFullScreenElement || document.msFullscreenElement;
      if (isFs) {
        if (fsIcon) fsIcon.innerText = "✖";
        if (fsLabel) fsLabel.innerText = "Exit Fullscreen";
        btnFullscreen.className = "glass-panel px-3 py-1.5 rounded-full bg-rose-950/80 hover:bg-rose-900/90 text-rose-300 text-xs font-bold transition flex items-center gap-1.5 border border-rose-500/50 shadow-[0_0_12px_rgba(244,63,94,0.4)]";
      } else {
        if (fsIcon) fsIcon.innerText = "⛶";
        if (fsLabel) fsLabel.innerText = "Fullscreen";
        btnFullscreen.className = "glass-panel px-3 py-1.5 rounded-full bg-teal-950/80 hover:bg-teal-900/90 text-teal-300 text-xs font-bold transition flex items-center gap-1.5 border border-teal-500/50 shadow-[0_0_12px_rgba(20,184,166,0.3)]";
      }
      setTimeout(resizeRenderer, 100);
      setTimeout(resizeRenderer, 400);
    }

    document.addEventListener('fullscreenchange', onFsChange);
    document.addEventListener('webkitfullscreenchange', onFsChange);
    document.addEventListener('mozfullscreenchange', onFsChange);
    document.addEventListener('MSFullscreenChange', onFsChange);

    // --- DOCTOR / PATIENT CONSULT LOGIC & FINDINGS ---
    let isPatientMode = false;

        
    // --- RSNA RADLEX IMPRESSION REPERTOIRE (Anti-Burnout Clinical Macros) ---
    const RADLEX_MACROS = {
      "ACL": "IMPRESSION: Complete tear of the midsubstance anterior cruciate ligament with associated pivot-shift impaction bone contusion of the lateral femoral condyle and posterior lateral tibial plateau. Moderate joint effusion without displaced meniscal fragment.",
      "MCL": "IMPRESSION: Grade II sprain of the superficial medial collateral ligament with periligamentous fluid and soft tissue edema. Tibiofemoral alignment maintained without valgus laxity.",
      "Medial Meniscus": "IMPRESSION: Complex radial tear involving the posterior horn and body of the medial meniscus extending to both superior and inferior articular surfaces. Mild adjacent parameniscal fluid tracking.",
      "Lateral Meniscus": "IMPRESSION: Longitudinal vertical tear of the lateral meniscus posterior horn without bucket-handle displacement. Popliteal hiatus and popliteus tendon appear intact.",
      "Medial OA": "IMPRESSION: Moderate medial compartment osteoarthritis with marked articular cartilage thinning, subchondral sclerosis, and marginal osteophytes. Associated medial meniscal extrusion.",
      "Lateral OA": "IMPRESSION: Lateral compartment osteoarthritis with moderate chondral thinning and mild lateral compartment joint space narrowing without significant subchondral collapse.",
      "PF OA": "IMPRESSION: Patellofemoral osteoarthritis with full-thickness chondral loss over the lateral patellar facet and prominent lateral trochlear osteophyte formation. Mild lateral patellar tilt.",
      "Effusion": "IMPRESSION: Moderate-to-large suprapatellar joint effusion measuring 7.5mm in anterior-posterior depth without evidence of intra-articular loose bodies or lipohemarthrosis.",
      "Synovitis": "IMPRESSION: Diffuse nodular synovial thickening and hypervascular enhancement within the infrapatellar fat pad (Hoffa fat pad) and suprapatellar pouch, consistent with active synovitis.",
      "Baker's": "IMPRESSION: Distended gastrocnemius-semimembranosus bursa (Baker's cyst) measuring 3.8 x 1.9 cm with internal fluid debris and no evidence of rupture into the deep posterior calf.",
      "Contusion": "IMPRESSION: Geographic bone marrow contusion pattern involving the posterior lateral tibial plateau and mid-lateral femoral condyle, characteristic of a pivot-shift impaction injury.",
      "Fracture": "IMPRESSION: Nondisplaced subchondral insufficiency fracture of the medial femoral condyle with extensive surrounding bone marrow edema. Articular cortex remains intact without step-off."
    };

    let activeFindingKey = "ACL";
    let isLeftKnee = false;

    // --- LATERALITY SWITCHER LOGIC ---
    const lateralityBtn = document.getElementById('btn-laterality');
    const lateralityLabel = document.getElementById('laterality-label');
    // Reusing existing depthLoLabel and depthHiLabel
    const compassMed = document.getElementById('compass-med');
    const compassLat = document.getElementById('compass-lat');
    const modalLatBadge = document.getElementById('modal-lat-badge');

    if (lateralityBtn) {
      lateralityBtn.addEventListener('click', () => {
        isLeftKnee = !isLeftKnee;
        if (lateralityLabel) lateralityLabel.innerText = isLeftKnee ? "Left Knee" : "Right Knee";
        if (modalLatBadge) modalLatBadge.innerText = isLeftKnee ? "Left Knee" : "Right Knee";
        
        // Mirror 3D Knee Anatomy across X axis
        if (typeof kneeGroup !== 'undefined') {
          kneeGroup.scale.x = isLeftKnee ? -1 : 1;
        }

        // Swap boundary labels on depth slider
        if (activePlaneMode === 'sagittal') {
          if (depthLoLabel) depthLoLabel.innerText = isLeftKnee ? "Lateral" : "Medial";
          if (depthHiLabel) depthHiLabel.innerText = isLeftKnee ? "Medial" : "Lateral";
        }

        // Swap compass labels
        if (compassMed && compassLat) {
          compassMed.innerText = isLeftKnee ? "L" : "M";
          compassLat.innerText = isLeftKnee ? "M" : "L";
        }

        requestRender();
      });
    }

    // --- 👁️ CIRCADIAN OCULAR & 3D LAYER SOFTENING ENGINE ---
    // Grounded in Helmholtz Retinal Irradiation & PocketGull A11y Studio (a11y_studio.html)
    // Adapts color temperature, retinal halation compensation (-3.75% optical stroke dial-back),
    // and biological subsurface light scattering across the 24-hour day.

    function getLocalCircadianTone() {
      const hr = new Date().getHours();
      if (hr >= 8 && hr < 17) return 'day';
      if (hr >= 17 && hr < 21) return 'dusk';
      return 'night';
    }

    function applyCircadianSoftening(targetTone) {
      currentCircadianTone = targetTone;
      const isNight = targetTone === 'night';
      const isDusk = targetTone === 'dusk';
      const isDay = targetTone === 'day';

      // 1. UI Retinal Irradiation & Typography Softening
      document.body.classList.remove('circadian-day', 'circadian-dusk', 'circadian-night');
      if (isNight) {
        document.body.classList.add('circadian-night', 'pg-irrd-compensated');
      } else if (isDusk) {
        document.body.classList.add('circadian-dusk', 'pg-irrd-compensated');
      } else {
        document.body.classList.add('circadian-day');
        document.body.classList.remove('pg-irrd-compensated');
      }

      // 2. Three.js Background, Volumetric Fog & Atmospheric Lighting
      if (isNight) {
        if (scene.background) scene.background.set(0x120e0a); // Deep warm obsidian umber
        if (scene.fog) {
          scene.fog.color.set(0x120e0a);
          scene.fog.density = 0.024; // Soft atmospheric envelopment
        }
        if (ambLight) {
          ambLight.color.set(0xfef3c7); // Warm candlelight amber phosphor (2200K)
          ambLight.intensity = 0.85;
        }
        if (dirLight1) {
          dirLight1.color.set(0xfbbf24); // Warm golden key (3000K)
          dirLight1.intensity = 0.72;
        }
        if (dirLight2) {
          dirLight2.color.set(0xd97706); // Warm terra-cotta rim
          dirLight2.intensity = 0.58;
        }
      } else if (isDusk) {
        if (scene.background) scene.background.set(0x0e0e14); // Twilight obsidian
        if (scene.fog) {
          scene.fog.color.set(0x0e0e14);
          scene.fog.density = 0.030;
        }
        if (ambLight) {
          ambLight.color.set(0xffedd5); // Golden hour fill (3500K)
          ambLight.intensity = 0.78;
        }
        if (dirLight1) {
          dirLight1.color.set(0xf59e0b); // Golden amber key (3800K)
          dirLight1.intensity = 0.82;
        }
        if (dirLight2) {
          dirLight2.color.set(0x2dd4bf); // Soft teal back
          dirLight2.intensity = 0.50;
        }
      } else {
        // Crisp Clinical Daylight (5500K - 6500K)
        if (scene.background) scene.background.set(0x09090b);
        if (scene.fog) {
          scene.fog.color.set(0x09090b);
          scene.fog.density = 0.035;
        }
        if (ambLight) {
          ambLight.color.set(0xffffff);
          ambLight.intensity = 0.70;
        }
        if (dirLight1) {
          dirLight1.color.set(0x5eead4); // Clinical teal key
          dirLight1.intensity = 0.90;
        }
        if (dirLight2) {
          dirLight2.color.set(0xf59e0b); // Subtle gold rim
          dirLight2.intensity = 0.45;
        }
      }

      // 3. Volumetric 3D Anatomical Layer Softening (Subsurface Scattering & Roughness)
      if (boneMat) {
        if (isNight) {
          boneMat.roughness = 0.68; // Diffuse warm living ivory (softens harsh specular glints)
          boneMat.metalness = 0.06;
          if (boneMode === "ghost") boneMat.opacity = 0.18;
        } else if (isDusk) {
          boneMat.roughness = 0.55;
          boneMat.metalness = 0.10;
          if (boneMode === "ghost") boneMat.opacity = 0.22;
        } else {
          boneMat.roughness = 0.45;
          boneMat.metalness = 0.15;
          if (boneMode === "ghost") boneMat.opacity = 0.26;
        }
      }

      if (cartilageMat) {
        cartilageMat.roughness = isNight ? 0.42 : (isDusk ? 0.28 : 0.15);
      }

      if (muscleMat) {
        muscleMat.roughness = isNight ? 0.82 : (isDusk ? 0.72 : 0.62);
        if (muscleMode === "ghost") {
          muscleMat.opacity = isNight ? 0.16 : (isDusk ? 0.19 : 0.22);
        }
      }

      updateChronoDockUI();
      requestRender();
    }

    function updateChronoDockUI() {
      const btnDay = document.getElementById('btn-chrono-day');
      const btnDusk = document.getElementById('btn-chrono-dusk');
      const btnNight = document.getElementById('btn-chrono-night');
      const btnAuto = document.getElementById('btn-chrono-auto');
      const autoLabel = document.getElementById('chrono-auto-label');

      const activeClass = "px-2.5 py-1 rounded-full bg-teal-600 text-white font-bold transition text-[11px] shadow-sm";
      const inactiveClass = "px-2.5 py-1 rounded-full text-zinc-400 hover:text-white transition text-[11px]";

      if (btnDay) btnDay.className = (currentCircadianTone === 'day' && !isCircadianAuto) ? activeClass : inactiveClass;
      if (btnDusk) btnDusk.className = (currentCircadianTone === 'dusk' && !isCircadianAuto) ? activeClass : inactiveClass;
      if (btnNight) btnNight.className = (currentCircadianTone === 'night' && !isCircadianAuto) ? activeClass : inactiveClass;
      if (btnAuto) {
        btnAuto.className = isCircadianAuto 
          ? "px-1.5 py-1 rounded-xl bg-teal-900/80 text-teal-200 border border-teal-400 font-bold transition text-[11px] font-mono shadow-[0_0_10px_rgba(20,184,166,0.3)]" 
          : "px-1.5 py-1 rounded-xl text-zinc-400 hover:text-white transition text-[11px] font-mono";
      }
      if (autoLabel) {
        autoLabel.innerText = isCircadianAuto ? `Auto (${currentCircadianTone.toUpperCase()})` : "Auto";
      }
    }

    const btnChronoDay = document.getElementById('btn-chrono-day');
    const btnChronoDusk = document.getElementById('btn-chrono-dusk');
    const btnChronoNight = document.getElementById('btn-chrono-night');
    const btnChronoAuto = document.getElementById('btn-chrono-auto');

    if (btnChronoDay) {
      btnChronoDay.addEventListener('click', () => {
        isCircadianAuto = false;
        applyCircadianSoftening('day');
      });
    }
    if (btnChronoDusk) {
      btnChronoDusk.addEventListener('click', () => {
        isCircadianAuto = false;
        applyCircadianSoftening('dusk');
      });
    }
    if (btnChronoNight) {
      btnChronoNight.addEventListener('click', () => {
        isCircadianAuto = false;
        applyCircadianSoftening('night');
      });
    }
    if (btnChronoAuto) {
      btnChronoAuto.addEventListener('click', () => {
        isCircadianAuto = true;
        applyCircadianSoftening(getLocalCircadianTone());
      });
    }

    // Mobile Diatom Micro-CT HUD Toggle
    const btnMobileDiatomToggle = document.getElementById('btn-mobile-diatom-toggle');
    if (btnMobileDiatomToggle) {
      btnMobileDiatomToggle.addEventListener('click', () => {
        const hud = document.getElementById('diatom-hud');
        if (hud) {
          hud.classList.toggle('hidden');
          hud.classList.toggle('flex');
        }
      });
    }

    // Initialize with auto circadian tone at boot!
    applyCircadianSoftening(getLocalCircadianTone());

    // --- 1-CLICK RADLEX COPY TO PACS ---
    const btnCopyRadlex = document.getElementById('btn-copy-radlex');
    const dockRadlexText = document.getElementById('dock-radlex-text');
    const copyBtnLabel = document.getElementById('copy-btn-label');

    if (btnCopyRadlex && dockRadlexText) {
      btnCopyRadlex.addEventListener('click', () => {
        const textToCopy = dockRadlexText.innerText;
        navigator.clipboard.writeText(textToCopy).then(() => {
          if (copyBtnLabel) copyBtnLabel.innerText = "✓ Copied to PACS!";
          btnCopyRadlex.classList.remove('bg-teal-600', 'hover:bg-teal-500');
          btnCopyRadlex.classList.add('bg-emerald-600', 'hover:bg-emerald-500');
          setTimeout(() => {
            if (copyBtnLabel) copyBtnLabel.innerText = "Copy to PACS";
            btnCopyRadlex.classList.remove('bg-emerald-600', 'hover:bg-emerald-500');
            btnCopyRadlex.classList.add('bg-teal-600', 'hover:bg-teal-500');
          }, 2000);
        });
      });
    }

    // --- 1-PAGE PATIENT CARE CARD MODAL LOGIC ---
    const careCardModal = document.getElementById('care-card-modal');
    const btnPrintCareCard = document.getElementById('btn-print-care-card');
    const modalBtnClose = document.getElementById('modal-btn-close');
    const modalBtnPrint = document.getElementById('modal-btn-print');
    const modalFindingTitle = document.getElementById('modal-finding-title');
    const modalWhy = document.getElementById('modal-why');
    const modalSens = document.getElementById('modal-sens');
    const modalRec = document.getElementById('modal-rec');

    if (btnPrintCareCard && careCardModal) {
      btnPrintCareCard.addEventListener('click', () => {
        const info = FINDING_INFOS[activeFindingKey] || FINDING_INFOS["ACL"];
        if (modalFindingTitle) modalFindingTitle.innerText = info.patLabel;
        if (modalWhy) modalWhy.innerText = info.patWhy;
        if (modalSens) modalSens.innerText = info.patSensation;
        if (modalRec) modalRec.innerText = info.patRecovery;
        careCardModal.classList.remove('hidden');
      });
    }

    if (modalBtnClose && careCardModal) {
      modalBtnClose.addEventListener('click', () => careCardModal.classList.add('hidden'));
    }

    if (modalBtnPrint) {
      modalBtnPrint.addEventListener('click', () => {
        window.print();
      });
    }

    // --- SMOOTH AUTOPILOT SLICING DEPTH TELEPORT ---
    let depthAnimFrame = null;
    function autopilotDepthTeleport(targetDepth) {
      if (depthAnimFrame) cancelAnimationFrame(depthAnimFrame);
      const startDepth = currentDepth;
      const startTime = performance.now();
      const duration = 300; // ms

      function step(now) {
        const elapsed = now - startTime;
        const progress = Math.min(elapsed / duration, 1.0);
        // Ease-out cubic
        const ease = 1 - Math.pow(1 - progress, 3);
        currentDepth = startDepth + (targetDepth - startDepth) * ease;
        if (depthSlider) depthSlider.value = currentDepth;
        updateSlicingPlane();

        if (progress < 1.0) {
          depthAnimFrame = requestAnimationFrame(step);
        } else {
          currentDepth = targetDepth;
          if (depthSlider) depthSlider.value = targetDepth;
          updateSlicingPlane();
        }
      }
      depthAnimFrame = requestAnimationFrame(step);
    }

    const FINDING_INFOS = {
      "ACL": {
        plane: "sagittal", depth: 0.50, cam: [1.8, 1.2, 5.5],
        targetMesh: acl, companionMeshes: [contusionMesh, medMeniscus],
        kaizenClass: "red",
        kaizenBadge: "🔴 ANDON RED | ACUTE CRUCIATE TEAR",
        heuristicName: "Blumensaat Line & Pivot-Shift Rule",
        docLabel: "ACL Tear (Intercondylar Notch)",
        docDesc: "Central sagittal cuts show fiber discontinuity, fluid sign, and Blumensaat angle deviation >15 degrees.",
        triadStat: "ACL tears co-occur with <b class='text-amber-400'>Bone Contusions at 88%</b> (lateral plateau impaction) and <b class='text-rose-400'>Medial Meniscus at 64%</b>.",
        patLabel: "Torn Front Stabilizer (ACL)",
        patWhy: "When your knee pivoted during your twist, the main central stabilizer ligament stretched past its limit and tore.",
        patSensation: "Explains why your knee felt like it 'gave way' and why walking down stairs feels unstable.",
        patRecovery: "Act I: Reduce fluid swelling with RICE. Act II: Hamstring strengthening to serve as an internal seatbelt. Act III: Safe return to sport."
      },
      "MCL": {
        plane: "coronal", depth: 0.50, cam: [-4.5, 0.5, 3.5],
        targetMesh: mcl, companionMeshes: [medMeniscus, medFemCart],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | PERILIGAMENTOUS STRAIN",
        heuristicName: "Pellegrini-Stieda & Feathery Edema Rule",
        docLabel: "MCL Sprain (Medial Collateral)",
        docDesc: "Coronal cuts reveal periligamentous fluid and thickening along medial femoral condyle without complete fiber gap.",
        triadStat: "MCL sprains frequently involve <b class='text-amber-400'>Medial Meniscal shear (52%)</b> and <b class='text-rose-400'>Medial OA progression</b>.",
        patLabel: "Strained Inner Band (MCL)",
        patWhy: "The inner strap holding the knee stable was pulled taut when force pushed your knee inward.",
        patSensation: "Causes direct tenderness when touching the inner edge of your knee and pain when walking.",
        patRecovery: "MCLs have rich blood flow and usually heal without surgery using a protective hinged brace for 4 to 6 weeks."
      },
      "Medial Meniscus": {
        plane: "sagittal", depth: 0.22, cam: [-3.2, 0.8, 4.2],
        targetMesh: medMeniscus, companionMeshes: [medFemCart, acl],
        kaizenClass: "red",
        kaizenBadge: "🔴 ANDON RED | MENISCAL TEAR",
        heuristicName: "Bow-Tie 2-Slice & Articular Cleft Rule",
        docLabel: "Medial Meniscal Tear (Posterior Horn)",
        docDesc: "Loss of standard 2-slice bow-tie contour and linear high T2 signal extending to the inferior articular surface.",
        triadStat: "Medial tears co-occur with <b class='text-amber-400'>Medial Compartment OA at 71%</b> and <b class='text-rose-400'>ACL insufficiency at 64%</b>.",
        patLabel: "Torn Inner Cushion (Medial Meniscus)",
        patWhy: "The inner C-shaped shock-absorbing cushion was pinched and torn during the twisting force.",
        patSensation: "Causes catching, clicking, and a sharp ache when squatting or bending the knee deeply.",
        patRecovery: "Protected weight-bearing, physical therapy to strengthen quad shock-absorption, and arthroscopic repair if locked."
      },
      "Lateral Meniscus": {
        plane: "sagittal", depth: 0.78, cam: [3.2, 0.8, 4.2],
        targetMesh: latMeniscus, companionMeshes: [latFemCart, contusionMesh],
        kaizenClass: "red",
        kaizenBadge: "🔴 ANDON RED | MENISCAL TEAR",
        heuristicName: "Popliteus Hiatus vs Peripheral Detachment Rule",
        docLabel: "Lateral Meniscal Tear",
        docDesc: "Radial or longitudinal disruption distinct from the normal popliteus tendon hiatus.",
        triadStat: "Lateral root tears increase <b class='text-amber-400'>Peak Lateral Compartment Contact Stress by 300%</b>.",
        patLabel: "Torn Outer Cushion (Lateral Meniscus)",
        patWhy: "The outer shock absorber was damaged, reducing the smooth gliding of your thigh bone on the shin.",
        patSensation: "Causes outer joint line sharpness when pivoting or turning while walking.",
        patRecovery: "Low-impact cycling, unweighted kinetic chain therapy, and surgical re-anchoring if extruded."
      },
      "Medial OA": {
        plane: "coronal", depth: 0.50, cam: [-2.5, 0.2, 4.0],
        targetMesh: medFemCart, companionMeshes: [medMeniscus, plateau],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | DEGENERATIVE JOINT WATCH",
        heuristicName: "Kellgren-Lawrence Joint Space Ratio",
        docLabel: "Medial Compartment Osteoarthritis",
        docDesc: "Joint space narrowing >50%, subchondral bone sclerosis, and marginal osteophytic lipping.",
        triadStat: "Medial OA strongly correlates with <b class='text-amber-400'>Medial Meniscus extrusion (78%)</b> and varus deformity.",
        patLabel: "Inner Joint Wear (Medial OA)",
        patWhy: "The protective cartilage cap over the bone has thinned over time from repetitive loading.",
        patSensation: "Morning stiffness that eases with gentle warmth and aching after prolonged standing.",
        patRecovery: "Act I: Low-impact pool walking. Act II: Gluteal and vastus medialis strengthening to redistribute knee pressure."
      },
      "Lateral OA": {
        plane: "coronal", depth: 0.50, cam: [2.5, 0.2, 4.0],
        targetMesh: latFemCart, companionMeshes: [latMeniscus, plateau],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | DEGENERATIVE JOINT WATCH",
        heuristicName: "Valgus Kinematic Load & Lateral Chondral Fissure",
        docLabel: "Lateral Compartment Osteoarthritis",
        docDesc: "Lateral joint space narrowing with full-thickness chondral loss over the lateral tibial plateau.",
        triadStat: "Lateral OA co-occurs with <b class='text-amber-400'>Lateral Meniscal fraying (62%)</b> and IT band tightness.",
        patLabel: "Outer Joint Wear (Lateral OA)",
        patWhy: "Cartilage smoothing along the outside of the knee has worn down.",
        patSensation: "Dull outer knee ache after distance walking or uneven ground.",
        patRecovery: "Orthotic shoe inserts to realign foot strike, hamstring lengthening, and low-impact swimming."
      },
      "PF OA": {
        plane: "axial", depth: 0.40, cam: [0, 4.5, 3.5],
        targetMesh: patella, companionMeshes: [effusionMesh, vl],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | PATELLOFEMORAL CHONDROMALACIA",
        heuristicName: "Merchant Tilt & Trochlear Dysplasia Rule",
        docLabel: "Patellofemoral Osteoarthritis",
        docDesc: "Retro-patellar cartilage loss with lateral facet subchondral bone contact during deep flexion.",
        triadStat: "PF OA co-occurs with <b class='text-amber-400'>Suprapatellar Synovitis (68%)</b> and quad atrophy.",
        patLabel: "Kneecap Cartilage Wear (PF OA)",
        patWhy: "The slippery cartilage lining behind your kneecap is rough, creating friction as it slides across the thigh bone.",
        patSensation: "Creaking, grinding sensations ('crepitus') when climbing stairs or rising from a low chair.",
        patRecovery: "Strengthening the inner quad (VMO) to pull the kneecap into a centered, pain-free gliding groove."
      },
      "Effusion": {
        plane: "axial", depth: 0.25, cam: [0, 3.8, 4.2],
        targetMesh: effusionMesh, companionMeshes: [bakerCyst, medFemCart],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | CAPSULAR DISTENSION",
        heuristicName: "3-5-10 mm Suprapatellar Fluid Rule",
        docLabel: "Joint Effusion (Fluid Distension)",
        docDesc: "Suprapatellar pouch distension with fluid depth >10mm, indicating acute internal trauma or chronic synovitis.",
        triadStat: "Tense effusion accompanies <b class='text-rose-400'>Acute ACL Tears in 85%</b> of cases.",
        patLabel: "Joint Fluid Swelling (Water on the Knee)",
        patWhy: "Your knee produced extra synovial fluid to protect itself, acting like an internal fluid splint.",
        patSensation: "Tight, heavy, 'balloon-like' pressure inside the joint that restricts full bending or straightening.",
        patRecovery: "Act I: Ice, elevation, and gentle ankle pumps to circulate fluid back into the lymphatic system."
      },
      "Synovitis": {
        plane: "axial", depth: 0.32, cam: [0, 3.2, 4.0],
        targetMesh: effusionMesh, companionMeshes: [effusionMesh, medFemCart],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | INFLAMMATORY HYPERTROPHY",
        heuristicName: "Hoffa Fat Pad Infiltration & Fronding",
        docLabel: "Synovitis (Synovial Thickening)",
        docDesc: "Hypertrophied, thickened synovial lining with high T2 frond-like projection into the infrapatellar fat pad.",
        triadStat: "Synovitis is the primary driver of <b class='text-amber-400'>Osteoarthritic pain flares (82%)</b>.",
        patLabel: "Inflamed Joint Lining (Synovitis)",
        patWhy: "The paper-thin lubricating membrane lining your knee is inflamed, thickened, and tender.",
        patSensation: "Warm, puffy feeling around the kneecap that throbs after activity.",
        patRecovery: "Anti-inflammatory pacing, cryotherapy, and gentle range-of-motion to flush metabolic waste."
      },
      "Baker's": {
        plane: "coronal", depth: 0.85, cam: [-1.8, 0.5, -4.5],
        targetMesh: bakerCyst, companionMeshes: [medMeniscus, effusionMesh],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | POPLITEAL VALVE ESCAPE",
        heuristicName: "Gastrocnemius-Semimembranosus Stalk Rule",
        docLabel: "Baker's Cyst (Popliteal Fluid)",
        docDesc: "Fluid distension of the gastrocnemius-semimembranosus bursa through a one-way posteromedial capsular valve.",
        triadStat: "Baker's cysts are secondary to <b class='text-amber-400'>Medial Meniscal Tears (84%)</b> or chronic effusion.",
        patLabel: "Popliteal Relief Pocket (Baker's Cyst)",
        patWhy: "Excess fluid escaped out the back of the knee joint through a natural pressure valve, creating a fluid pocket.",
        patSensation: "Fullness or tightness behind the knee, especially noticeable when fully extending your leg.",
        patRecovery: "Treat the underlying meniscus or inflammation in the front; once joint pressure drops, the cyst reabsorbs."
      },
      "Contusion": {
        plane: "coronal", depth: 0.55, cam: [2.8, -0.4, 3.8],
        targetMesh: contusionMesh, companionMeshes: [acl, latMeniscus],
        kaizenClass: "amber",
        kaizenBadge: "🟡 KAIZEN AMBER | BONE MARROW EDEMA",
        heuristicName: "Kissing Bone Contusion Vector",
        docLabel: "Bone Contusion (Trabecular Microfracture)",
        docDesc: "High T2/STIR bone marrow edema in the posterolateral tibial plateau and lateral femoral condyle from pivot impact.",
        triadStat: "Bone contusion is present in <b class='text-rose-400'>88% of Acute ACL Tears</b>, proving rotational impact.",
        patLabel: "Deep Bone Bruise (Trabecular Impact)",
        patWhy: "During your injury twist, the thigh and shin bones slammed together like bumper cars, bruising the bone inside.",
        patSensation: "Deep, aching bone pain that lingers even when sitting still or resting.",
        patRecovery: "Act I: Partial crutch unweighting to allow internal micro-bone trabeculae to remodel without stress fracture."
      },
      "Fracture": {
        plane: "coronal", depth: 0.50, cam: [2.5, -0.8, 3.8],
        targetMesh: plateau, companionMeshes: [contusionMesh, fibHead],
        kaizenClass: "red",
        kaizenBadge: "🔴 ANDON RED | CORTICAL STRUCTURAL BREAK",
        heuristicName: "Schatzker Plateau Depression & Step-Off Rule",
        docLabel: "Tibial Plateau Fracture",
        docDesc: "Sharp cortical disruption with >2mm plateau depression; risk of associated cruciate and meniscal avulsion.",
        triadStat: "Plateau fractures have a <b class='text-rose-400'>55% risk of concomitant soft-tissue cruciate/meniscal tears</b>.",
        patLabel: "Bone Fracture (Tibial Plateau)",
        patWhy: "The force of impact cracked the weight-bearing bone shelf at the top of your shin.",
        patSensation: "Severe, immediate inability to bear weight with localized bone tenderness.",
        patRecovery: "Strict non-weight-bearing in an immobilizer brace; immediate orthopedic evaluation for surgical fixation."
      }
    };

        // --- DOCTOR / PATIENT CONSULT LOGIC & LISTENERS ---
    const docModeBtn = document.getElementById('btn-mode-doc');
    const patModeBtn = document.getElementById('btn-mode-pat');
    const dockDocContent = document.getElementById('dock-doc-content');
    const dockPatContent = document.getElementById('dock-pat-content');
    const headerModeBadge = document.getElementById('header-mode-badge');

    docModeBtn.addEventListener('click', () => {
      isPatientMode = false;
      docModeBtn.className = "px-2.5 py-1 rounded-xl bg-teal-600 text-white font-bold transition";
      patModeBtn.className = "px-2.5 py-1 rounded-xl text-zinc-400 hover:text-white transition";
      if (headerModeBadge) headerModeBadge.innerText = "3D MRI SCANNER";
      if (dockDocContent) dockDocContent.classList.remove('hidden');
      if (dockPatContent) dockPatContent.classList.add('hidden');
      updateButtonLabels();
      updateSlicingPlane();
    });

    patModeBtn.addEventListener('click', () => {
      isPatientMode = true;
      patModeBtn.className = "px-2.5 py-1 rounded-xl bg-teal-600 text-white font-bold transition";
      docModeBtn.className = "px-2.5 py-1 rounded-xl text-zinc-400 hover:text-white transition";
      if (headerModeBadge) headerModeBadge.innerText = "3D PATIENT CONSULT";
      if (dockDocContent) dockDocContent.classList.add('hidden');
      if (dockPatContent) dockPatContent.classList.remove('hidden');
      updateButtonLabels();
      updateSlicingPlane();
    });

    function updateButtonLabels() {
      const btns = document.querySelectorAll('.finding-btn');
      btns.forEach(btn => {
        const fKey = btn.getAttribute('data-f');
        const info = FINDING_INFOS[fKey];
        if (!info) return;
        const dotColor = info.kaizenClass === 'red' ? 'bg-rose-500' : 'bg-amber-400';
        let labelText = '';
        if (isPatientMode) {
          labelText = info.patLabel.split('(')[0].trim();
        } else {
          labelText = fKey === 'Medial Meniscus' ? 'Med. Meniscus' : (fKey === 'Lateral Meniscus' ? 'Lat. Meniscus' : fKey);
        }
        btn.innerHTML = `<span class="w-1.5 h-1.5 rounded-full ${dotColor} shrink-0"></span><span>${labelText}</span>`;
      });
    }

    const findingBtns = document.querySelectorAll('.finding-btn');
    findingBtns.forEach(btn => {
      btn.addEventListener('click', () => {
        findingBtns.forEach(b => {
          b.className = "finding-btn py-1 px-1.5 rounded-lg bg-zinc-900 text-zinc-300 border border-zinc-800 hover:border-teal-500 hover:text-white transition text-center truncate font-medium";
        });
        btn.className = "finding-btn py-1 px-1.5 rounded-lg bg-teal-500/30 text-teal-300 border border-teal-500 font-bold hover:bg-teal-500/40 transition text-center truncate shadow-[0_0_12px_rgba(20,184,166,0.4)]";

        const fKey = btn.getAttribute('data-f');
        const info = FINDING_INFOS[fKey];
        if (!info) return;

        activePlaneMode = info.plane;
        highlightPlaneBtn(`btn-plane-${info.plane}`);

        activeFindingKey = fKey;
        if (dockRadlexText) {
          dockRadlexText.innerText = RADLEX_MACROS[fKey] || "IMPRESSION: Evaluating findings.";
        }
        autopilotDepthTeleport(info.depth);


        gsapAnimateCam(info.cam[0], info.cam[1], info.cam[2]);

        [acl, pcl, mcl, lcl, medMeniscus, latMeniscus, medFemCart, latFemCart, bakerCyst, effusionMesh, contusionMesh, plateau, patella].forEach(m => {
          if (m === acl || m === pcl || m === mcl || m === lcl) m.material = ligamentMat;
          else if (m === medMeniscus || m === latMeniscus || m === medFemCart || m === latFemCart) m.material = cartilageMat;
          else if (m === plateau || m === patella) m.material = boneMat;
        });

        // Apply Kaizen Color Coded Emissive Materials
        const primaryMat = info.kaizenClass === "red" ? kaizenRedMat : kaizenAmberMat;
        if (info.targetMesh) info.targetMesh.material = primaryMat;
        if (info.companionMeshes) info.companionMeshes.forEach(cm => cm.material = kaizenAmberMat);

        // Update Dock Doctor View Details with Kaizen & Heuristic Engine
        const docTitle = document.getElementById('dock-doc-title');
        const docDesc = document.getElementById('dock-doc-desc');
        const docPlane = document.getElementById('dock-doc-plane');
        const docTriad = document.getElementById('dock-doc-triad');
        const docKaizenBadge = document.getElementById('dock-doc-kaizen-badge');
        const docHeuristic = document.getElementById('dock-doc-heuristic');

        if (docTitle) docTitle.innerText = info.docLabel;
        if (docDesc) docDesc.innerText = info.docDesc;
        if (docPlane) docPlane.innerText = `${info.plane.toUpperCase()} ${info.depth.toFixed(2)}`;
        if (docTriad) docTriad.innerHTML = info.triadStat;
        if (docKaizenBadge) {
          docKaizenBadge.innerText = info.kaizenBadge || (info.kaizenClass === 'red' ? '🔴 ANDON RED | ACUTE TEAR' : '🟡 KAIZEN AMBER | SECONDARY WATCH');
          docKaizenBadge.innerText = info.kaizenBadge;
          docKaizenBadge.className = info.kaizenClass === "red"
            ? "px-2 py-0.5 rounded-full bg-rose-950/80 text-rose-300 border border-rose-800/80 font-mono tracking-wider uppercase text-[10px]"
            : "px-2 py-0.5 rounded-full bg-amber-950/80 text-amber-300 border border-amber-800/80 font-mono tracking-wider uppercase text-[10px]";
        }
        if (docHeuristic) docHeuristic.innerText = info.heuristicName || 'Blumensaat Line & Pivot-Shift Rule';

        // Update Dock Patient View Details
        const patTitle = document.getElementById('dock-pat-title');
        const patWhy = document.getElementById('dock-pat-why');
        const patSens = document.getElementById('dock-pat-sens');
        const patRec = document.getElementById('dock-pat-rec');
        if (patTitle) patTitle.innerText = info.patLabel;
        if (patWhy) patWhy.innerText = info.patWhy;
        if (patSens) patSens.innerText = info.patSensation;
        if (patRec) patRec.innerText = info.patRecovery;

        headerStructBadge.innerText = isPatientMode ? info.patLabel : info.docLabel;
        requestRender();
      });
    });

    // --- KAIZEN ACUITY FILTER LOGIC ---
    const filterAll = document.getElementById('filter-kaizen-all');
    const filterRed = document.getElementById('filter-kaizen-red');
    const filterAmber = document.getElementById('filter-kaizen-amber');

    function applyKaizenFilter(filterType) {
      if (filterAll) filterAll.className = "px-2 py-0.5 rounded-md bg-zinc-950 text-zinc-400 border border-zinc-800 font-bold transition";
      if (filterRed) filterRed.className = "px-2 py-0.5 rounded-md bg-zinc-950 text-zinc-400 border border-zinc-800 font-bold transition flex items-center gap-1";
      if (filterAmber) filterAmber.className = "px-2 py-0.5 rounded-md bg-zinc-950 text-zinc-400 border border-zinc-800 font-bold transition flex items-center gap-1";

      if (filterType === 'all' && filterAll) {
        filterAll.className = "px-2 py-0.5 rounded-md bg-zinc-800 text-teal-300 font-bold border border-teal-500/40 transition";
      } else if (filterType === 'red' && filterRed) {
        filterRed.className = "px-2 py-0.5 rounded-md bg-rose-950/90 text-rose-300 border border-rose-600 font-bold transition flex items-center gap-1";
      } else if (filterType === 'amber' && filterAmber) {
        filterAmber.className = "px-2 py-0.5 rounded-md bg-amber-950/90 text-amber-300 border border-amber-600 font-bold transition flex items-center gap-1";
      }

      findingBtns.forEach(btn => {
        const k = btn.getAttribute('data-k');
        if (filterType === 'all' || k === filterType) {
          btn.style.display = "flex";
        } else {
          btn.style.display = "none";
        }
      });
    }

    if (filterAll) filterAll.addEventListener('click', () => applyKaizenFilter('all'));
    if (filterRed) filterRed.addEventListener('click', () => applyKaizenFilter('red'));
    if (filterAmber) filterAmber.addEventListener('click', () => applyKaizenFilter('amber'));


    // --- GREENOPS ON-DEMAND RENDER LOOP ---
    function render() {
      renderRequested = false;
      if (controls) controls.update();

      // Dynamic 3D Anatomical Compass Azimuth Rotation
      const compassEl = document.getElementById('anatomical-compass');
      if (compassEl && camera) {
        const angle = Math.atan2(camera.position.x, camera.position.z);
        const deg = -angle * (180 / Math.PI);
        compassEl.style.transform = `rotate(${deg.toFixed(1)}deg)`;
      }

      // Dynamic Screen-Space Hardware Callout Tracking
      const calloutsContainer = document.getElementById('hardware-callouts');
      if (isProcedureActive && calloutsContainer) {
        calloutsContainer.style.display = 'block';
        const w = (container && container.clientWidth) || window.innerWidth;
        const h = (container && container.clientHeight) || window.innerHeight;
        const tmpVec = new THREE.Vector3();

        // 1. Femoral Suspensory Button (With Direct Smooth Curve)
        const elCalloutBtn = document.getElementById('callout-button');
        const pathBtn = document.getElementById('curve-path-button');
        const dotBtn = document.getElementById('curve-dot-button');
        if (elCalloutBtn) {
          if (corticalButton && corticalButton.visible && currentProcedureStage === 4) {
            corticalButton.getWorldPosition(tmpVec);
            tmpVec.project(camera);
            if (tmpVec.z <= 1.0) {
              const sx = (tmpVec.x * 0.5 + 0.5) * w;
              const sy = (-(tmpVec.y * 0.5) + 0.5) * h;
              const bx = sx + 80;
              const by = sy - 40;
              elCalloutBtn.style.display = 'flex';
              elCalloutBtn.style.left = `${bx}px`;
              elCalloutBtn.style.top = `${by}px`;
              if (pathBtn && dotBtn) {
                pathBtn.style.display = 'block';
                dotBtn.style.display = 'block';
                const cx1 = sx + 30;
                const cy1 = sy - 10;
                const cx2 = bx - 20;
                const cy2 = by + 15;
                pathBtn.setAttribute('d', `M ${sx.toFixed(1)},${sy.toFixed(1)} C ${cx1.toFixed(1)},${cy1.toFixed(1)} ${cx2.toFixed(1)},${cy2.toFixed(1)} ${bx.toFixed(1)},${by.toFixed(1)}`);
                dotBtn.setAttribute('cx', sx.toFixed(1));
                dotBtn.setAttribute('cy', sy.toFixed(1));
              }
            } else {
              elCalloutBtn.style.display = 'none';
              if (pathBtn) pathBtn.style.display = 'none';
              if (dotBtn) dotBtn.style.display = 'none';
            }
          } else {
            elCalloutBtn.style.display = 'none';
            if (pathBtn) pathBtn.style.display = 'none';
            if (dotBtn) dotBtn.style.display = 'none';
          }
        }

        // 2. Quadrupled Autograft (With Direct Smooth Curve)
        const elCalloutGraft = document.getElementById('callout-graft');
        const pathGraft = document.getElementById('curve-path-graft');
        const dotGraft = document.getElementById('curve-dot-graft');
        if (elCalloutGraft) {
          if (reconstructedGraft && reconstructedGraft.visible && (currentProcedureStage === 3 || currentProcedureStage === 4)) {
            tmpVec.set(-0.05, 0.05, 0.12);
            kneeGroup.localToWorld(tmpVec);
            tmpVec.project(camera);
            if (tmpVec.z <= 1.0) {
              const sx = (tmpVec.x * 0.5 + 0.5) * w;
              const sy = (-(tmpVec.y * 0.5) + 0.5) * h;
              const bx = sx - 110;
              const by = sy - 30;
              elCalloutGraft.style.display = 'flex';
              elCalloutGraft.style.left = `${bx}px`;
              elCalloutGraft.style.top = `${by}px`;
              if (pathGraft && dotGraft) {
                pathGraft.style.display = 'block';
                dotGraft.style.display = 'block';
                const cx1 = sx - 35;
                const cy1 = sy;
                const cx2 = bx + 25;
                const cy2 = by + 15;
                pathGraft.setAttribute('d', `M ${sx.toFixed(1)},${sy.toFixed(1)} C ${cx1.toFixed(1)},${cy1.toFixed(1)} ${cx2.toFixed(1)},${cy2.toFixed(1)} ${bx.toFixed(1)},${by.toFixed(1)}`);
                dotGraft.setAttribute('cx', sx.toFixed(1));
                dotGraft.setAttribute('cy', sy.toFixed(1));
              }
            } else {
              elCalloutGraft.style.display = 'none';
              if (pathGraft) pathGraft.style.display = 'none';
              if (dotGraft) dotGraft.style.display = 'none';
            }
          } else {
            elCalloutGraft.style.display = 'none';
            if (pathGraft) pathGraft.style.display = 'none';
            if (dotGraft) dotGraft.style.display = 'none';
          }
        }

        // 3. Diatom Biosilica Interference Anchor (With Direct Smooth Curve)
        const elCalloutAnchor = document.getElementById('callout-anchor');
        const pathAnchor = document.getElementById('curve-path-anchor');
        const dotAnchor = document.getElementById('curve-dot-anchor');
        if (elCalloutAnchor) {
          if (diatomAnchorGroup && diatomAnchorGroup.visible && currentProcedureStage === 4) {
            diatomAnchorGroup.getWorldPosition(tmpVec);
            tmpVec.project(camera);
            if (tmpVec.z <= 1.0) {
              const sx = (tmpVec.x * 0.5 + 0.5) * w;
              const sy = (-(tmpVec.y * 0.5) + 0.5) * h;
              const bx = sx + 95;
              const by = sy + 30;
              elCalloutAnchor.style.display = 'flex';
              elCalloutAnchor.style.left = `${bx}px`;
              elCalloutAnchor.style.top = `${by}px`;
              if (pathAnchor && dotAnchor) {
                pathAnchor.style.display = 'block';
                dotAnchor.style.display = 'block';
                const cx1 = sx + 30;
                const cy1 = sy + 15;
                const cx2 = bx - 25;
                const cy2 = by - 10;
                pathAnchor.setAttribute('d', `M ${sx.toFixed(1)},${sy.toFixed(1)} C ${cx1.toFixed(1)},${cy1.toFixed(1)} ${cx2.toFixed(1)},${cy2.toFixed(1)} ${bx.toFixed(1)},${by.toFixed(1)}`);
                dotAnchor.setAttribute('cx', sx.toFixed(1));
                dotAnchor.setAttribute('cy', sy.toFixed(1));
              }
            } else {
              elCalloutAnchor.style.display = 'none';
              if (pathAnchor) pathAnchor.style.display = 'none';
              if (dotAnchor) dotAnchor.style.display = 'none';
            }
          } else {
            elCalloutAnchor.style.display = 'none';
            if (pathAnchor) pathAnchor.style.display = 'none';
            if (dotAnchor) dotAnchor.style.display = 'none';
          }
        }

        // 4. Arthroscopic Portals (With Direct Smooth Curve)
        const elCalloutPortals = document.getElementById('callout-portals');
        const pathPortals = document.getElementById('curve-path-portals');
        const dotPortals = document.getElementById('curve-dot-portals');
        if (elCalloutPortals) {
          if (portalGroup && portalGroup.visible && currentProcedureStage === 1) {
            tmpVec.set(0.0, -0.15, 1.4);
            kneeGroup.localToWorld(tmpVec);
            tmpVec.project(camera);
            if (tmpVec.z <= 1.0) {
              const sx = (tmpVec.x * 0.5 + 0.5) * w;
              const sy = (-(tmpVec.y * 0.5) + 0.5) * h;
              const bx = sx + 85;
              const by = sy - 25;
              elCalloutPortals.style.display = 'flex';
              elCalloutPortals.style.left = `${bx}px`;
              elCalloutPortals.style.top = `${by}px`;
              if (pathPortals && dotPortals) {
                pathPortals.style.display = 'block';
                dotPortals.style.display = 'block';
                const cx1 = sx + 25;
                const cy1 = sy - 15;
                const cx2 = bx - 20;
                const cy2 = by + 10;
                pathPortals.setAttribute('d', `M ${sx.toFixed(1)},${sy.toFixed(1)} C ${cx1.toFixed(1)},${cy1.toFixed(1)} ${cx2.toFixed(1)},${cy2.toFixed(1)} ${bx.toFixed(1)},${by.toFixed(1)}`);
                dotPortals.setAttribute('cx', sx.toFixed(1));
                dotPortals.setAttribute('cy', sy.toFixed(1));
              }
            } else {
              elCalloutPortals.style.display = 'none';
              if (pathPortals) pathPortals.style.display = 'none';
              if (dotPortals) dotPortals.style.display = 'none';
            }
          } else {
            elCalloutPortals.style.display = 'none';
            if (pathPortals) pathPortals.style.display = 'none';
            if (dotPortals) dotPortals.style.display = 'none';
          }
        }
      } else if (calloutsContainer) {
        calloutsContainer.style.display = 'none';
      }

      if (renderer && scene && camera) {
        renderer.render(scene, camera);
      }
    }

    if (controls) {
      controls.addEventListener('change', requestRender);
    }

    function gsapAnimateCam(tx, ty, tz) {
      const startPos = camera.position.clone();
      const endPos = new THREE.Vector3(tx, ty, tz);
      let progress = 0;
      isAnimating = true;

      function animStep() {
        if (document.hidden) {
          camera.position.copy(endPos);
          isAnimating = false;
          render();
          return;
        }
        progress += 0.05;
        camera.position.lerpVectors(startPos, endPos, Math.min(progress, 1.0));
        render();
        if (progress < 1.0) {
          requestAnimationFrame(animStep);
        } else {
          isAnimating = false;
        }
      }
      animStep();
    }

    document.addEventListener('visibilitychange', () => {
      if (!document.hidden) requestRender();
    });

    function resizeRenderer() {
      const w = container.clientWidth || window.innerWidth || 800;
      const h = container.clientHeight || window.innerHeight || 600;
      camera.aspect = w / h;
      camera.updateProjectionMatrix();
      renderer.setSize(w, h);
      render();
    }

    window.addEventListener('resize', resizeRenderer);

    // Initial plane setup and multiple layout kicks for iframe reliability
    updateSlicingPlane();
    resizeRenderer();
    setTimeout(resizeRenderer, 60);
    setTimeout(resizeRenderer, 300);
    setTimeout(render, 500);
    // Dismiss Skeleton Splash Screen smoothly
    setTimeout(() => {
      const splash = document.getElementById('pocketgull-skeleton-splash');
      if (splash) {
        splash.style.opacity = '0';
        splash.style.visibility = 'hidden';
        setTimeout(() => splash.remove(), 450);
      }
    }, 650);
  </script>
</body>
</html>"""

display(HTML(f'''
<div id="knee-sim-wrapper" style="position: relative; margin-bottom: 24px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;">
  <!-- 3D WebGL Specimen Bench Container -->
  <div id="knee-iframe-box" style="width: 100%; height: 720px; border-radius: 4px 4px 0 0; overflow: hidden; border: 1px solid #1e293b; border-bottom: none; box-shadow: 0 16px 40px rgba(5,8,14,0.8); background: #05080e; position: relative;">
    <iframe id="knee-3d-iframe" srcdoc="{html_content.replace('"', '&quot;')}" style="width: 100%; height: 100%; border: none;" allowfullscreen="true" webkitallowfullscreen="true" mozallowfullscreen="true" allow="fullscreen; accelerometer; gyroscope"></iframe>
  </div>

  <!-- PocketShoal Console Control Dock (NO PILLS, Crisp 3px Corners, Inset Tray) -->
  <div style="background: #090d16; border: 1px solid #1e293b; border-radius: 0 0 4px 4px; padding: 12px 18px; display: flex; align-items: center; justify-content: space-between; flex-wrap: wrap; gap: 12px; font-family: monospace; font-size: 12px;">
    <div style="display: flex; align-items: center; gap: 10px;">
      <span style="font-weight: 700; color: #38bdf8; background: #131d2e; border: 1px solid #1e293b; padding: 3px 8px; border-radius: 2px;">3.0T MRI</span>
      <span style="color: #94a3b8; font-size: 11.5px;">Superconducting Cryo-Gantry & 16-Channel Knee RF Coil active in 3D scene</span>
    </div>
    
    <div style="display: flex; align-items: center; flex-wrap: wrap; gap: 8px;">
      <!-- Native Outer Fullscreen Trigger (Crisp 3px Button) -->
      <button onclick="(function(){{
        var box = document.getElementById('knee-iframe-box');
        if (!document.fullscreenElement) {{
          if (box.requestFullscreen) box.requestFullscreen();
          else if (box.webkitRequestFullscreen) box.webkitRequestFullscreen();
          else if (box.mozRequestFullScreen) box.mozRequestFullScreen();
          else if (box.msRequestFullscreen) box.msRequestFullscreen();
        }} else {{
          if (document.exitFullscreen) document.exitFullscreen();
        }}
      }})()" style="display: inline-flex; align-items: center; gap: 6px; padding: 6px 12px; border-radius: 3px; background: #131d2e; border: 1px solid #334155; color: #38bdf8; font-size: 11.5px; font-weight: 700; cursor: pointer; font-family: monospace; transition: all 0.15s ease;">
        <span>[⛶ FULLSCREEN BENCH]</span>
      </button>

      <!-- Direct Jump Link to PyTorch Code (The one just before 3) -->
      <a href="#section-2-depth-module" style="display: inline-flex; align-items: center; gap: 6px; padding: 6px 14px; border-radius: 3px; background: #0c4a6e; border: 1px solid #0284c7; color: #ffffff; text-decoration: none; font-size: 11.5px; font-weight: 700; font-family: monospace; box-shadow: 0 0 10px rgba(2,132,199,0.3); transition: all 0.15s ease;">
        <span>[02. PYTORCH DEPTH MODULE ↓]</span>
      </a>

      <!-- Direct Jump Link to Section 3 Matrix -->
      <a href="#section-3-matrix" style="display: inline-flex; align-items: center; gap: 5px; padding: 6px 12px; border-radius: 3px; background: #131d2e; border: 1px solid #334155; color: #f8fafc; text-decoration: none; font-size: 11.5px; font-weight: 600; font-family: monospace;">
        <span>[03. CO-OCCURRENCE ↓]</span>
      </a>
    </div>
  </div>
</div>
'''))
```

  </div>
  <div style="background: #05080e; border: 1px solid #1e293b; border-radius: 2px; padding: 12px; text-align: center; font-size: 16px; color: #38bdf8;">

  </div>
  <div style="font-size: 12px; color: #94a3b8; margin: 0 0 10px 0;">

  </div>
  <div style="background: #05080e; border: 1px solid #1e293b; border-radius: 2px; padding: 12px; text-align: center; font-size: 15px; color: #6ee7b7;">
    $$\tilde{z}_i^{\text{canon}} = \begin{cases} \tilde{z}_i & \text{if Right Knee} \\ 1.0 - \tilde{z}_i & \text{if Left Knee} \end{cases}$$
  </div>

<div style="font-family: monospace; font-size: 11.5px; color: #64748b; background: #090d16; border-radius: 2px; padding: 10px 14px; border: 1px solid #1e293b;">
  [SPECIMEN] Drop-In PyTorch Module compiled below with learned Fourier position embeddings and laterality flags.
</div>

```python
# =====================================================================
# DROP-IN PYTORCH MODULE: 3D Continuous Depth Embedding & Laterality
# =====================================================================
import torch
import torch.nn as nn

class Anatomical3DDepthEncoder(nn.Module):
    """
    Encodes continuous 1D spatial depth z in [0, 1] into a high-dimensional
    anatomical coordinate embedding and injects it into slice token representations.
    """
    def __init__(self, embed_dim: int = 512, hidden_dim: int = 64):
        super().__init__()
        self.depth_mlp = nn.Sequential(
            nn.Linear(1, hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, embed_dim),
            nn.LayerNorm(embed_dim)
        )
        
    def forward(self, slice_tokens: torch.Tensor, depths: torch.Tensor) -> torch.Tensor:
        """
        slice_tokens: [Batch, N_slices, Embed_dim]
        depths:       [Batch, N_slices] (continuous float in [0.0, 1.0])
        """
        depth_emb = self.depth_mlp(depths.unsqueeze(-1))  # [Batch, N_slices, Embed_dim]
        return slice_tokens + depth_emb


def compute_normalized_depth(iop: list[float], ipp_list: list[list[float]]) -> np.ndarray:
    """
    Given ImageOrientationPatient (6 floats) and a list of ImagePositionPatient (3 floats each),
    returns the normalized 1D slice depth array in [0.0, 1.0].
    """
    rx, ry, rz, cx, cy, cz = iop[:6]
    nx = ry * cz - rz * cy
    ny = rz * cx - rx * cz
    nz = rx * cy - ry * cx
    norm = math.sqrt(nx*nx + ny*ny + nz*nz) or 1.0
    normal = np.array([nx/norm, ny/norm, nz/norm])
    
    raw_depths = np.array([np.dot(normal, p) for p in ipp_list])
    span = raw_depths.max() - raw_depths.min()
    if span < 1e-4:
        return np.linspace(0.0, 1.0, len(ipp_list), dtype=np.float32)
    return ((raw_depths - raw_depths.min()) / span).astype(np.float32)

print("Anatomical3DDepthEncoder module ready to drop into your PyTorch models!")
```

```python
# Embed the Salutogenic Cinema of Hills & 12 Knee Quests
html_cinema_content = """<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0, maximum-scale=1.0, user-scalable=no">
  <title>Knee Slapper: Cinema of Hills | The 12 Knee Quests & Knee Rider</title>
  <style>
    /* Direct Curves & Glassmorphism Global Reset */
    *, *::before, *::after {
      box-sizing: border-box;
      margin: 0;
      padding: 0;
      -webkit-tap-highlight-color: transparent;
      user-select: none;
    }

    :root {
      --emerald-glow: #10b981;
      --emerald-dark: #064e3b;
      --spring-green: #34d399;
      --golden-amber: #f59e0b;
      --sun-rose: #f43f5e;
      --sky-top: #0f172a;
      --sky-bottom: #1e293b;
      --glass-bg: rgba(15, 23, 42, 0.72);
      --glass-border: rgba(255, 255, 255, 0.14);
      --text-main: #f8fafc;
      --text-muted: #94a3b8;
      --squircle: 28px;
    }

    body {
      background: #020617;
      color: var(--text-main);
      font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, "Helvetica Neue", Arial, sans-serif;
      overflow: hidden;
      width: 100vw;
      height: 100vh;
      touch-action: none;
      transition: background 0.5s ease;
    }

    body.synthwave-mode {
      background: #05050f;
    }

    /* Fullscreen High-DPI Canvas */
    #game-canvas {
      position: absolute;
      top: 0;
      left: 0;
      width: 100%;
      height: 100%;
      display: block;
      z-index: 1;
    }

    /* Floating UI Container */
    .hud-layer {
      position: absolute;
      inset: 0;
      z-index: 10;
      pointer-events: none;
      display: flex;
      flex-direction: column;
      justify-content: space-between;
      padding: 14px 20px;
    }

    /* Direct Curve Pills */
    .pill-btn, .pill-badge {
      pointer-events: auto;
      display: inline-flex;
      align-items: center;
      gap: 7px;
      padding: 8px 16px;
      border-radius: 9999px;
      background: var(--glass-bg);
      backdrop-filter: blur(20px);
      -webkit-backdrop-filter: blur(20px);
      border: 1px solid var(--glass-border);
      color: var(--text-main);
      font-size: 13px;
      font-weight: 600;
      letter-spacing: 0.25px;
      cursor: pointer;
      box-shadow: 0 8px 32px rgba(0, 0, 0, 0.35);
      transition: all 0.25s cubic-bezier(0.16, 1, 0.3, 1);
    }

    .pill-btn:hover {
      transform: translateY(-2px);
      background: rgba(30, 41, 59, 0.90);
      border-color: rgba(52, 211, 153, 0.4);
      box-shadow: 0 12px 36px rgba(16, 185, 129, 0.25);
    }

    .pill-btn:active {
      transform: scale(0.96) translateY(0);
    }

    .pill-btn.active {
      background: linear-gradient(135deg, #10b981 0%, #059669 100%);
      color: #ffffff;
      border-color: rgba(255, 255, 255, 0.3);
      box-shadow: 0 0 24px rgba(16, 185, 129, 0.5);
    }

    /* Knee Rider Red Neon Theme */
    .pill-btn.rider-active {
      background: linear-gradient(135deg, #dc2626 0%, #991b1b 100%) !important;
      color: #ffffff !important;
      border-color: rgba(248, 113, 113, 0.6) !important;
      box-shadow: 0 0 28px rgba(239, 68, 68, 0.65) !important;
    }

    .pill-badge {
      cursor: default;
      font-variant-numeric: tabular-nums;
    }

    /* Green, Yellow, Red Clinical Difficulty Tiers */
    .tier-badge {
      display: inline-flex;
      align-items: center;
      gap: 5px;
      padding: 3px 10px;
      border-radius: 9999px;
      font-size: 11px;
      font-weight: 700;
      text-transform: uppercase;
      letter-spacing: 0.6px;
    }

    .tier-green {
      background: rgba(16, 185, 129, 0.2);
      color: #34d399;
      border: 1px solid rgba(52, 211, 153, 0.45);
    }

    .tier-yellow {
      background: rgba(245, 158, 11, 0.2);
      color: #facc15;
      border: 1px solid rgba(250, 204, 21, 0.45);
    }

    .tier-red {
      background: rgba(239, 68, 68, 0.2);
      color: #f87171;
      border: 1px solid rgba(248, 113, 113, 0.45);
    }

    /* Top Navigation Bar */
    .top-bar {
      display: flex;
      align-items: center;
      justify-content: space-between;
      width: 100%;
      max-width: 1480px;
      margin: 0 auto;
      gap: 10px;
      flex-wrap: wrap;
    }

    .nav-group {
      display: flex;
      align-items: center;
      gap: 8px;
    }

    .brand-pill {
      background: linear-gradient(135deg, rgba(16, 185, 129, 0.25) 0%, rgba(6, 78, 59, 0.4) 100%);
      border: 1px solid rgba(52, 211, 153, 0.35);
    }

    .brand-pill.rider {
      background: linear-gradient(135deg, rgba(220, 38, 38, 0.3) 0%, rgba(15, 23, 42, 0.6) 100%);
      border: 1px solid rgba(239, 68, 68, 0.45);
    }

    .brand-icon {
      font-size: 18px;
      display: inline-block;
      animation: gentle-wobble 3s infinite ease-in-out;
    }

    @keyframes gentle-wobble {
      0%, 100% { transform: rotate(0deg) scale(1); }
      25% { transform: rotate(-6deg) scale(1.08); }
      75% { transform: rotate(6deg) scale(1.08); }
    }

    /* Telemetry Deck */
    .telemetry-deck {
      display: flex;
      align-items: center;
      gap: 8px;
    }

    .metric-value {
      color: #34d399;
      font-weight: 700;
    }

    .metric-value.rider {
      color: #f87171;
    }

    /* Bottom Control Dock */
    .bottom-dock {
      display: flex;
      align-items: center;
      justify-content: space-between;
      width: 100%;
      max-width: 1480px;
      margin: 0 auto;
      gap: 12px;
    }

    .hint-capsule {
      font-size: 12px;
      color: var(--text-muted);
      background: rgba(15, 23, 42, 0.65);
      padding: 8px 18px;
      border-radius: 9999px;
      border: 1px solid rgba(255, 255, 255, 0.08);
      backdrop-filter: blur(16px);
      -webkit-backdrop-filter: blur(16px);
    }

    .hint-key {
      display: inline-block;
      padding: 2px 7px;
      background: rgba(255, 255, 255, 0.15);
      border-radius: 6px;
      color: #fff;
      font-weight: 700;
      margin: 0 3px;
    }

    /* Speech Bubble / Knee Slapper Card Overlay */
    #joke-banner {
      position: absolute;
      top: 80px;
      left: 50%;
      transform: translateX(-50%) translateY(-20px);
      max-width: 600px;
      width: calc(100% - 40px);
      background: rgba(15, 23, 42, 0.92);
      backdrop-filter: blur(28px);
      -webkit-backdrop-filter: blur(28px);
      border: 1.5px solid rgba(52, 211, 153, 0.4);
      border-radius: var(--squircle);
      padding: 18px 24px;
      box-shadow: 0 20px 60px rgba(0, 0, 0, 0.7), 0 0 40px rgba(16, 185, 129, 0.2);
      z-index: 50;
      pointer-events: auto;
      opacity: 0;
      visibility: hidden;
      transition: all 0.4s cubic-bezier(0.16, 1, 0.3, 1);
    }

    #joke-banner.rider {
      background: rgba(15, 23, 42, 0.95);
      border-color: rgba(239, 68, 68, 0.55);
      box-shadow: 0 20px 60px rgba(0, 0, 0, 0.8), 0 0 50px rgba(239, 68, 68, 0.35);
    }

    #joke-banner.show {
      opacity: 1;
      visibility: visible;
      transform: translateX(-50%) translateY(0);
    }

    .joke-header {
      display: flex;
      align-items: center;
      justify-content: space-between;
      margin-bottom: 10px;
    }

    .joke-badge {
      display: inline-flex;
      align-items: center;
      gap: 6px;
      font-size: 11px;
      font-weight: 700;
      text-transform: uppercase;
      letter-spacing: 0.8px;
      color: #34d399;
      background: rgba(16, 185, 129, 0.15);
      padding: 4px 12px;
      border-radius: 9999px;
    }

    .joke-badge.rider {
      color: #f87171;
      background: rgba(220, 38, 38, 0.2);
    }

    /* K.N.E.E. 2000 Voice Box (K.I.T.T. 3-bar Audio Visualizer) */
    .kitt-voice-box {
      display: none;
      align-items: center;
      gap: 3px;
      height: 18px;
    }

    .kitt-voice-box.active {
      display: inline-flex;
    }

    .kitt-bar {
      width: 4px;
      height: 100%;
      background: #ef4444;
      border-radius: 2px;
      box-shadow: 0 0 8px #ef4444;
      animation: kitt-bounce 0.6s infinite ease-in-out alternate;
    }

    .kitt-bar:nth-child(2) { animation-delay: 0.2s; height: 75%; }
    .kitt-bar:nth-child(3) { animation-delay: 0.4s; height: 50%; }

    @keyframes kitt-bounce {
      0% { transform: scaleY(0.3); opacity: 0.5; }
      100% { transform: scaleY(1.0); opacity: 1; }
    }

    .joke-close {
      background: none;
      border: none;
      color: var(--text-muted);
      cursor: pointer;
      font-size: 18px;
      line-height: 1;
      padding: 4px;
      border-radius: 50%;
      transition: color 0.2s;
    }

    .joke-close:hover {
      color: #fff;
    }

    .joke-setup {
      font-size: 15px;
      font-weight: 600;
      color: #f1f5f9;
      line-height: 1.45;
      margin-bottom: 8px;
    }

    .joke-punchline {
      font-size: 16px;
      font-weight: 700;
      color: #34d399;
      line-height: 1.45;
      padding: 10px 14px;
      background: rgba(16, 185, 129, 0.08);
      border-radius: 16px;
      border-left: 3px solid #10b981;
    }

    .joke-punchline.rider {
      color: #f87171;
      background: rgba(239, 68, 68, 0.1);
      border-left-color: #ef4444;
    }

    .joke-actions {
      display: flex;
      align-items: center;
      justify-content: flex-end;
      gap: 10px;
      margin-top: 12px;
    }

    .joke-btn {
      padding: 6px 14px;
      font-size: 12px;
      font-weight: 600;
      border-radius: 9999px;
      background: rgba(255, 255, 255, 0.08);
      border: 1px solid rgba(255, 255, 255, 0.1);
      color: #e2e8f0;
      cursor: pointer;
      transition: all 0.2s;
    }

    .joke-btn:hover {
      background: rgba(255, 255, 255, 0.18);
      color: #fff;
    }

    /* Modal Windows */
    .modal-overlay {
      position: absolute;
      inset: 0;
      background: rgba(2, 6, 23, 0.78);
      backdrop-filter: blur(14px);
      -webkit-backdrop-filter: blur(14px);
      z-index: 100;
      display: flex;
      align-items: center;
      justify-content: center;
      padding: 20px;
      opacity: 0;
      visibility: hidden;
      transition: all 0.3s cubic-bezier(0.16, 1, 0.3, 1);
    }

    .modal-overlay.open {
      opacity: 1;
      visibility: visible;
    }

    .modal-card {
      width: 100%;
      max-width: 680px;
      max-height: 88vh;
      display: flex;
      flex-direction: column;
      background: #0f172a;
      border: 1.5px solid rgba(52, 211, 153, 0.3);
      border-radius: var(--squircle);
      box-shadow: 0 25px 80px rgba(0, 0, 0, 0.8), 0 0 50px rgba(16, 185, 129, 0.15);
      overflow: hidden;
      transform: scale(0.95);
      transition: transform 0.3s cubic-bezier(0.16, 1, 0.3, 1);
    }

    .modal-overlay.open .modal-card {
      transform: scale(1);
    }

    .modal-header {
      padding: 18px 24px;
      display: flex;
      align-items: center;
      justify-content: space-between;
      border-bottom: 1px solid rgba(255, 255, 255, 0.08);
      background: rgba(15, 23, 42, 0.95);
    }

    .modal-title {
      font-size: 17px;
      font-weight: 700;
      display: flex;
      align-items: center;
      gap: 10px;
    }

    .modal-body {
      padding: 20px 24px;
      overflow-y: auto;
      flex: 1;
    }

    /* 12 Knee Quests Level Cards Grid */
    .quests-filter-bar {
      display: flex;
      align-items: center;
      gap: 8px;
      margin-bottom: 16px;
      overflow-x: auto;
      padding-bottom: 4px;
    }

    .filter-tab {
      padding: 6px 14px;
      border-radius: 9999px;
      font-size: 12px;
      font-weight: 700;
      background: rgba(255, 255, 255, 0.06);
      border: 1px solid rgba(255, 255, 255, 0.1);
      color: var(--text-muted);
      cursor: pointer;
      transition: all 0.2s;
      white-space: nowrap;
    }

    .filter-tab:hover, .filter-tab.active {
      background: rgba(16, 185, 129, 0.2);
      border-color: rgba(52, 211, 153, 0.4);
      color: #34d399;
    }

    .quest-card {
      background: rgba(30, 41, 59, 0.55);
      border: 1px solid rgba(255, 255, 255, 0.08);
      border-radius: 20px;
      padding: 16px 18px;
      margin-bottom: 12px;
      transition: all 0.2s;
      cursor: pointer;
    }

    .quest-card:hover {
      background: rgba(30, 41, 59, 0.9);
      border-color: rgba(52, 211, 153, 0.4);
      transform: translateY(-2px);
    }

    .quest-card.tier-border-green:hover { border-color: #34d399; box-shadow: 0 8px 24px rgba(16, 185, 129, 0.2); }
    .quest-card.tier-border-yellow:hover { border-color: #facc15; box-shadow: 0 8px 24px rgba(245, 158, 11, 0.2); }
    .quest-card.tier-border-red:hover { border-color: #f87171; box-shadow: 0 8px 24px rgba(239, 68, 68, 0.2); }

    .quest-card-header {
      display: flex;
      align-items: center;
      justify-content: space-between;
      margin-bottom: 8px;
    }

    .quest-num {
      font-size: 15px;
      font-weight: 800;
      color: #fff;
    }

    .quest-metaphor {
      font-size: 13px;
      font-weight: 700;
      color: #38bdf8;
      margin-bottom: 4px;
    }

    .quest-desc {
      font-size: 13px;
      color: #cbd5e1;
      line-height: 1.4;
      margin-bottom: 10px;
    }

    .quest-footer {
      display: flex;
      align-items: center;
      justify-content: space-between;
      border-top: 1px solid rgba(255, 255, 255, 0.06);
      padding-top: 8px;
      font-size: 12px;
      color: var(--text-muted);
    }

    /* Shift Roster List */
    .roster-item {
      display: flex;
      align-items: center;
      justify-content: space-between;
      padding: 14px 18px;
      background: rgba(30, 41, 59, 0.5);
      border: 1px solid rgba(255, 255, 255, 0.06);
      border-radius: 20px;
      margin-bottom: 10px;
      transition: all 0.2s;
    }

    .roster-item:hover {
      background: rgba(30, 41, 59, 0.85);
      border-color: rgba(52, 211, 153, 0.3);
      transform: translateX(4px);
    }

    .roster-left {
      display: flex;
      align-items: center;
      gap: 14px;
    }

    .roster-rank {
      font-size: 18px;
      font-weight: 800;
      color: #34d399;
      width: 28px;
      text-align: center;
    }

    .roster-name {
      font-weight: 700;
      font-size: 15px;
      color: #f8fafc;
    }

    .roster-role {
      font-size: 12px;
      color: var(--text-muted);
      margin-top: 2px;
    }

    .roster-note {
      font-size: 12px;
      color: #38bdf8;
      font-style: italic;
      margin-top: 4px;
    }

    .roster-stat {
      text-align: right;
    }

    .roster-distance {
      font-size: 16px;
      font-weight: 800;
      color: #facc15;
      font-variant-numeric: tabular-nums;
    }

    .roster-slaps {
      font-size: 11px;
      color: var(--text-muted);
    }

    /* Input Form for Logging a Shift */
    .log-form {
      margin-top: 18px;
      padding: 16px;
      background: rgba(16, 185, 129, 0.06);
      border: 1px dashed rgba(52, 211, 153, 0.3);
      border-radius: 20px;
    }

    .form-row {
      display: flex;
      gap: 10px;
      margin-bottom: 10px;
    }

    .form-input {
      flex: 1;
      padding: 10px 14px;
      background: rgba(15, 23, 42, 0.8);
      border: 1px solid rgba(255, 255, 255, 0.15);
      border-radius: 12px;
      color: #fff;
      font-size: 13px;
      outline: none;
    }

    .form-input:focus {
      border-color: #34d399;
    }

    .form-select {
      padding: 10px 14px;
      background: #0f172a;
      border: 1px solid rgba(255, 255, 255, 0.15);
      border-radius: 12px;
      color: #fff;
      font-size: 13px;
      outline: none;
    }

    /* Screen-Slap Flash Effect */
    #slap-flash {
      position: absolute;
      inset: 0;
      background: rgba(255, 255, 255, 0.35);
      z-index: 20;
      pointer-events: none;
      opacity: 0;
      transition: opacity 0.15s ease-out;
    }

    #slap-flash.active {
      opacity: 1;
    }

    /* Mobile Floating Action Button (Tap To Slap / Dive) */
    .mobile-slap-zone {
      display: none;
    }

    @media (max-width: 768px) {
      .hud-layer {
        padding: 10px 12px;
      }
      .telemetry-deck .pill-badge:nth-child(2),
      .telemetry-deck .pill-badge:nth-child(3) {
        display: none;
      }
      .hint-capsule {
        display: none;
      }
      .mobile-slap-zone {
        display: flex;
        align-items: center;
        justify-content: center;
        width: 64px;
        height: 64px;
        border-radius: 50%;
        background: linear-gradient(135deg, #10b981, #059669);
        box-shadow: 0 0 30px rgba(16, 185, 129, 0.6);
        color: #fff;
        font-size: 26px;
        pointer-events: auto;
        cursor: pointer;
        position: absolute;
        bottom: 24px;
        right: 24px;
      }
      .mobile-slap-zone.rider {
        background: linear-gradient(135deg, #ef4444, #991b1b);
        box-shadow: 0 0 30px rgba(239, 68, 68, 0.7);
      }
    }
  </style>
</head>
<body>

  <!-- Fullscreen Canvas -->
  <canvas id="game-canvas"></canvas>

  <!-- Slap Flash Overlay -->
  <div id="slap-flash"></div>

  <!-- HUD Interface Layer -->
  <div class="hud-layer">
    <!-- Top Navigation Bar -->
    <header class="top-bar">
      <div class="nav-group">
        <div id="brand-pill" class="pill-badge brand-pill">
          <span id="brand-icon" class="brand-icon">🌱</span>
          <span id="brand-title">Patty's Knee Slapper</span>
        </div>

        <!-- 🏥 THE 12 KNEE QUESTS (Waiting Room Patient Level System) -->
        <button id="btn-quests" class="pill-btn active" title="Open The 12 Knee Quests (Waiting Room Patient Education & Levels)">
          <span>🏥</span>
          <span id="quests-btn-text">12 Knee Quests</span>
        </button>

        <!-- HERO SWITCHER: Happy Knee vs KNEE RIDER -->
        <button id="btn-hero" class="pill-btn" title="Switch Hero Mode: Green Happy Knee ⟷ KNEE RIDER (K.N.E.E. 2000)">
          <span id="hero-icon">🏎️</span>
          <span id="hero-text">KNEE RIDER</span>
        </button>

        <button id="btn-mode" class="pill-btn" title="Toggle Auto-Cinema Screensaver vs Interactive Shift Play">
          <span id="mode-icon">🎬</span>
          <span id="mode-text">Auto-Cinema</span>
        </button>

        <!-- Pacing / Speed Control (Waiting Room Gentle Mode) -->
        <button id="btn-pacing" class="pill-btn" title="Adjust Gliding Tempo: Serene (Calm) ⟷ Flow ⟷ Swift">
          <span id="pacing-icon">🐢</span>
          <span id="pacing-text">Serene (0.6x)</span>
        </button>

        <button id="btn-chrono" class="pill-btn" title="Circadian Sky Light (Dew / Sun / Dusk / PACS Midnight)">
          <span id="chrono-icon">☀️</span>
          <span id="chrono-text">Sunlight</span>
        </button>
      </div>

      <!-- Telemetry Deck -->
      <div class="telemetry-deck">
        <!-- Active Quest Target Indicator -->
        <div id="quest-hud-pill" class="pill-badge" style="display: none; background: rgba(16, 185, 129, 0.2); border-color: rgba(52, 211, 153, 0.4);">
          <span id="quest-hud-icon">🎯</span>
          <span id="quest-hud-label" class="metric-value">Quest Goal: 0/8</span>
        </div>

        <div class="pill-badge">
          <span>Glide:</span>
          <span id="stat-dist" class="metric-value">0 m</span>
        </div>
        <div class="pill-badge">
          <span>Knee-Slaps:</span>
          <span id="stat-slaps" class="metric-value">0</span>
        </div>
        <div class="pill-badge">
          <span id="stat-special-label">Harmony:</span>
          <span id="stat-special-val" class="metric-value">100%</span>
        </div>
      </div>

      <!-- Right Utility Actions -->
      <div class="nav-group">
        <!-- Gamepad Indicator Pill -->
        <div id="gamepad-pill" class="pill-badge" style="display: none; background: rgba(59, 130, 246, 0.2); border-color: rgba(96, 165, 250, 0.4);">
          <span>🎮</span>
          <span id="gamepad-name">Gamepad</span>
        </div>

        <button id="btn-sound" class="pill-btn" title="Synthesized Web Audio (Zen Chimes / 80s Synth Bass / Rimshot)">
          <span id="sound-icon">🔔</span>
          <span id="sound-text">Sound: On</span>
        </button>

        <button id="btn-roster" class="pill-btn" title="Collaborative Shift Roster & Breakroom Streaks">
          <span>📋</span>
          <span>Shift Board</span>
        </button>

        <button id="btn-jokes" class="pill-btn" title="View Unlocked Knee Slapper Joke Book">
          <span>📖</span>
          <span>Jokes</span>
        </button>
      </div>
    </header>

    <!-- Bottom Dock -->
    <footer class="bottom-dock">
      <div id="hint-capsule" class="hint-capsule">
        <span>🎮</span>
        <span id="hint-text">Hold <span class="hint-key">SPACE</span> / <span class="hint-key">CLICK</span> / Gamepad <span class="hint-key">A</span> to dive slopes &bull; Release to launch! Click Patty anytime for a <b>Knee Slapper</b>!</span>
      </div>

      <div class="nav-group">
        <!-- Turbo Boost Action Button -->
        <button id="btn-turbo" class="pill-btn" style="display: none; background: linear-gradient(135deg, #ef4444, #b91c1c); color: #fff; box-shadow: 0 0 20px rgba(239, 68, 68, 0.5);" title="Turbo Boost Extension (Shift / T key or Gamepad RT/Y)">
          <span>⚡</span>
          <span>TURBO BOOST</span>
        </button>

        <a href="interactive_3d_knee_slicer.html" class="pill-btn" title="Return to Clinical 3D Knee Anatomy & Slicer">
          <span>🔬</span>
          <span>Back to 3D Anatomy</span>
        </a>
      </div>
    </footer>
  </div>

  <!-- Mobile Floating Slap Button -->
  <div id="btn-mobile-slap" class="mobile-slap-zone" title="Tap to Slap!">
    🦵
  </div>

  <!-- Knee-Slapper Joke Popout Card -->
  <div id="joke-banner">
    <div class="joke-header">
      <div style="display: flex; align-items: center; gap: 8px;">
        <span id="joke-badge" class="joke-badge">⚡ KNEE SLAPPER UNLOCKED!</span>
        <div id="kitt-voice-box" class="kitt-voice-box">
          <div class="kitt-bar"></div>
          <div class="kitt-bar"></div>
          <div class="kitt-bar"></div>
        </div>
      </div>
      <button id="joke-close-btn" class="joke-close" title="Close Joke">&times;</button>
    </div>
    <div id="joke-setup-text" class="joke-setup">Why did the patella go to school?</div>
    <div id="joke-punchline-text" class="joke-punchline">To get a little more flexible!</div>
    <div class="joke-actions">
      <button id="joke-share-btn" class="joke-btn">📋 Copy Joke</button>
      <button id="joke-next-btn" class="joke-btn">🎲 Another Slapper!</button>
    </div>
  </div>

  <!-- 🏥 THE 12 KNEE QUESTS MODAL (Waiting Room Patient Level Selector) -->
  <div id="quests-modal" class="modal-overlay">
    <div class="modal-card" style="max-width: 780px;">
      <div class="modal-header">
        <div class="modal-title">
          <span>🏥</span>
          <span>The 12 Knee Quests (Waiting Room Patient Guide)</span>
        </div>
        <button id="quests-close-btn" class="joke-close">&times;</button>
      </div>
      <div class="modal-body">
        <p style="font-size: 13px; color: var(--text-muted); margin-bottom: 14px;">
          Waiting to see your doctor or radiologist? Tap on your injury or condition below to play its dedicated quest, learn what happened in plain English, and get 3 smart questions for your visit!
        </p>

        <!-- Filter Bar: Green, Yellow, Red Tiers -->
        <div class="quests-filter-bar">
          <button class="filter-tab active" data-filter="all">All 12 Conditions</button>
          <button class="filter-tab" data-filter="green">🟢 Gentle (Green Tiers 1–4)</button>
          <button class="filter-tab" data-filter="yellow">🟡 Dynamic (Yellow Tiers 5–8)</button>
          <button class="filter-tab" data-filter="red">🔴 Advanced (Red Tiers 9–12)</button>
        </div>

        <!-- Quests Grid -->
        <div id="quests-list-container">
          <!-- Dynamically populated from QUEST_LEVELS -->
        </div>
      </div>
    </div>
  </div>

  <!-- 📋 QUEST BRIEFING & DOCTOR CONSULTATION MODAL -->
  <div id="quest-brief-modal" class="modal-overlay">
    <div class="modal-card" style="max-width: 650px;">
      <div class="modal-header">
        <div class="modal-title">
          <span id="brief-tier-pill" class="tier-badge tier-green">🟢 GENTLE</span>
          <span id="brief-modal-title">Level 1: Joint Effusion</span>
        </div>
        <button id="brief-close-btn" class="joke-close">&times;</button>
      </div>
      <div class="modal-body">
        <div style="font-size: 16px; font-weight: 700; color: #38bdf8; margin-bottom: 6px;" id="brief-metaphor">
          "The Body's Natural Airbag"
        </div>

        <div style="background: rgba(30, 41, 59, 0.6); padding: 14px; border-radius: 16px; border: 1px solid rgba(255,255,255,0.08); margin-bottom: 14px;">
          <div style="font-size: 12px; font-weight: 700; color: #34d399; text-transform: uppercase; margin-bottom: 4px;">
            💡 What Happened to My Body?
          </div>
          <p id="brief-anatomy-explain" style="font-size: 13.5px; color: #e2e8f0; line-height: 1.45;">
            <!-- Plain English explanation -->
          </p>
        </div>

        <div style="background: rgba(16, 185, 129, 0.08); padding: 14px; border-radius: 16px; border: 1px solid rgba(52, 211, 153, 0.25); margin-bottom: 14px;">
          <div style="font-size: 12px; font-weight: 700; color: #34d399; text-transform: uppercase; margin-bottom: 4px;">
            🌱 How Your Body Heals
          </div>
          <p id="brief-healing-explain" style="font-size: 13px; color: #cbd5e1; line-height: 1.45;">
            <!-- Reassuring healing science -->
          </p>
        </div>

        <div style="background: rgba(56, 189, 248, 0.08); padding: 14px; border-radius: 16px; border: 1px solid rgba(56, 189, 248, 0.25); margin-bottom: 16px;">
          <div style="font-size: 12px; font-weight: 700; color: #38bdf8; text-transform: uppercase; margin-bottom: 6px;">
            🩺 3 Smart Questions for Your Doctor Today
          </div>
          <ul id="brief-doctor-questions" style="font-size: 13px; color: #f1f5f9; padding-left: 18px; line-height: 1.5;">
            <!-- 3 tailored questions -->
          </ul>
        </div>

        <div style="display: flex; align-items: center; justify-content: space-between; gap: 10px;">
          <button id="btn-copy-doc-questions" class="pill-btn" style="background: rgba(255,255,255,0.08);">
            📋 Copy Questions to Phone
          </button>
          <button id="btn-launch-quest" class="pill-btn active" style="padding: 10px 24px; font-size: 14px;">
            ▶ Play Quest Now
          </button>
        </div>
      </div>
    </div>
  </div>

  <!-- Shift Harmony Roster Modal -->
  <div id="roster-modal" class="modal-overlay">
    <div class="modal-card">
      <div class="modal-header">
        <div class="modal-title">
          <span>📋</span>
          <span>Shift Harmony Roster & Downtime Streaks</span>
        </div>
        <button id="roster-close-btn" class="joke-close">&times;</button>
      </div>
      <div class="modal-body">
        <p style="font-size: 13px; color: var(--text-muted); margin-bottom: 16px;">
          No patients waiting on the floor or in the scanner? Leave an uplifting streak and note for the oncoming shift!
        </p>
        <div id="roster-list"></div>

        <div class="log-form">
          <div style="font-size: 13px; font-weight: 700; color: #34d399; margin-bottom: 10px;">
            ✍️ Record Your Shift Streak
          </div>
          <div class="form-row">
            <input type="text" id="input-roster-name" class="form-input" placeholder="Your Name or Initials" maxlength="24">
            <select id="select-roster-role" class="form-select">
              <option value="🩻 Radiology">🩻 Radiology</option>
              <option value="🦴 Orthopedics">🦴 Orthopedics</option>
              <option value="🩺 Nursing">🩺 Nursing</option>
              <option value="🚑 Emergency">🚑 Emergency</option>
              <option value="🧪 Lab / Pharmacy">🧪 Lab / Pharmacy</option>
              <option value="☕ Night Shift">☕ Night Shift</option>
              <option value="🏎️ Knee Rider Unit">🏎️ Knee Rider Unit</option>
              <option value="✨ Radiologist Fellow">✨ Radiologist Fellow</option>
            </select>
          </div>
          <div class="form-row">
            <input type="text" id="input-roster-note" class="form-input" placeholder="Shift handover note / positive message..." maxlength="80">
          </div>
          <div style="display: flex; justify-content: flex-end; gap: 8px; margin-top: 8px;">
            <button id="btn-save-roster" class="pill-btn active" style="padding: 8px 18px;">
              💾 Save to Breakroom Board
            </button>
          </div>
        </div>
      </div>
    </div>
  </div>

  <!-- Joke Archive Modal -->
  <div id="jokes-modal" class="modal-overlay">
    <div class="modal-card">
      <div class="modal-header">
        <div class="modal-title">
          <span>📖</span>
          <span id="jokes-modal-title">Patty's Knee-Slapper Joke Book</span>
        </div>
        <button id="jokes-close-btn" class="joke-close">&times;</button>
      </div>
      <div class="modal-body" id="jokes-book-list"></div>
    </div>
  </div>

  <script>
    /* =========================================================================
       1. SYNTHESIZED WEB AUDIO ENGINE (ZERO EXTERNAL ASSETS / ZERO TRACKING)
       ========================================================================= */
    class KneeAudioEngine {
      constructor() {
        this.ctx = null;
        this.enabled = true;
        this.synthwaveInterval = null;
        this.step = 0;
        this.isRider = false;
        this.initAudioContext();
      }

      initAudioContext() {
        if (!this.ctx && (window.AudioContext || window.webkitAudioContext)) {
          const AudioCtx = window.AudioContext || window.webkitAudioContext;
          this.ctx = new AudioCtx();
        }
      }

      resume() {
        if (this.ctx && this.ctx.state === 'suspended') {
          this.ctx.resume();
        }
      }

      setRiderMode(isRider) {
        this.isRider = isRider;
        if (isRider && this.enabled) {
          this.startSynthwaveBass();
        } else {
          this.stopSynthwaveBass();
        }
      }

      // Whimsical Launch Boing (Happy Knee)
      playLaunch(pitch = 320) {
        if (!this.enabled || !this.ctx) return;
        this.resume();
        try {
          const osc = this.ctx.createOscillator();
          const gain = this.ctx.createGain();
          osc.type = 'sine';
          osc.frequency.setValueAtTime(pitch, this.ctx.currentTime);
          osc.frequency.exponentialRampToValueAtTime(pitch * 2.2, this.ctx.currentTime + 0.28);

          gain.gain.setValueAtTime(0.2, this.ctx.currentTime);
          gain.gain.exponentialRampToValueAtTime(0.001, this.ctx.currentTime + 0.32);

          osc.connect(gain);
          gain.connect(this.ctx.destination);
          osc.start();
          osc.stop(this.ctx.currentTime + 0.33);
        } catch (e) {}
      }

      // Collectible Chime / Quest Item Chime
      playChime(freq = 528) {
        if (!this.enabled || !this.ctx) return;
        this.resume();
        try {
          const osc = this.ctx.createOscillator();
          const gain = this.ctx.createGain();
          osc.type = this.isRider ? 'sawtooth' : 'triangle';
          osc.frequency.setValueAtTime(freq, this.ctx.currentTime);
          gain.gain.setValueAtTime(0.18, this.ctx.currentTime);
          gain.gain.exponentialRampToValueAtTime(0.0001, this.ctx.currentTime + 0.45);
          osc.connect(gain);
          gain.connect(this.ctx.destination);
          osc.start();
          osc.stop(this.ctx.currentTime + 0.46);
        } catch (e) {}
      }

      // Quest Complete Fanfare
      playQuestVictory() {
        if (!this.enabled || !this.ctx) return;
        this.resume();
        const now = this.ctx.currentTime;
        const notes = [523.25, 659.25, 783.99, 1046.50]; // C5, E5, G5, C6
        notes.forEach((freq, idx) => {
          setTimeout(() => {
            this.playChime(freq);
          }, idx * 110);
        });
      }

      // KNEE RIDER Turbo Boost Turbine Jet Sound
      playTurboBoostSound() {
        if (!this.enabled || !this.ctx) return;
        this.resume();
        const now = this.ctx.currentTime;
        try {
          const osc = this.ctx.createOscillator();
          const gain = this.ctx.createGain();
          osc.type = 'sawtooth';
          osc.frequency.setValueAtTime(140, now);
          osc.frequency.exponentialRampToValueAtTime(880, now + 0.65);

          const filter = this.ctx.createBiquadFilter();
          filter.type = 'lowpass';
          filter.frequency.setValueAtTime(400, now);
          filter.frequency.exponentialRampToValueAtTime(4000, now + 0.65);
          filter.Q.setValueAtTime(6.0, now);

          gain.gain.setValueAtTime(0.3, now);
          gain.gain.exponentialRampToValueAtTime(0.001, now + 0.7);

          osc.connect(filter);
          filter.connect(gain);
          gain.connect(this.ctx.destination);
          osc.start(now);
          osc.stop(now + 0.72);
        } catch (e) {}
      }

      // KNEE RIDER Scanner Sound (K.I.T.T. whoosh-whoosh sweep)
      playScannerWhoosh() {
        if (!this.enabled || !this.ctx) return;
        this.resume();
        const now = this.ctx.currentTime;
        try {
          const bSize = this.ctx.sampleRate * 0.18;
          const nBuffer = this.ctx.createBuffer(1, bSize, this.ctx.sampleRate);
          const out = nBuffer.getChannelData(0);
          for (let i = 0; i < bSize; i++) out[i] = Math.random() * 2 - 1;
          const src = this.ctx.createBufferSource();
          src.buffer = nBuffer;

          const filter = this.ctx.createBiquadFilter();
          filter.type = 'bandpass';
          filter.frequency.setValueAtTime(600, now);
          filter.frequency.exponentialRampToValueAtTime(1800, now + 0.09);
          filter.frequency.exponentialRampToValueAtTime(500, now + 0.18);
          filter.Q.setValueAtTime(5.0, now);

          const gain = this.ctx.createGain();
          gain.gain.setValueAtTime(0.2, now);
          gain.gain.exponentialRampToValueAtTime(0.01, now + 0.18);

          src.connect(filter);
          filter.connect(gain);
          gain.connect(this.ctx.destination);
          src.start(now);
          src.stop(now + 0.19);
        } catch (e) {}
      }

      // Driving 80s Synthwave Bass Arpeggio in D Minor
      startSynthwaveBass() {
        this.stopSynthwaveBass();
        if (!this.enabled) return;
        const notes = [
          73.4, 73.4, 146.8, 73.4,
          87.3, 73.4, 98.0, 73.4,
          73.4, 73.4, 130.8, 73.4,
          110.0, 98.0, 87.3, 65.4
        ];
        this.step = 0;
        this.synthwaveInterval = setInterval(() => {
          if (!this.enabled || !this.isRider) return;
          const freq = notes[this.step % notes.length];
          this.playSynthNote(freq);

          if (this.step % 8 === 4) this.scheduleSnare(0.01);
          if (this.step % 8 === 0) this.scheduleKick(0.01);

          this.step++;
        }, 125);
      }

      stopSynthwaveBass() {
        if (this.synthwaveInterval) {
          clearInterval(this.synthwaveInterval);
          this.synthwaveInterval = null;
        }
      }

      playSynthNote(freq) {
        if (!this.ctx) return;
        this.resume();
        const now = this.ctx.currentTime;
        try {
          const osc1 = this.ctx.createOscillator();
          const osc2 = this.ctx.createOscillator();
          const filter = this.ctx.createBiquadFilter();
          const gain = this.ctx.createGain();

          osc1.type = 'sawtooth';
          osc2.type = 'square';
          osc1.frequency.setValueAtTime(freq, now);
          osc2.frequency.setValueAtTime(freq * 1.004, now);

          filter.type = 'lowpass';
          filter.frequency.setValueAtTime(1400, now);
          filter.frequency.exponentialRampToValueAtTime(180, now + 0.11);
          filter.Q.setValueAtTime(4.0, now);

          gain.gain.setValueAtTime(0.12, now);
          gain.gain.exponentialRampToValueAtTime(0.001, now + 0.11);

          osc1.connect(filter);
          osc2.connect(filter);
          filter.connect(gain);
          gain.connect(this.ctx.destination);

          osc1.start(now);
          osc2.start(now);
          osc1.stop(now + 0.12);
          osc2.stop(now + 0.12);
        } catch (e) {}
      }

      scheduleKick(offset = 0) {
        if (!this.ctx) return;
        const now = this.ctx.currentTime + offset;
        try {
          const osc = this.ctx.createOscillator();
          const gain = this.ctx.createGain();
          osc.frequency.setValueAtTime(120, now);
          osc.frequency.exponentialRampToValueAtTime(35, now + 0.09);
          gain.gain.setValueAtTime(0.3, now);
          gain.gain.exponentialRampToValueAtTime(0.001, now + 0.1);
          osc.connect(gain);
          gain.connect(this.ctx.destination);
          osc.start(now);
          osc.stop(now + 0.11);
        } catch (e) {}
      }

      scheduleSnare(offset = 0) {
        if (!this.ctx) return;
        const now = this.ctx.currentTime + offset;
        try {
          const bSize = this.ctx.sampleRate * 0.08;
          const nBuffer = this.ctx.createBuffer(1, bSize, this.ctx.sampleRate);
          const out = nBuffer.getChannelData(0);
          for (let i = 0; i < bSize; i++) out[i] = Math.random() * 2 - 1;
          const src = this.ctx.createBufferSource();
          src.buffer = nBuffer;

          const hp = this.ctx.createBiquadFilter();
          hp.type = 'highpass';
          hp.frequency.setValueAtTime(2200, now);

          const gain = this.ctx.createGain();
          gain.gain.setValueAtTime(0.15, now);
          gain.gain.exponentialRampToValueAtTime(0.001, now + 0.08);

          src.connect(hp);
          hp.connect(gain);
          gain.connect(this.ctx.destination);
          src.start(now);
          src.stop(now + 0.09);
        } catch (e) {}
      }

      // THE CLASSIC KNEE SLAP + RIMSHOT (Thwack! Ba-dum-tss!)
      playKneeSlap() {
        if (!this.enabled || !this.ctx) return;
        this.resume();
        const now = this.ctx.currentTime;
        try {
          const bufferSize = this.ctx.sampleRate * 0.08;
          const noiseBuffer = this.ctx.createBuffer(1, bufferSize, this.ctx.sampleRate);
          const output = noiseBuffer.getChannelData(0);
          for (let i = 0; i < bufferSize; i++) output[i] = Math.random() * 2 - 1;
          const whiteNoise = this.ctx.createBufferSource();
          whiteNoise.buffer = noiseBuffer;

          const filter = this.ctx.createBiquadFilter();
          filter.type = 'bandpass';
          filter.frequency.setValueAtTime(1400, now);
          filter.Q.setValueAtTime(3.0, now);

          const slapGain = this.ctx.createGain();
          slapGain.gain.setValueAtTime(0.7, now);
          slapGain.gain.exponentialRampToValueAtTime(0.01, now + 0.07);

          whiteNoise.connect(filter);
          filter.connect(slapGain);
          slapGain.connect(this.ctx.destination);
          whiteNoise.start(now);

          const thumbOsc = this.ctx.createOscillator();
          const thumbGain = this.ctx.createGain();
          thumbOsc.frequency.setValueAtTime(260, now);
          thumbOsc.frequency.exponentialRampToValueAtTime(40, now + 0.12);
          thumbGain.gain.setValueAtTime(0.8, now);
          thumbGain.gain.exponentialRampToValueAtTime(0.01, now + 0.12);
          thumbOsc.connect(thumbGain);
          thumbGain.connect(this.ctx.destination);
          thumbOsc.start(now);
          thumbOsc.stop(now + 0.13);

          this.scheduleDrum(now + 0.22, 180, 0.35, 0.08);
          this.scheduleDrum(now + 0.36, 140, 0.45, 0.09);
          this.scheduleCymbal(now + 0.50, 0.3);
        } catch (e) {}
      }

      scheduleDrum(time, freq, gainVal, dur) {
        const osc = this.ctx.createOscillator();
        const gain = this.ctx.createGain();
        osc.frequency.setValueAtTime(freq, time);
        osc.frequency.exponentialRampToValueAtTime(freq * 0.4, time + dur);
        gain.gain.setValueAtTime(gainVal, time);
        gain.gain.exponentialRampToValueAtTime(0.01, time + dur);
        osc.connect(gain);
        gain.connect(this.ctx.destination);
        osc.start(time);
        osc.stop(time + dur + 0.01);
      }

      scheduleCymbal(time, dur) {
        const bSize = this.ctx.sampleRate * dur;
        const nBuffer = this.ctx.createBuffer(1, bSize, this.ctx.sampleRate);
        const out = nBuffer.getChannelData(0);
        for (let i = 0; i < bSize; i++) out[i] = Math.random() * 2 - 1;
        const src = this.ctx.createBufferSource();
        src.buffer = nBuffer;

        const hp = this.ctx.createBiquadFilter();
        hp.type = 'highpass';
        hp.frequency.setValueAtTime(7500, time);

        const cymGain = this.ctx.createGain();
        cymGain.gain.setValueAtTime(0.4, time);
        cymGain.gain.exponentialRampToValueAtTime(0.001, time + dur);

        src.connect(hp);
        hp.connect(cymGain);
        cymGain.connect(this.ctx.destination);
        src.start(time);
        src.stop(time + dur + 0.01);
      }
    }

    /* =========================================================================
       2. THE 12 KNEE QUESTS (WAITING ROOM CLINICAL EDUCATION SUITE)
          Green (Gentle: 1–4) | Yellow (Dynamic: 5–8) | Red (Advanced: 9–12)
       ========================================================================= */
    const QUEST_LEVELS = [
      // 🟢 GREEN TIERS (1–4: Gentle Relief & Acute Management)
      {
        id: 1,
        tier: 'green',
        tierLabel: '🟢 GENTLE',
        title: 'Level 1: Joint Effusion',
        structure: 'Synovial Fluid Capsule',
        metaphor: "The Body's Natural Airbag",
        anatomyExplain: "When the knee gets irritated or twisted, it creates extra synovial fluid. This swelling isn't an enemy—it acts like an emergency airbag to protect internal cartilage and deliver healing cells.",
        healingExplain: "Resting the leg elevated and doing gentle non-weight-bearing ankle pumps stimulates the lymphatic system to safely absorb and circulate fluid without harsh drugs.",
        doctorQuestions: [
          "Is this swelling protecting an acute twist or gradual inflammation?",
          "Should I use cold packs, compression wrap, or gentle elevation at home?",
          "Can I do gentle quad sets while seated to help the lymphatic system pump fluid out?"
        ],
        objectiveName: "Synovial Drops",
        targetCount: 8,
        terrainAmp: 45,
        terrainWave: 920,
        itemColor: '#38bdf8'
      },
      {
        id: 2,
        tier: 'green',
        tierLabel: '🟢 GENTLE',
        title: 'Level 2: MCL Sprain (Grade 1/2)',
        structure: 'Medial Collateral Ligament',
        metaphor: 'The Inner Shield & Seatbelt',
        anatomyExplain: "The MCL is a wide, tough band on the inside of your knee that stops your leg from bowing inward when knocked from the outside. Unlike inner ligaments, the MCL has a fantastic rich blood supply!",
        healingExplain: "Because of its generous blood flow, the vast majority of MCL sprains heal completely without surgery through early protected movement and targeted quad strengthening.",
        doctorQuestions: [
          "What grade is my MCL sprain (Grade 1, 2, or 3)?",
          "Do I need a hinged brace to protect lateral movement while walking?",
          "How soon can I start stationary cycling or physical therapy?"
        ],
        objectiveName: "Capillary Healing Droplets",
        targetCount: 8,
        terrainAmp: 50,
        terrainWave: 880,
        itemColor: '#f43f5e'
      },
      {
        id: 3,
        tier: 'green',
        tierLabel: '🟢 GENTLE',
        title: "Level 3: Runner's Knee / Chondromalacia",
        structure: 'Patellofemoral Cartilage',
        metaphor: 'The Polished Ice Rink',
        anatomyExplain: "Articular cartilage under your kneecap is 5 times smoother than ice on ice. When it softens or gets slightly irritated, the kneecap wanders out of its smooth femoral groove.",
        healingExplain: "Strengthening the hip stabilizers (gluteus medius) and inner quad keeps the kneecap gliding centered in the groove like a high-speed train on smooth rails.",
        doctorQuestions: [
          "Are weak hip stabilizers letting my kneecap track slightly off-center?",
          "What quad-strengthening exercises won't compress my kneecap?",
          "Would patellar taping or shoe orthotics help in the short term?"
        ],
        objectiveName: "Glide Crystals",
        targetCount: 8,
        terrainAmp: 48,
        terrainWave: 860,
        itemColor: '#06b6d4'
      },
      {
        id: 4,
        tier: 'green',
        tierLabel: '🟢 GENTLE',
        title: 'Level 4: Patellar Tendonitis',
        structure: 'Patellar Tendon / Extensor Mechanism',
        metaphor: 'The Elastic Bungee Cord',
        anatomyExplain: "The patellar tendon is a thick rubber-band-like cable connecting your kneecap to your shin. Repetitive jumping or rapid mileage increases can cause micro-strain and tenderness.",
        healingExplain: "Tendons hate sudden rest followed by sudden jumps; they love progressive, isometric holds (like 45-second wall sits) which relieve pain and stimulate collagen realignment.",
        doctorQuestions: [
          "Should I do isometric wall-sits to reduce tendon pain before exercising?",
          "What jump or running volume is safe for me right now?",
          "Is a patellar tendon strap or band helpful for stairs?"
        ],
        objectiveName: "Tendon Collagen Bundles",
        targetCount: 8,
        terrainAmp: 52,
        terrainWave: 840,
        itemColor: '#facc15'
      },

      // 🟡 YELLOW TIERS (5–8: Dynamic Stability & Meniscal Shock Absorption)
      {
        id: 5,
        tier: 'yellow',
        tierLabel: '🟡 DYNAMIC',
        title: 'Level 5: Medial Meniscus Tear',
        structure: 'Medial Meniscal Cartilage',
        metaphor: 'The Inner Sneaker Insole',
        anatomyExplain: "Your medial meniscus is a C-shaped fibrocartilage shock absorber. Like a high-end sneaker insole, it cushions up to 50% of your body weight during squats, turns, and steps.",
        healingExplain: "The outer rim of the meniscus (the 'Red-Red' vascular zone) has direct blood vessels that can heal or be sutured. Even degenerative tears often become completely asymptomatic with physical therapy!",
        doctorQuestions: [
          "Is my tear in the outer 'Red-Red' vascular zone (which can heal) or the inner white zone?",
          "Does this tear catch, lock, or prevent me from straightening my knee?",
          "Can we try 6 to 8 weeks of targeted physical therapy before considering arthroscopy?"
        ],
        objectiveName: "Fibrocartilage Sponges",
        targetCount: 10,
        terrainAmp: 56,
        terrainWave: 820,
        itemColor: '#10b981'
      },
      {
        id: 6,
        tier: 'yellow',
        tierLabel: '🟡 DYNAMIC',
        title: 'Level 6: Lateral Meniscus Tear',
        structure: 'Lateral Meniscal Cartilage',
        metaphor: 'The Mobile Outer Shock Absorber',
        anatomyExplain: "The lateral meniscus is more circular and mobile than the medial side, sliding back and forth to absorb impact during quick pivots, direction changes, and downhill strides.",
        healingExplain: "Low-impact cyclic motion (like stationary cycling or pool exercise) pumps synovial fluid across the meniscus, providing continuous oxygen and nutrients to maintain tissue hydration.",
        doctorQuestions: [
          "How does lateral meniscus treatment differ from medial tears?",
          "What low-impact exercises will keep synovial fluid bathing the cushion?",
          "Are deep squats or kneeling safe for my specific tear pattern?"
        ],
        objectiveName: "Lateral Buffer Pads",
        targetCount: 10,
        terrainAmp: 58,
        terrainWave: 800,
        itemColor: '#a855f7'
      },
      {
        id: 7,
        tier: 'yellow',
        tierLabel: '🟡 DYNAMIC',
        title: 'Level 7: Bone Bruise / Contusion',
        structure: 'Subchondral Trabecular Bone',
        metaphor: 'The Micro-Honeycomb Sponge',
        anatomyExplain: "Bones aren't solid porcelain; inside they are a living, flexible lattice of microscopic struts. A heavy impact micro-fractures these internal struts, creating bone marrow swelling.",
        healingExplain: "Bone bruises heal from the inside out over 6 to 12 weeks. Giving the bone time without high-impact pounding allows osteoblasts to rebuild the microscopic honeycomb struts stronger than before.",
        doctorQuestions: [
          "Why does a bone bruise take 6 to 12 weeks to heal compared to normal bruises?",
          "How much weight-bearing or walking is recommended while the bone marrow heals?",
          "Are there nutritional factors (Vitamin D, calcium, protein) that support bone recovery?"
        ],
        objectiveName: "Trabecular Honeycomb Seeds",
        targetCount: 10,
        terrainAmp: 60,
        terrainWave: 780,
        itemColor: '#fbbf24'
      },
      {
        id: 8,
        tier: 'yellow',
        tierLabel: '🟡 DYNAMIC',
        title: 'Level 8: PCL Tear / Sprain',
        structure: 'Posterior Cruciate Ligament',
        metaphor: 'The Reverse Anchor Cable',
        anatomyExplain: "The PCL is the thick anchor cable holding your shin bone from sliding backward toward your calf. It is commonly strained by hitting a car dashboard or landing hard on a bent knee.",
        healingExplain: "The PCL has an extraordinary intrinsic healing capacity! In many cases, specialized dynamic bracing and aggressive quad strengthening stabilize the shin completely without surgical reconstruction.",
        doctorQuestions: [
          "Can my PCL heal without surgery using non-operative dynamic bracing?",
          "Which hamstring exercises should I avoid temporarily so the shin doesn't pull backward?",
          "When can I begin closed-chain quad strengthening?"
        ],
        objectiveName: "Quad Torque Anchors",
        targetCount: 10,
        terrainAmp: 62,
        terrainWave: 760,
        itemColor: '#f97316'
      },

      // 🔴 RED TIERS (9–12: High Acuity, Complex Reconstruction & Long-Term Joint)
      {
        id: 9,
        tier: 'red',
        tierLabel: '🔴 ADVANCED',
        title: 'Level 9: Complete ACL Tear',
        structure: 'Anterior Cruciate Ligament',
        metaphor: 'The Pivot Canyon & Forward Brake',
        anatomyExplain: "The ACL is the primary forward brake cable inside your joint, stopping the shin from sliding forward or twisting out of control when planting or pivoting.",
        healingExplain: "Modern ACL restoration uses minimally invasive techniques with robust biological scaffolds (like diatom biosilica anchors) paired with targeted pre-habilitation to guarantee a strong, stable return to play.",
        doctorQuestions: [
          "What are the pros and cons of autograft (patellar tendon, quad, hamstring) vs allograft for my activity level?",
          "How important is 'pre-hab' to restore full extension and reduce swelling before surgery?",
          "What is my realistic timeline for returning to cutting and pivoting sports?"
        ],
        objectiveName: "Collagen Graft Strands",
        targetCount: 12,
        terrainAmp: 68,
        terrainWave: 740,
        itemColor: '#ef4444'
      },
      {
        id: 10,
        tier: 'red',
        tierLabel: '🔴 ADVANCED',
        title: 'Level 10: LCL & Posterolateral Corner',
        structure: 'Lateral Collateral & Popliteus Complex',
        metaphor: 'The Outer Fortress Towers',
        anatomyExplain: "The LCL and posterolateral corner form the outer defense wall preventing the knee from twisting outward or bowing into a severe 'bowleg' posture.",
        healingExplain: "Accurate physical examination and early stabilization protect the nearby peroneal nerve and restore rotational equilibrium for confident, stable walking.",
        doctorQuestions: [
          "Did my outer injury affect the peroneal nerve (any numbness or foot drop)?",
          "Is the posterolateral corner (PLC) intact or involved with the LCL?",
          "Does this require surgical repair or can a specialized brace stabilize it?"
        ],
        objectiveName: "Rotational Balance Gyros",
        targetCount: 12,
        terrainAmp: 66,
        terrainWave: 750,
        itemColor: '#ec4899'
      },
      {
        id: 11,
        tier: 'red',
        tierLabel: '🔴 ADVANCED',
        title: 'Level 11: Quadriceps Tendon Rupture / Strain',
        structure: 'Quadriceps Muscle-Tendon Junction',
        metaphor: 'The Big Engine Cable',
        anatomyExplain: "The quadriceps tendon is the heavy transmission cable transferring power from your 4 massive thigh muscles to straighten your leg and lift your body out of deep chairs.",
        healingExplain: "Following surgical repair or high-grade rehab, progressive closed-chain strength programs systematically rebuild muscle bulk and neuromuscular control.",
        doctorQuestions: [
          "Is the tendon fully or partially torn, and what is the surgical window if full?",
          "How long will the knee need to remain locked in full extension during early healing?",
          "When can we begin passive range-of-motion exercises?"
        ],
        objectiveName: "Quad Power Cells",
        targetCount: 12,
        terrainAmp: 70,
        terrainWave: 720,
        itemColor: '#f43f5e'
      },
      {
        id: 12,
        tier: 'red',
        tierLabel: '🔴 ADVANCED',
        title: 'Level 12: Osteoarthritis / The Long Glide',
        structure: 'Full Joint Articular Cartilage',
        metaphor: 'The Silver Glider: Movement is Medicine',
        anatomyExplain: "Articular cartilage is a living sponge with zero blood vessels—it eats only through cyclic compression and release. Sitting motionless starves it; rhythmic, low-impact gliding nourishes it!",
        healingExplain: "Movement is medicine! Low-impact activities (swimming, cycling, walking) flood the joint with protective lubricants, strengthen supporting muscles, and dramatically reduce joint stiffness.",
        doctorQuestions: [
          "What low-impact activities (swimming, cycling, walking on grass) are best for my joint stage?",
          "Could unloader braces, physical therapy, or targeted injections relieve pain?",
          "What are the milestones for knowing if or when joint replacement makes sense?"
        ],
        objectiveName: "Synovial Nourishment Spheres",
        targetCount: 15,
        terrainAmp: 46,
        terrainWave: 980,
        itemColor: '#facc15'
      }
    ];

    /* =========================================================================
       3. GAMEPAD CONTROLLER ENGINE (XBOX, PLAYSTATION, SWITCH, ADAPTIVE)
       ========================================================================= */
    class GamepadManager {
      constructor() {
        this.gamepadIndex = null;
        this.lastButtons = {};
        this.init();
      }

      init() {
        window.addEventListener("gamepadconnected", (e) => {
          this.gamepadIndex = e.gamepad.index;
          const name = e.gamepad.id.toLowerCase().includes("xbox") ? "Xbox Controller" :
                       e.gamepad.id.toLowerCase().includes("dual") || e.gamepad.id.toLowerCase().includes("playstation") ? "PlayStation Controller" :
                       e.gamepad.id.toLowerCase().includes("switch") || e.gamepad.id.toLowerCase().includes("joy-con") ? "Switch Controller" : "Gamepad";
          document.getElementById('gamepad-name').textContent = name;
          document.getElementById('gamepad-pill').style.display = 'inline-flex';
          this.rumble(200, 0.4, 0.4);
        });

        window.addEventListener("gamepaddisconnected", (e) => {
          if (this.gamepadIndex === e.gamepad.index) {
            this.gamepadIndex = null;
            document.getElementById('gamepad-pill').style.display = 'none';
          }
        });
      }

      rumble(durationMs = 120, weak = 0.5, strong = 0.5) {
        if (this.gamepadIndex === null) return;
        const gp = navigator.getGamepads ? navigator.getGamepads()[this.gamepadIndex] : null;
        if (gp && gp.vibrationActuator) {
          try {
            gp.vibrationActuator.playEffect("dual-rumble", {
              startDelay: 0,
              duration: durationMs,
              weakMagnitude: weak,
              strongMagnitude: strong
            });
          } catch(e) {}
        }
      }

      poll(onDive, onSlap, onTurbo) {
        if (this.gamepadIndex === null) return;
        const gamepads = navigator.getGamepads ? navigator.getGamepads() : [];
        const gp = gamepads[this.gamepadIndex];
        if (!gp) return;

        // Button 0 (A / Cross / Switch B): Dive
        const btnA = gp.buttons[0]?.pressed;
        // Button 1 (B / Circle) or Button 2 (X / Square): Knee Slap
        const btnSlap = gp.buttons[1]?.pressed || gp.buttons[2]?.pressed;
        // Button 7 (RT / R2) or Button 3 (Y / Triangle): Turbo Boost
        const btnTurbo = gp.buttons[3]?.pressed || gp.buttons[7]?.pressed;

        // Left Analog Stick Down
        const stickY = gp.axes[1] || 0;
        const stickDive = stickY > 0.4;

        onDive(btnA || stickDive);

        if (btnSlap && !this.lastButtons.slap) {
          onSlap();
          this.rumble(150, 0.9, 0.4);
        }

        if (btnTurbo && !this.lastButtons.turbo) {
          onTurbo();
          this.rumble(240, 0.5, 0.8);
        }

        this.lastButtons.slap = btnSlap;
        this.lastButtons.turbo = btnTurbo;
      }
    }

    /* =========================================================================
       4. JOKES REPOSITORY
       ========================================================================= */
    const PATTY_JOKES = [
      { setup: "Why did the patella go to medical school?", punchline: "To get a little more flexible and find its groove in the trochlea!", category: "🦴 Orthopedics" },
      { setup: "What did the femur say to the patella before running the marathon?", punchline: "'Don't worry, I've got your back — just keep tracking straight!'", category: "🦴 Orthopedics" },
      { setup: "Why are orthopedic surgeons always so calm and collected?", punchline: "Because nothing gets under their articular cartilage!", category: "🩺 Hospital Life" },
      { setup: "What is a radiologist's absolute favorite beverage during a night shift?", punchline: "A piping hot mug of Contrast-Tea!", category: "🩻 Radiology" },
      { setup: "Why did the ACL break up with the PCL?", punchline: "Because they were always crossing each other!", category: "🦴 Orthopedics" },
      { setup: "Why did the radiologist bring an umbrella to the MRI suite?", punchline: "They heard there was a strong gradient field with scattered proton spins!", category: "🩻 Radiology" },
      { setup: "What did the medial meniscus say to the lateral meniscus during a pivot?", punchline: "'Hang in there buddy, we're absorbing high shock today!'", category: "🦴 Orthopedics" },
      { setup: "Why couldn't the skeleton start a fight at the hospital party?", punchline: "He had no guts, and his knees were too busy doing the twist!", category: "✨ Anatomy" },
      { setup: "What is a knee's favorite genre of music?", punchline: "Hip-Hop... but only with proper quadriceps support!", category: "🌱 Dad Jokes" },
      { setup: "Why did the nurse bring a red marker to the triage desk?", punchline: "In case they had to draw blood in 4K resolution!", category: "🩺 Nursing" },
      { setup: "What did the synovial fluid say during cold morning rounds?", punchline: "'Just warming up my viscosity, ready to glide!'", category: "✨ Physiology" },
      { setup: "How does a happy knee answer the phone when clinic calls?", punchline: "'Yellow-cartilage! How may I articulate today?'", category: "🌱 Dad Jokes" },
      { setup: "What do you call two knee surgeons who agree on everything?", punchline: "A joint decision!", category: "🦴 Orthopedics" },
      { setup: "Why do knees make the best comedians in the entire hospital?", punchline: "Because every single one of their jokes is a genuine Knee-Slapper!", category: "🌱 Classic" }
    ];

    const KNEE_RIDER_JOKES = [
      { setup: "Michael, I have scanned the breakroom on floor 3...", punchline: "The last blueberry muffin has been compromised. Commencing tactical pursuit.", category: "🏎️ K.N.E.E. 2000" },
      { setup: "Why does K.N.E.E. 2000 never experience ligament tears?", punchline: "Dual-redundant carbon fiber composite tracking and molecular defense shielding, Michael.", category: "🏎️ K.N.E.E. 2000" },
      { setup: "What is K.N.E.E.'s preferred diagnostic sequence on the MRI scanner?", punchline: "Turbo Spin Echo, Michael. Naturally. Everything I do is Turbo.", category: "🏎️ K.N.E.E. 2000" },
      { setup: "Why did K.N.E.E. refuse to scan the patient's elbow?", punchline: "'I only ride for lower extremity joint justice, Michael. Foundation directives.'", category: "🏎️ K.N.E.E. 2000" },
      { setup: "Michael, my optical sensors indicate an acute emergency in PACS Reading Room 4...", punchline: "The espresso roast carafe is at 4% capacity. Engaging Turbo Boost immediately.", category: "🏎️ K.N.E.E. 2000" },
      { setup: "How does K.N.E.E. calibrate its condylar hill flight trajectory?", punchline: "With 99.8% algorithmic perfection. Human manual descent is mathematically chaotic, Michael.", category: "🏎️ K.N.E.E. 2000" }
    ];

    /* =========================================================================
       5. CIRCADIAN & SYNTHWAVE PALETTES
       ========================================================================= */
    const CIRCADIAN_MODES = [
      { id: 'daylight', name: 'Sunlight', icon: '☀️', skyTop: '#0ea5e9', skyBottom: '#bae6fd', sunColor: '#fef08a', hill1: 'rgba(52, 211, 153, 0.35)', hill2: 'rgba(16, 185, 129, 0.65)', hill3: '#059669', wireframe: false },
      { id: 'golden', name: 'Golden Hour', icon: '🌅', skyTop: '#7c2d12', skyBottom: '#fdba74', sunColor: '#f97316', hill1: 'rgba(180, 83, 9, 0.35)', hill2: 'rgba(16, 185, 129, 0.55)', hill3: '#065f46', wireframe: false },
      { id: 'midnight', name: 'PACS Midnight', icon: '🌙', skyTop: '#020617', skyBottom: '#0f172a', sunColor: '#38bdf8', hill1: 'rgba(15, 23, 42, 0.8)', hill2: 'rgba(6, 78, 59, 0.6)', hill3: '#064e3b', wireframe: false },
      { id: 'dew', name: 'Morning Dew', icon: '🌿', skyTop: '#0d9488', skyBottom: '#ccfbf1', sunColor: '#fef9c3', hill1: 'rgba(45, 212, 191, 0.35)', hill2: 'rgba(13, 148, 136, 0.65)', hill3: '#0f766e', wireframe: false }
    ];

    const SYNTHWAVE_PALETTE = {
      id: 'synthwave', name: 'Cyber Outrun', icon: '🏎️', skyTop: '#090514', skyBottom: '#2d0838', sunColor: '#f43f5e',
      hill1: 'rgba(236, 72, 153, 0.25)', hill2: 'rgba(168, 85, 247, 0.45)', hill3: '#180728', gridColor: '#06b6d4', wireframe: true
    };

    /* =========================================================================
       6. MAIN ENGINE & KINEMATICS
       ========================================================================= */
    const canvas = document.getElementById('game-canvas');
    const ctx = canvas.getContext('2d');
    const audio = new KneeAudioEngine();
    const gamepads = new GamepadManager();

    let width = 0;
    let height = 0;
    let dpr = window.devicePixelRatio || 1;

    function resizeCanvas() {
      width = window.innerWidth;
      height = window.innerHeight;
      dpr = window.devicePixelRatio || 1;
      canvas.width = width * dpr;
      canvas.height = height * dpr;
      ctx.scale(dpr, dpr);
    }
    window.addEventListener('resize', resizeCanvas);
    resizeCanvas();

    // Core States
    let heroMode = 'patty';
    let isCinemaMode = true;
    let currentChronoIndex = 0;
    let isDiving = false;
    let isTurbo = false;
    let turboCharge = 100;
    let distanceMeters = 0;
    let kneeSlapsCount = 0;
    // --- 🌊 SALUTOGENIC FLOW STATE & STRATEGIC KNEE SLAP SCREEN-BOMB ---
    let flowCombo = 0;
    let isFlowActive = false;
    let flowTimer = 0;
    let lastLandType = 'none'; // 'downslope', 'upslope'
    let slapCooldownTimer = 0; // 10s cooldown
    let shockwaveRadius = 0;
    let isShockwaveActive = false;
    let pendingCompletedQuest = null; // For non-interruptive banner
    let harmonyMeter = 100;

    // Active Quest Level State
    let activeQuest = null; // null = free flight
    let questCollected = 0;

    // Pacing & Speed Controller (Calm Waiting Room Standard)
    const PACING_MODES = [
      { name: "Serene (0.6x)", icon: "🐢", mult: 0.60 },
      { name: "Gentle (0.8x)", icon: "🌿", mult: 0.80 },
      { name: "Flow (1.0x)", icon: "🚶", mult: 1.0 },
      { name: "Swift (1.3x)", icon: "🏎️", mult: 1.3 }
    ];
    let currentPacingIdx = 0; // Starts in Serene (0.6x)
    let pacingMultiplier = PACING_MODES[0].mult;

    const avatar = {
      x: 220,
      y: 300,
      radius: 28,
      vx: 1.6, // Calm, smooth, gentle drift
      vy: 0,
      angle: 0,
      squashX: 1,
      squashY: 1,
      expression: 'happy',
      blushing: true,
      lastSlapTime: 0,
      scannerPos: 0,
      scannerDir: 1
    };

    // Parallax Hills: Wide, majestic, slow-moving condylar curves
    const hills = [
      { amplitude: 110, wavelength: 1400, speed: 0.12, offset: 0, baseY: 0.60 },
      { amplitude: 80, wavelength: 980, speed: 0.25, offset: 0, baseY: 0.68 },
      { amplitude: 60, wavelength: 700, speed: 0.45, offset: 0, baseY: 0.76 }
    ];

    const collectibles = [];
    const particles = [];
    const confetti = [];

    function getForegroundHillY(worldX) {
      const h = hills[2];
      const normalizedX = (worldX + h.offset);
      const yBase = height * h.baseY;
      return yBase +
             Math.sin(normalizedX / (h.wavelength / (Math.PI * 2))) * (h.amplitude * 0.75) +
             Math.cos(normalizedX / (h.wavelength * 0.45)) * (h.amplitude * 0.25);
    }

    function getForegroundSlope(worldX) {
      const delta = 4;
      const y1 = getForegroundHillY(worldX - delta);
      const y2 = getForegroundHillY(worldX + delta);
      return (y2 - y1) / (delta * 2);
    }

    function spawnCollectibles() {
      collectibles.length = 0;
      // Golden Path: Group collectibles in 4 fluid ballistic flight arcs
      for (let arc = 0; arc < 4; arc++) {
        const startX = avatar.x + 350 + arc * 720;
        const arcCount = 5;
        for (let i = 0; i < arcCount; i++) {
          const t = i / (arcCount - 1); // 0.0 to 1.0 along the jump
          const cx = startX + i * 90;
          // Parabolic arc matching optimal launch flight trajectory
          const groundBase = getForegroundHillY(cx);
          const apexHeight = 160 + (i === 2 ? 30 : 0); // Apex at center
          const arcHeight = Math.sin(t * Math.PI) * apexHeight + 50;
          const cy = groundBase - arcHeight;

          const isApex = (i === Math.floor(arcCount / 2));
          collectibles.push({
            x: cx,
            y: cy,
            radius: isApex ? 16 : 12,
            type: activeQuest ? 'quest' : (isApex ? 'synovial' : 'biosilica'),
            isApex: isApex,
            collected: false,
            pulse: i * 0.4
          });
        }
      }
    }
    spawnCollectibles();

    /* =========================================================================
       7. JOKES, SLAPS & QUEST HUD
       ========================================================================= */
    const jokeBanner = document.getElementById('joke-banner');
    const jokeBadge = document.getElementById('joke-badge');
    const kittVoiceBox = document.getElementById('kitt-voice-box');
    const jokeSetup = document.getElementById('joke-setup-text');
    const jokePunchline = document.getElementById('joke-punchline-text');
    const slapFlash = document.getElementById('slap-flash');

    let currentPattyJokeIdx = 0;
    let currentRiderJokeIdx = 0;
    let unlockedPattyJokes = new Set([0]);
    let unlockedRiderJokes = new Set([0]);

    function triggerKneeSlap(x, y) {
      audio.playKneeSlap();
      gamepads.rumble(150, 0.9, 0.4);
      kneeSlapsCount++;
      document.getElementById('stat-slaps').textContent = kneeSlapsCount;

      slapFlash.classList.add('active');
      setTimeout(() => slapFlash.classList.remove('active'), 120);

      avatar.expression = 'slapped';
      avatar.squashX = 1.35;
      avatar.squashY = 0.65;
      avatar.vy = -12;
      avatar.lastSlapTime = Date.now();

      const isRider = (heroMode === 'knee_rider');
      const burstColors = isRider
        ? ['#ef4444', '#f43f5e', '#06b6d4', '#ec4899', '#ffffff']
        : ['#10b981', '#34d399', '#facc15', '#f43f5e', '#38bdf8'];

      for (let i = 0; i < 36; i++) {
        confetti.push({
          x: x || avatar.x,
          y: y || avatar.y,
          vx: (Math.random() - 0.5) * 16,
          vy: (Math.random() - 0.8) * 16,
          color: burstColors[Math.floor(Math.random() * burstColors.length)],
          size: Math.random() * 8 + 4,
          rotation: Math.random() * Math.PI * 2,
          vRot: (Math.random() - 0.5) * 0.2,
          life: 1.0
        });
      }

      let joke;
      if (isRider) {
        currentRiderJokeIdx = (currentRiderJokeIdx + 1) % KNEE_RIDER_JOKES.length;
        unlockedRiderJokes.add(currentRiderJokeIdx);
        joke = KNEE_RIDER_JOKES[currentRiderJokeIdx];

        jokeBanner.classList.add('rider');
        jokeBadge.classList.add('rider');
        jokeBadge.textContent = "⚡ K.N.E.E. 2000 DIRECTIVE";
        kittVoiceBox.classList.add('active');
        jokePunchline.classList.add('rider');
      } else {
        currentPattyJokeIdx = (currentPattyJokeIdx + 1) % PATTY_JOKES.length;
        unlockedPattyJokes.add(currentPattyJokeIdx);
        joke = PATTY_JOKES[currentPattyJokeIdx];

        jokeBanner.classList.remove('rider');
        jokeBadge.classList.remove('rider');
        jokeBadge.textContent = "⚡ KNEE SLAPPER UNLOCKED!";
        kittVoiceBox.classList.remove('active');
        jokePunchline.classList.remove('rider');
      }

      jokeSetup.textContent = joke.setup;
      jokePunchline.textContent = joke.punchline;
      jokeBanner.classList.add('show');

      clearTimeout(window.jokeTimer);
      window.jokeTimer = setTimeout(() => {
        jokeBanner.classList.remove('show');
      }, 9500);
    }

    document.getElementById('joke-close-btn').addEventListener('click', () => jokeBanner.classList.remove('show'));
    document.getElementById('joke-next-btn').addEventListener('click', () => triggerKneeSlap(avatar.x, avatar.y));
    document.getElementById('joke-share-btn').addEventListener('click', () => {
      const isRider = (heroMode === 'knee_rider');
      const joke = isRider ? KNEE_RIDER_JOKES[currentRiderJokeIdx] : PATTY_JOKES[currentPattyJokeIdx];
      const speaker = isRider ? 'K.N.E.E. 2000' : 'Patty the Green Happy Knee';
      const text = `🦵 Knee-Slapper from ${speaker}:\\n${joke.setup}\\n👉 ${joke.punchline}\\n(PocketGull Shift Sanctuary)`;
      navigator.clipboard.writeText(text).then(() => {
        const btn = document.getElementById('joke-share-btn');
        btn.textContent = '✅ Copied!';
        setTimeout(() => btn.textContent = '📋 Copy Joke', 2000);
      });
    });

    /* =========================================================================
       8. QUEST SYSTEM CONTROLLER & WAITING ROOM CONSULTATION CARD
       ========================================================================= */
    const questsModal = document.getElementById('quests-modal');
    const questBriefModal = document.getElementById('quest-brief-modal');
    const questHudPill = document.getElementById('quest-hud-pill');
    const questHudLabel = document.getElementById('quest-hud-label');
    const questsBtnText = document.getElementById('quests-btn-text');

    let selectedQuestForBriefing = null;

    function renderQuestsList(filter = 'all') {
      const container = document.getElementById('quests-list-container');
      const filtered = QUEST_LEVELS.filter(q => filter === 'all' || q.tier === filter);

      container.innerHTML = filtered.map(q => `
        <div class="quest-card tier-border-${q.tier}" onclick="openQuestBriefing(${q.id})">
          <div class="quest-card-header">
            <span class="quest-num">${q.title}</span>
            <span class="tier-badge tier-${q.tier}">${q.tierLabel}</span>
          </div>
          <div class="quest-metaphor">${q.metaphor}</div>
          <div class="quest-desc">${q.anatomyExplain}</div>
          <div class="quest-footer">
            <span>🎯 Mission: Collect ${q.targetCount} ${q.objectiveName}</span>
            <span style="color: #38bdf8; font-weight: 700;">Tap for Patient Guide &rarr;</span>
          </div>
        </div>
      `).join('');
    }

    window.openQuestBriefing = function(questId) {
      const q = QUEST_LEVELS.find(item => item.id === questId);
      if (!q) return;
      selectedQuestForBriefing = q;

      const tierPill = document.getElementById('brief-tier-pill');
      tierPill.className = `tier-badge tier-${q.tier}`;
      tierPill.textContent = q.tierLabel;

      document.getElementById('brief-modal-title').textContent = q.title;
      document.getElementById('brief-metaphor').textContent = `"${q.metaphor}"`;
      document.getElementById('brief-anatomy-explain').textContent = q.anatomyExplain;
      document.getElementById('brief-healing-explain').textContent = q.healingExplain;

      const questionsList = document.getElementById('brief-doctor-questions');
      questionsList.innerHTML = q.doctorQuestions.map(ques => `<li style="margin-bottom: 6px;">${ques}</li>`).join('');

      questsModal.classList.remove('open');
      questBriefModal.classList.add('open');
    };

    document.getElementById('btn-quests').addEventListener('click', () => {
      renderQuestsList('all');
      questsModal.classList.add('open');
    });
    document.getElementById('quests-close-btn').addEventListener('click', () => questsModal.classList.remove('open'));
    document.getElementById('brief-close-btn').addEventListener('click', () => questBriefModal.classList.remove('open'));

    // Filter Tabs
    document.querySelectorAll('.filter-tab').forEach(tab => {
      tab.addEventListener('click', (e) => {
        document.querySelectorAll('.filter-tab').forEach(t => t.classList.remove('active'));
        tab.classList.add('active');
        renderQuestsList(tab.dataset.filter);
      });
    });

    // Copy Doctor Questions to Clipboard
    document.getElementById('btn-copy-doc-questions').addEventListener('click', () => {
      if (!selectedQuestForBriefing) return;
      const q = selectedQuestForBriefing;
      const text = `🏥 PocketGull Knee Patient Guide: ${q.title}\\n"${q.metaphor}"\\n\\n🩺 3 Questions for My Doctor:\\n1. ${q.doctorQuestions[0]}\\n2. ${q.doctorQuestions[1]}\\n3. ${q.doctorQuestions[2]}\\n\\n(Generated in Waiting Room)`;
      navigator.clipboard.writeText(text).then(() => {
        const btn = document.getElementById('btn-copy-doc-questions');
        btn.textContent = '✅ Copied to Phone!';
        setTimeout(() => btn.textContent = '📋 Copy Questions to Phone', 2500);
      });
    });

    // Launch Quest Gameplay
    document.getElementById('btn-launch-quest').addEventListener('click', () => {
      if (!selectedQuestForBriefing) return;
      activeQuest = selectedQuestForBriefing;
      questCollected = 0;

      // Adjust terrain parameters to reflect the condition!
      hills[2].amplitude = activeQuest.terrainAmp;
      hills[2].wavelength = activeQuest.terrainWave;

      questHudLabel.textContent = `${activeQuest.title}: 0/${activeQuest.targetCount}`;
      questHudPill.style.display = 'inline-flex';
      questsBtnText.textContent = `🎯 Level ${activeQuest.id}`;

      spawnCollectibles();
      questBriefModal.classList.remove('open');

      // Switch to interactive play
      if (isCinemaMode) {
        document.getElementById('btn-mode').click();
      }
    });

    function checkQuestCompletion() {
      if (!activeQuest) return;
      questCollected++;
      questHudLabel.textContent = `${activeQuest.title}: ${questCollected}/${activeQuest.targetCount}`;
      gamepads.rumble(80, 0.4, 0.4);

      if (questCollected >= activeQuest.targetCount) {
        audio.playQuestVictory();
        gamepads.rumble(400, 0.8, 0.9);

        // Burst celebration confetti
        for (let i = 0; i < 60; i++) {
          confetti.push({
            x: avatar.x,
            y: avatar.y,
            vx: (Math.random() - 0.5) * 20,
            vy: (Math.random() - 0.8) * 20,
            color: ['#10b981', '#34d399', '#facc15', '#38bdf8', '#ef4444'][Math.floor(Math.random() * 5)],
            size: Math.random() * 9 + 4,
            rotation: Math.random() * Math.PI * 2,
            vRot: (Math.random() - 0.5) * 0.2,
            life: 1.2
          });
        }

        // Flow-Preserving Victory Banner (No Hard-Stop Modal Context Switch!)
        pendingCompletedQuest = activeQuest;
        const banner = document.getElementById('quest-complete-flow-banner');
        const bannerTitle = document.getElementById('flow-banner-title');
        if (banner && bannerTitle) {
          bannerTitle.textContent = `${activeQuest.name} Fully Restored!`;
          banner.style.display = 'flex';
          banner.style.opacity = '1';
        }
        audio.playChime(880);
        hills[2].amplitude = 60;
        hills[2].wavelength = 700;
        activeQuest = null;
        questHudPill.style.display = 'none';
        questsBtnText.textContent = '12 Knee Quests';
        spawnCollectibles();
      }
    }

    /* =========================================================================
       9. HERO MODE SWITCHER: GREEN HAPPY KNEE ⟷ KNEE RIDER
       ========================================================================= */
    const btnHero = document.getElementById('btn-hero');
    const heroIcon = document.getElementById('hero-icon');
    const heroText = document.getElementById('hero-text');
    const brandPill = document.getElementById('brand-pill');
    const brandIcon = document.getElementById('brand-icon');
    const brandTitle = document.getElementById('brand-title');
    const hintText = document.getElementById('hint-text');
    const btnTurbo = document.getElementById('btn-turbo');
    const statSpecialLabel = document.getElementById('stat-special-label');
    const statSpecialVal = document.getElementById('stat-special-val');
    const statDist = document.getElementById('stat-dist');
    const statSlaps = document.getElementById('stat-slaps');

    function switchHeroMode(mode) {
      heroMode = mode;
      const isRider = (heroMode === 'knee_rider');
      document.body.classList.toggle('synthwave-mode', isRider);

      if (isRider) {
        heroIcon.textContent = '🌱';
        heroText.textContent = 'Happy Knee';
        btnHero.classList.add('rider-active');

        brandPill.classList.add('rider');
        brandIcon.textContent = '🏎️';
        brandTitle.textContent = 'KNEE RIDER 2000';

        statSpecialLabel.textContent = 'Turbo Boost:';
        statSpecialVal.textContent = `${Math.round(turboCharge)}%`;
        statSpecialVal.classList.add('rider');
        statDist.classList.add('rider');
        statSlaps.classList.add('rider');

        hintText.innerHTML = `Hold <span class="hint-key">SPACE</span> / <span class="hint-key">A</span> to dive &bull; <span class="hint-key">SHIFT</span> / <span class="hint-key">T</span> / <span class="hint-key">RT</span> for <b>TURBO BOOST</b>! Click K.N.E.E. for 80s tactical jokes!`;
        btnTurbo.style.display = 'inline-flex';

        audio.setRiderMode(true);
        audio.playScannerWhoosh();
      } else {
        heroIcon.textContent = '🏎️';
        heroText.textContent = 'KNEE RIDER';
        btnHero.classList.remove('rider-active');

        brandPill.classList.remove('rider');
        brandIcon.textContent = '🌱';
        brandTitle.textContent = "Patty's Knee Slapper";

        statSpecialLabel.textContent = 'Harmony:';
        statSpecialVal.textContent = `${harmonyMeter}%`;
        statSpecialVal.classList.remove('rider');
        statDist.classList.remove('rider');
        statSlaps.classList.remove('rider');

        hintText.innerHTML = `Hold <span class="hint-key">SPACE</span> / <span class="hint-key">CLICK</span> / Gamepad <span class="hint-key">A</span> to dive &bull; Release to launch! Click Patty for a <b>Knee Slapper</b>!`;
        btnTurbo.style.display = 'none';

        audio.setRiderMode(false);
      }
    }

    btnHero.addEventListener('click', () => {
      switchHeroMode(heroMode === 'patty' ? 'knee_rider' : 'patty');
    });

    function triggerTurboBoost() {
      if (turboCharge < 25) return;
      isTurbo = true;
      audio.playTurboBoostSound();
      gamepads.rumble(250, 0.5, 0.8);
      avatar.vx = 4.5;
      avatar.vy = -120 * pacingMultiplier;
      turboCharge = Math.max(0, turboCharge - 35);
      if (heroMode === 'knee_rider') {
        statSpecialVal.textContent = `${Math.round(turboCharge)}%`;
      }
      setTimeout(() => { isTurbo = false; }, 950);
    }

    btnTurbo.addEventListener('click', triggerTurboBoost);

    /* =========================================================================
       10. INPUT HANDLING & GAMEPAD POLLING
       ========================================================================= */
    canvas.addEventListener('pointerdown', (e) => {
      const rect = canvas.getBoundingClientRect();
      const clickX = e.clientX - rect.left;
      const clickY = e.clientY - rect.top;

      const distToAvatar = Math.hypot(clickX - avatar.x, clickY - avatar.y);
      if (distToAvatar < avatar.radius * 2.2) {
        triggerKneeSlap(avatar.x, avatar.y);
      } else {
        isDiving = true;
      }
    });

    window.addEventListener('pointerup', () => { isDiving = false; });

    window.addEventListener('keydown', (e) => {
      if (e.code === 'Space' || e.code === 'ArrowDown') isDiving = true;
      if (e.code === 'KeyK') triggerKneeSlap(avatar.x, avatar.y);
      if (e.code === 'KeyT' || e.code === 'ShiftLeft' || e.code === 'ShiftRight') triggerTurboBoost();
    });

    window.addEventListener('keyup', (e) => {
      if (e.code === 'Space' || e.code === 'ArrowDown') isDiving = false;
    });

    document.getElementById('btn-mobile-slap').addEventListener('click', () => {
      triggerKneeSlap(avatar.x, avatar.y);
    });

    /* =========================================================================
       11. SHIFT ROSTER & JOKE BOOK MODALS
       ========================================================================= */
    const DEFAULT_ROSTER = [
      { rank: 1, name: "Dr. Alex M.", role: "🩻 Radiology", distance: 4820, slaps: 16, note: "PACS room 3 is serene tonight, fresh roast coffee in breakroom!" },
      { rank: 2, name: "K.N.E.E. Unit 01", role: "🏎️ Knee Rider Unit", distance: 4450, slaps: 19, note: "All joint borders secured. Turbo Boost operational." },
      { rank: 3, name: "Nurse Elena", role: "🩺 Nursing", distance: 3640, slaps: 11, note: "Triage clear! Sending warmth to the oncoming morning crew 🌿" },
      { rank: 4, name: "Tech Marcus", role: "✨ Radiologist Fellow", distance: 3150, slaps: 9, note: "Patty caught the triple condyle jump on the hill crest!" },
      { rank: 5, name: "Jordan R.", role: "☕ Night Shift", distance: 2480, slaps: 7, note: "Survivor of the 3 AM CT queue. We survived!" }
    ];

    function getShiftRoster() {
      try {
        const stored = localStorage.getItem('pocketgull_shift_roster');
        return stored ? JSON.parse(stored) : DEFAULT_ROSTER;
      } catch (e) {
        return DEFAULT_ROSTER;
      }
    }

    function saveShiftRoster(roster) {
      try {
        localStorage.setItem('pocketgull_shift_roster', JSON.stringify(roster));
      } catch (e) {}
    }

    function renderShiftRoster() {
      const listEl = document.getElementById('roster-list');
      const roster = getShiftRoster();
      roster.sort((a, b) => b.distance - a.distance);

      listEl.innerHTML = roster.map((item, idx) => `
        <div class="roster-item">
          <div class="roster-left">
            <div class="roster-rank">#${idx + 1}</div>
            <div>
              <div class="roster-name">${item.name}</div>
              <div class="roster-role">${item.role}</div>
              ${item.note ? `<div class="roster-note">"${item.note}"</div>` : ''}
            </div>
          </div>
          <div class="roster-stat">
            <div class="roster-distance">${item.distance.toLocaleString()} m</div>
            <div class="roster-slaps">${item.slaps} knee-slaps</div>
          </div>
        </div>
      `).join('');
    }

    const rosterModal = document.getElementById('roster-modal');
    document.getElementById('btn-roster').addEventListener('click', () => {
      renderShiftRoster();
      rosterModal.classList.add('open');
    });
    document.getElementById('roster-close-btn').addEventListener('click', () => rosterModal.classList.remove('open'));

    document.getElementById('btn-save-roster').addEventListener('click', () => {
      const name = document.getElementById('input-roster-name').value.trim() || 'Anonymous Colleague';
      const role = document.getElementById('select-roster-role').value;
      const note = document.getElementById('input-roster-note').value.trim();

      const roster = getShiftRoster();
      roster.push({
        rank: roster.length + 1,
        name,
        role,
        distance: Math.max(distanceMeters, 120),
        slaps: kneeSlapsCount,
        note
      });
      saveShiftRoster(roster);
      renderShiftRoster();
      document.getElementById('input-roster-name').value = '';
      document.getElementById('input-roster-note').value = '';
    });

    const jokesModal = document.getElementById('jokes-modal');
    document.getElementById('btn-jokes').addEventListener('click', () => {
      const listEl = document.getElementById('jokes-book-list');
      const isRider = (heroMode === 'knee_rider');
      const jokes = isRider ? KNEE_RIDER_JOKES : PATTY_JOKES;
      const unlocked = isRider ? unlockedRiderJokes : unlockedPattyJokes;

      document.getElementById('jokes-modal-title').textContent = isRider
        ? "K.N.E.E. 2000 Tactical Joke Database"
        : "Patty's Knee-Slapper Joke Book";

      listEl.innerHTML = jokes.map((j, idx) => `
        <div class="roster-item" style="flex-direction: column; align-items: flex-start; gap: 8px;">
          <div style="display: flex; justify-content: space-between; width: 100%;">
            <span class="joke-badge ${isRider ? 'rider' : ''}">${j.category} #${idx + 1}</span>
            <span style="font-size: 11px; color: ${unlocked.has(idx) ? '#34d399' : '#64748b'};">
              ${unlocked.has(idx) ? '✅ Unlocked' : '🔒 Slap to Unlock'}
            </span>
          </div>
          <div style="font-size: 14px; font-weight: 600; color: #f1f5f9;">${j.setup}</div>
          <div style="font-size: 14px; font-weight: 700; color: ${isRider ? '#f87171' : '#34d399'};">👉 ${j.punchline}</div>
        </div>
      `).join('');
      jokesModal.classList.add('open');
    });
    document.getElementById('jokes-close-btn').addEventListener('click', () => jokesModal.classList.remove('open'));

    [rosterModal, jokesModal, questsModal, questBriefModal].forEach(modal => {
      modal.addEventListener('click', (e) => {
        if (e.target === modal) modal.classList.remove('open');
      });
    });

    const btnSound = document.getElementById('btn-sound');
    btnSound.addEventListener('click', () => {
      audio.enabled = !audio.enabled;
      btnSound.classList.toggle('active', audio.enabled);
      document.getElementById('sound-icon').textContent = audio.enabled ? '🔔' : '🔕';
      document.getElementById('sound-text').textContent = audio.enabled ? 'Sound: On' : 'Muted';
      if (!audio.enabled) {
        audio.stopSynthwaveBass();
      } else if (heroMode === 'knee_rider') {
        audio.startSynthwaveBass();
      }
    });

    const btnChrono = document.getElementById('btn-chrono');
    btnChrono.addEventListener('click', () => {
      currentChronoIndex = (currentChronoIndex + 1) % CIRCADIAN_MODES.length;
      const mode = CIRCADIAN_MODES[currentChronoIndex];
      document.getElementById('chrono-icon').textContent = mode.icon;
      document.getElementById('chrono-text').textContent = mode.name;
    });

    const btnMode = document.getElementById('btn-mode');
    btnMode.addEventListener('click', () => {
      isCinemaMode = !isCinemaMode;
      btnMode.classList.toggle('active', isCinemaMode);
      document.getElementById('mode-icon').textContent = isCinemaMode ? '🎬' : '🕹️';
      document.getElementById('mode-text').textContent = isCinemaMode ? 'Auto-Cinema' : 'Shift Play';
    });
    btnMode.classList.add('active');

    const btnPacing = document.getElementById('btn-pacing');
    btnPacing.addEventListener('click', () => {
      currentPacingIdx = (currentPacingIdx + 1) % PACING_MODES.length;
      const mode = PACING_MODES[currentPacingIdx];
      pacingMultiplier = mode.mult;
      document.getElementById('pacing-icon').textContent = mode.icon;
      document.getElementById('pacing-text').textContent = mode.name;
    });

    /* =========================================================================
       12. MAIN 60/120 FPS RENDERING & KINEMATICS LOOP
       ========================================================================= */
    let lastTime = performance.now();

    function renderLoop(now) {
      const dt = Math.min((now - lastTime) / 1000, 0.1);
      lastTime = now;

      // Poll Gamepad Controller
      gamepads.poll(
        (diveState) => { if (!isCinemaMode) isDiving = diveState; },
        () => { triggerKneeSlap(avatar.x, avatar.y); },
        () => { triggerTurboBoost(); }
      );

      const isRider = (heroMode === 'knee_rider');
      const pal = isRider ? SYNTHWAVE_PALETTE : CIRCADIAN_MODES[currentChronoIndex];

      // A. Sky Gradient
      const skyGrad = ctx.createLinearGradient(0, 0, 0, height);
      skyGrad.addColorStop(0, pal.skyTop);
      skyGrad.addColorStop(1, pal.skyBottom);
      ctx.fillStyle = skyGrad;
      ctx.fillRect(0, 0, width, height);

      // B. Celestial Sun / Synthwave Horizon
      if (isRider) {
        ctx.save();
        const sunX = width * 0.72;
        const sunY = height * 0.32;
        const sunR = 64;

        const sunGlow = ctx.createRadialGradient(sunX, sunY, sunR * 0.3, sunX, sunY, sunR * 2.2);
        sunGlow.addColorStop(0, '#f43f5e');
        sunGlow.addColorStop(0.5, '#ec4899');
        sunGlow.addColorStop(1, 'transparent');
        ctx.fillStyle = sunGlow;
        ctx.beginPath();
        ctx.arc(sunX, sunY, sunR * 2.2, 0, Math.PI * 2);
        ctx.fill();

        const sunDisc = ctx.createLinearGradient(sunX, sunY - sunR, sunX, sunY + sunR);
        sunDisc.addColorStop(0, '#fde047');
        sunDisc.addColorStop(0.6, '#f43f5e');
        sunDisc.addColorStop(1, '#7e22ce');
        ctx.fillStyle = sunDisc;
        ctx.beginPath();
        ctx.arc(sunX, sunY, sunR, 0, Math.PI * 2);
        ctx.fill();

        ctx.fillStyle = pal.skyBottom;
        for (let i = 0; i < 7; i++) {
          const cutY = sunY - 10 + i * 11;
          const cutH = 2 + i * 1.2;
          ctx.fillRect(sunX - sunR, cutY, sunR * 2, cutH);
        }
        ctx.restore();
      } else {
        ctx.save();
        ctx.fillStyle = pal.sunColor;
        ctx.shadowColor = pal.sunColor;
        ctx.shadowBlur = 40;
        ctx.beginPath();
        ctx.arc(width * 0.78, height * 0.22, 44, 0, Math.PI * 2);
        ctx.fill();
        ctx.restore();
      }

      // C. Terrain Measurement & Auto-Cinema
      const groundY = getForegroundHillY(avatar.x);
      const slope = getForegroundSlope(avatar.x);
      const dtClamped = Math.min(dt, 0.033);

      if (isCinemaMode) {
        if (slope > 0.06) {
          isDiving = true;
        } else if (slope < -0.04 || avatar.y < groundY - 140) {
          isDiving = false;
        }
        if (Date.now() - avatar.lastSlapTime > 35000) {
          triggerKneeSlap(avatar.x, avatar.y);
        }
      }

      // D. Physics (Paced, Serene, Floaty Kinematics)
      if (isFlowActive) {
        flowTimer -= dtClamped;
        if (flowTimer <= 0) {
          isFlowActive = false;
          flowCombo = 0;
        }
      }
      if (slapCooldownTimer > 0) {
        slapCooldownTimer -= dtClamped;
      }
      if (isShockwaveActive) {
        shockwaveRadius += 900 * dtClamped;
        if (shockwaveRadius > width * 1.5) {
          isShockwaveActive = false;
          shockwaveRadius = 0;
        }
      }
      const forwardSpeed = avatar.vx * pacingMultiplier;
      distanceMeters += Math.round(forwardSpeed * 0.4);
      statDist.textContent = `${distanceMeters.toLocaleString()} m`;

      const gravity = (isDiving ? 420 : 190) * pacingMultiplier;
      avatar.vy += gravity * dtClamped;

      if (avatar.y < 110) {
        avatar.vy += 450 * dtClamped;
      }

      avatar.y += avatar.vy * dtClamped;

      if (avatar.y + avatar.radius >= groundY) {
        avatar.y = groundY - avatar.radius;

        if (slope > 0.05) {
          // Perfect Downslope Landing!
          if (isDiving && lastLandType !== 'downslope') {
            flowCombo++;
            lastLandType = 'downslope';
            if (flowCombo >= 3) {
              isFlowActive = true;
              flowTimer = 6.0; // 6s duration
              audio.playLaunch(440);
              gamepads.rumble(90, 0.4, 0.2);
            }
          }
          const boostRate = (isDiving || isFlowActive) ? 4.6 : 1.6;
          const maxVx = isTurbo ? 4.8 : (isFlowActive ? 3.0 : (isDiving ? 2.6 : 2.0));
          avatar.vx = Math.min(avatar.vx + slope * boostRate * dtClamped, maxVx);
          avatar.expression = isDiving ? 'diving' : 'happy';
        } else {
          // Landed on upslope or flat - combo breaks
          if (slope < -0.06 && lastLandType !== 'upslope') {
            flowCombo = 0;
            isFlowActive = false;
            lastLandType = 'upslope';
          }

          if (avatar.vx > 1.35 && slope < -0.08) {
            avatar.vy = Math.max(slope * avatar.vx * 130 * (isRider ? 1.2 : 1.05), -290 * pacingMultiplier);
            audio.playLaunch(280 + Math.abs(slope) * 120);
            gamepads.rumble(60, 0.25, 0.15);
            avatar.expression = 'soaring';
            avatar.squashX = 0.85;
            avatar.squashY = 1.18;

            turboCharge = Math.min(100, turboCharge + 8);
            if (isRider) statSpecialVal.textContent = `${Math.round(turboCharge)}%`;
          }
          avatar.vx = Math.max(avatar.vx - 1.2 * dtClamped, 1.3);
        }
        avatar.angle = Math.atan2(slope, 1);
      } else {
        avatar.angle = Math.atan2(avatar.vy * 0.04, avatar.vx * 0.12);
        if (avatar.vy < -50) {
          avatar.expression = 'soaring';
        } else if (isDiving) {
          avatar.expression = 'diving';
        }
      }

      avatar.squashX += (1 - avatar.squashX) * 0.10;
      avatar.squashY += (1 - avatar.squashY) * 0.10;

      hills.forEach(h => {
        h.offset += forwardSpeed * h.speed;
      });

      // E. Draw Parallax Condylar Hills
      const hillColors = [pal.hill1, pal.hill2, pal.hill3];
      hills.forEach((h, lIdx) => {
        ctx.save();
        ctx.fillStyle = hillColors[lIdx];
        ctx.beginPath();
        ctx.moveTo(0, height);

        const step = 14;
        for (let x = 0; x <= width + step; x += step) {
          const worldX = x;
          const normalizedX = (worldX + h.offset);
          const yBase = height * h.baseY;
          const y = yBase +
                    Math.sin(normalizedX / (h.wavelength / (Math.PI * 2))) * (h.amplitude * 0.75) +
                    Math.cos(normalizedX / (h.wavelength * 0.45)) * (h.amplitude * 0.25);
          ctx.lineTo(x, y);
        }

        ctx.lineTo(width, height);
        ctx.closePath();
        ctx.fill();

        if (isRider && lIdx === 2) {
          ctx.strokeStyle = '#06b6d4';
          ctx.lineWidth = 2.5;
          ctx.shadowColor = '#06b6d4';
          ctx.shadowBlur = 12;
          ctx.stroke();

          ctx.beginPath();
          ctx.lineWidth = 1;
          ctx.strokeStyle = 'rgba(6, 182, 212, 0.4)';
          const gridSpacing = 48;
          const gridOffset = (h.offset % gridSpacing);
          for (let gx = -gridOffset; gx <= width + gridSpacing; gx += gridSpacing) {
            const gy = getForegroundHillY(gx);
            ctx.moveTo(gx, gy);
            ctx.lineTo(gx - 30, height);
          }
          ctx.stroke();
        }
        ctx.restore();
      });

      // F. Draw Collectibles & Quest Items
      collectibles.forEach((c) => {
        const screenX = c.x - hills[2].offset;

        if (screenX < -100) {
          c.x += width + 500 + Math.random() * 300;
          const groundBase = getForegroundHillY(c.x);
          const arcHeight = 60 + Math.sin((c.pulse || 0)) * 120;
          c.y = groundBase - arcHeight;
          c.collected = false;
        }

        if (screenX >= -50 && screenX <= width + 50 && !c.collected) {
          c.pulse += 0.05;
          const glow = Math.sin(c.pulse) * 4;

          // Magnetic Pull during Flow State (Radius 220px)
          if (isFlowActive && !c.collected) {
            const mDist = Math.hypot(screenX - avatar.x, c.y - avatar.y);
            if (mDist < 220) {
              const pullStrength = (1.0 - mDist / 220) * 12;
              c.x -= ((screenX - avatar.x) / mDist) * pullStrength;
              c.y -= ((c.y - avatar.y) / mDist) * pullStrength;
            }
          }

          // Knee Slap Shockwave Instant Collection
          if (isShockwaveActive && !c.collected) {
            const swDist = Math.hypot(screenX - avatar.x, c.y - avatar.y);
            if (swDist <= shockwaveRadius) {
              c.collected = true;
              audio.playChime(784);
              if (activeQuest) checkQuestCompletion();
            }
          }

          const dist = Math.hypot(screenX - avatar.x, c.y - avatar.y);
          if (dist < avatar.radius + c.radius) {
            c.collected = true;
            audio.playChime(c.type === 'synovial' ? 587 : 784);

            if (activeQuest) {
              checkQuestCompletion();
            } else if (isRider) {
              turboCharge = Math.min(100, turboCharge + 10);
              statSpecialVal.textContent = `${Math.round(turboCharge)}%`;
            } else {
              harmonyMeter = Math.min(harmonyMeter + 5, 100);
              statSpecialVal.textContent = `${harmonyMeter}%`;
            }

            for (let i = 0; i < 8; i++) {
              particles.push({
                x: screenX,
                y: c.y,
                vx: (Math.random() - 0.5) * 6,
                vy: (Math.random() - 0.5) * 6,
                life: 1.0,
                color: activeQuest ? activeQuest.itemColor : (isRider ? '#f43f5e' : (c.type === 'synovial' ? '#facc15' : '#38bdf8'))
              });
            }
          }

          ctx.save();
          ctx.beginPath();
          const itemColor = activeQuest ? activeQuest.itemColor : (isRider ? '#f43f5e' : (c.type === 'synovial' ? '#facc15' : '#38bdf8'));

          if (isRider || activeQuest?.tier === 'red') {
            ctx.fillStyle = itemColor;
            ctx.shadowColor = itemColor;
            ctx.shadowBlur = 18;
            ctx.moveTo(screenX, c.y - c.radius - glow);
            ctx.lineTo(screenX + c.radius + glow, c.y);
            ctx.lineTo(screenX, c.y + c.radius + glow);
            ctx.lineTo(screenX - c.radius - glow, c.y);
            ctx.closePath();
            ctx.fill();
          } else {
            ctx.arc(screenX, c.y, c.radius + glow, 0, Math.PI * 2);
            ctx.fillStyle = itemColor;
            ctx.shadowColor = ctx.fillStyle;
            ctx.shadowBlur = 16;
            ctx.fill();

            ctx.beginPath();
            ctx.arc(screenX - 3, c.y - 3, c.radius * 0.4, 0, Math.PI * 2);
            ctx.fillStyle = '#ffffff';
            ctx.fill();
          }
          ctx.restore();
        }
      });

      // G. Particles & Confetti
      for (let i = particles.length - 1; i >= 0; i--) {
        const p = particles[i];
        p.x += p.vx;
        p.y += p.vy;
        p.life -= dt * 1.5;
        if (p.life <= 0) {
          particles.splice(i, 1);
          continue;
        }
        ctx.save();
        ctx.globalAlpha = p.life;
        ctx.fillStyle = p.color;
        ctx.beginPath();
        ctx.arc(p.x, p.y, 3, 0, Math.PI * 2);
        ctx.fill();
        ctx.restore();
      }

      for (let i = confetti.length - 1; i >= 0; i--) {
        const cf = confetti[i];
        cf.x += cf.vx;
        cf.y += cf.vy;
        cf.vy += 12 * dt;
        cf.rotation += cf.vRot;
        cf.life -= dt * 0.6;
        if (cf.life <= 0) {
          confetti.splice(i, 1);
          continue;
        }
        ctx.save();
        ctx.globalAlpha = cf.life;
        ctx.translate(cf.x, cf.y);
        ctx.rotate(cf.rotation);
        ctx.fillStyle = cf.color;
        ctx.fillRect(-cf.size / 2, -cf.size / 2, cf.size, cf.size * 0.6);
        ctx.restore();
      }

      // H. Draw Avatar
      ctx.save();
      ctx.translate(avatar.x, avatar.y);
      ctx.rotate(avatar.angle);
      ctx.scale(avatar.squashX, avatar.squashY);

      if (isRider) {
        // KNEE RIDER 2000
        ctx.save();
        const jetWave = Math.sin(now * 0.05) * 4;
        const thrusterLen = isTurbo ? 64 : 26;
        const thrusterGrad = ctx.createLinearGradient(-avatar.radius, 0, -avatar.radius - thrusterLen, 0);
        thrusterGrad.addColorStop(0, '#ef4444');
        thrusterGrad.addColorStop(0.5, '#f43f5e');
        thrusterGrad.addColorStop(1, 'transparent');
        ctx.fillStyle = thrusterGrad;

        ctx.beginPath();
        ctx.moveTo(-avatar.radius * 0.7, -8);
        ctx.lineTo(-avatar.radius - thrusterLen, -8 + jetWave);
        ctx.lineTo(-avatar.radius * 0.7, -4);
        ctx.closePath();
        ctx.fill();

        ctx.beginPath();
        ctx.moveTo(-avatar.radius * 0.7, 4);
        ctx.lineTo(-avatar.radius - thrusterLen, 4 - jetWave);
        ctx.lineTo(-avatar.radius * 0.7, 8);
        ctx.closePath();
        ctx.fill();
        ctx.restore();

        const chassisGrad = ctx.createLinearGradient(-avatar.radius, -avatar.radius, avatar.radius, avatar.radius);
        chassisGrad.addColorStop(0, '#27272a');
        chassisGrad.addColorStop(0.3, '#09090b');
        chassisGrad.addColorStop(0.8, '#18181b');
        chassisGrad.addColorStop(1, '#09090b');

        ctx.beginPath();
        ctx.arc(0, 0, avatar.radius, 0, Math.PI * 2);
        ctx.fillStyle = chassisGrad;
        ctx.shadowColor = isTurbo ? '#ef4444' : 'rgba(0,0,0,0.8)';
        ctx.shadowBlur = isTurbo ? 32 : 16;
        ctx.fill();
        ctx.lineWidth = 2;
        ctx.strokeStyle = '#ef4444';
        ctx.stroke();

        ctx.beginPath();
        ctx.arc(0, -6, avatar.radius * 0.75, Math.PI * 1.1, Math.PI * 1.9);
        ctx.lineWidth = 1.5;
        ctx.strokeStyle = 'rgba(255, 255, 255, 0.45)';
        ctx.stroke();

        // Oscillating Red Scanner
        avatar.scannerPos += avatar.scannerDir * dt * 4.2;
        if (avatar.scannerPos > 1) { avatar.scannerPos = 1; avatar.scannerDir = -1; }
        else if (avatar.scannerPos < -1) { avatar.scannerPos = -1; avatar.scannerDir = 1; }

        const scannerWidth = 26;
        const scannerHeight = 6;
        const scannerX = 2;
        const scannerY = 0;

        ctx.fillStyle = '#000000';
        ctx.fillRect(scannerX, scannerY - scannerHeight / 2, scannerWidth, scannerHeight);
        ctx.strokeStyle = '#3f3f46';
        ctx.lineWidth = 1;
        ctx.strokeRect(scannerX, scannerY - scannerHeight / 2, scannerWidth, scannerHeight);

        const numLeds = 8;
        const targetLed = Math.round(((avatar.scannerPos + 1) / 2) * (numLeds - 1));
        const ledW = scannerWidth / numLeds;

        for (let i = 0; i < numLeds; i++) {
          const distToBeam = Math.abs(i - targetLed);
          let intensity = Math.max(0, 1 - distToBeam * 0.45);
          ctx.fillStyle = `rgba(239, 68, 68, ${0.15 + intensity * 0.85})`;
          if (distToBeam === 0) { ctx.shadowColor = '#ef4444'; ctx.shadowBlur = 12; }
          else ctx.shadowBlur = 0;
          ctx.fillRect(scannerX + i * ledW + 0.5, scannerY - scannerHeight / 2 + 0.5, ledW - 1, scannerHeight - 1);
        }

        ctx.font = 'bold 8px monospace';
        ctx.fillStyle = '#f87171';
        ctx.textAlign = 'center';
        ctx.shadowBlur = 0;
        ctx.fillText('K.N.E.E.', 0, -12);

      } else {
        // GREEN HAPPY KNEE (Patty)
        ctx.save();
        ctx.fillStyle = 'rgba(52, 211, 153, 0.45)';
        ctx.beginPath();
        ctx.moveTo(-avatar.radius * 0.8, -4);
        const capeWave = Math.sin(now * 0.01) * 8;
        ctx.quadraticCurveTo(-avatar.radius * 2.2, capeWave, -avatar.radius * 3.2, capeWave * 1.4);
        ctx.lineTo(-avatar.radius * 3.2, capeWave * 1.4 + 10);
        ctx.quadraticCurveTo(-avatar.radius * 2.0, capeWave + 6, -avatar.radius * 0.8, 6);
        ctx.closePath();
        ctx.fill();
        ctx.restore();

        const bodyGrad = ctx.createRadialGradient(-6, -8, 4, 0, 0, avatar.radius);
        bodyGrad.addColorStop(0, '#6ee7b7');
        bodyGrad.addColorStop(0.4, '#10b981');
        bodyGrad.addColorStop(1, '#047857');

        ctx.beginPath();
        ctx.arc(0, 0, avatar.radius, 0, Math.PI * 2);
        ctx.fillStyle = bodyGrad;
        ctx.shadowColor = 'rgba(16, 185, 129, 0.5)';
        ctx.shadowBlur = 18;
        ctx.fill();
        ctx.lineWidth = 2.5;
        ctx.strokeStyle = '#ffffff';
        ctx.stroke();

        if (avatar.blushing) {
          ctx.fillStyle = 'rgba(251, 113, 133, 0.5)';
          ctx.beginPath();
          ctx.arc(-11, 5, 5, 0, Math.PI * 2);
          ctx.arc(11, 5, 5, 0, Math.PI * 2);
          ctx.fill();
        }

        ctx.fillStyle = '#0f172a';
        if (avatar.expression === 'soaring') {
          ctx.lineWidth = 2.5;
          ctx.strokeStyle = '#0f172a';
          ctx.beginPath();
          ctx.arc(-8, -2, 5, Math.PI * 1.1, Math.PI * 1.9);
          ctx.stroke();
          ctx.beginPath();
          ctx.arc(8, -2, 5, Math.PI * 1.1, Math.PI * 1.9);
          ctx.stroke();
        } else if (avatar.expression === 'diving') {
          ctx.lineWidth = 2.5;
          ctx.strokeStyle = '#0f172a';
          ctx.beginPath();
          ctx.moveTo(-12, -4); ctx.lineTo(-6, -1); ctx.lineTo(-12, 2);
          ctx.stroke();
          ctx.beginPath();
          ctx.moveTo(12, -4); ctx.lineTo(6, -1); ctx.lineTo(12, 2);
          ctx.stroke();
        } else {
          ctx.beginPath();
          ctx.arc(-8, -2, 3.5, 0, Math.PI * 2);
          ctx.arc(8, -2, 3.5, 0, Math.PI * 2);
          ctx.fill();

          ctx.fillStyle = '#ffffff';
          ctx.beginPath();
          ctx.arc(-9, -3.5, 1.2, 0, Math.PI * 2);
          ctx.arc(7, -3.5, 1.2, 0, Math.PI * 2);
          ctx.fill();
        }

        ctx.beginPath();
        ctx.lineWidth = 2;
        ctx.strokeStyle = '#0f172a';
        ctx.arc(0, 4, 6, 0.15, Math.PI - 0.15);
        ctx.stroke();
      }

      ctx.restore();

      requestAnimationFrame(renderLoop);
    }

    requestAnimationFrame(renderLoop);

    setTimeout(() => {
      triggerKneeSlap(avatar.x, avatar.y);
    }, 1800);
  </script>
</body>
</html>
"""

display(HTML(f'''
<div style="width: 100%; height: 720px; border-radius: 24px; overflow: hidden; border: 1px solid #27272a; box-shadow: 0 20px 40px rgba(0,0,0,0.8);">
  <iframe srcdoc="{html_cinema_content.replace('"', '&quot;')}" style="width: 100%; height: 100%; border: none;"></iframe>
</div>
'''))
```
